همچنین یادگیری واژگان برنامه‌نویسی را شروع می‌کنید؛ اصطلاحاتی مثل عملگر (operator)، عبارت (expression)، مقدار (value) و نوع (type). این واژگان مهم‌اند: برای فهمیدن بقیه‌ی کتاب، برای گفتگو با برنامه‌نویس‌های دیگر و برای استفاده از دستیارهای مجازی و فهمیدن حرف‌هایشان به آن‌ها نیاز خواهید داشت.

## ۱.۱. عملگرهای حسابی

**عملگر حسابی** (arithmetic operator) نمادی است که یک محاسبه‌ی حسابی را نشان می‌دهد. برای مثال، علامت جمع، <code dir="ltr">+</code>، عمل جمع را انجام می‌دهد.

</div>

```python
30 + 12
```

<div dir="rtl" style="text-align: right;">

علامت منها، <code dir="ltr">-</code>، عملگری است که تفریق را انجام می‌دهد.

</div>

```python
43 - 1
```

<div dir="rtl" style="text-align: right;">

ستاره، <code dir="ltr">*</code>، ضرب را انجام می‌دهد.

</div>

```python
6 * 7
```

<div dir="rtl" style="text-align: right;">

و اسلش، <code dir="ltr">/</code>، تقسیم را انجام می‌دهد:

</div>

```python
84 / 2
```

<div dir="rtl" style="text-align: right;">

دقت کنید که نتیجه‌ی تقسیم <code dir="ltr">42.0</code> است، نه <code dir="ltr">42</code>. دلیلش این است که در پایتون دو نوع عدد وجود دارد:

- **عدد صحیح** (integer): اعدادی که بخش کسری یا اعشاری ندارند، و

- **عدد اعشاری** (floating-point): اعداد صحیح و اعدادی که ممیز (نقطه‌ی اعشار) دارند.

اگر دو عدد صحیح را جمع، تفریق یا ضرب کنید، نتیجه عدد صحیح است. اما اگر دو عدد صحیح را بر هم تقسیم کنید، نتیجه عدد اعشاری است. پایتون عملگر دیگری هم دارد، <code dir="ltr">//</code>، که **تقسیم صحیح** (integer division) را انجام می‌دهد. نتیجه‌ی تقسیم صحیح همیشه یک عدد صحیح است.

</div>

```python
84 // 2
```

<div dir="rtl" style="text-align: right;">

تقسیم صحیح را «تقسیم کفی» (floor division) هم می‌گویند، چون همیشه به پایین (به سمت «کف») گرد می‌کند.

</div>

```python
85 // 2
```

<div dir="rtl" style="text-align: right;">

سرانجام، عملگر <code dir="ltr">**</code> توان‌رسانی را انجام می‌دهد؛ یعنی یک عدد را به یک توان می‌رساند:

</div>

```python
7 ** 2
```

<div dir="rtl" style="text-align: right;">

در بعضی زبان‌های دیگر برای توان‌رسانی از علامت <code dir="ltr">^</code> (کِرت) استفاده می‌شود، اما در پایتون این علامت یک عملگر بیتی به نام XOR است. اگر با عملگرهای بیتی آشنا نباشید، نتیجه ممکن است غیرمنتظره باشد:

</div>

```python
7 ^ 2
```

<div dir="rtl" style="text-align: right;">

در این کتاب به عملگرهای بیتی نمی‌پردازم، اما می‌توانید درباره‌ی آن‌ها در <http://wiki.python.org/moin/BitwiseOperators> بخوانید.

</div>

<div dir="rtl" style="text-align: right;">

## ۱.۲. عبارت‌ها

مجموعه‌ای از عملگرها و اعداد را **عبارت** (expression) می‌نامند. یک عبارت می‌تواند هر تعداد عملگر و عدد داشته باشد. برای مثال، این یک عبارت با دو عملگر است.

</div>

```python
6 + 6 ** 2
```

<div dir="rtl" style="text-align: right;">

دقت کنید که توان‌رسانی قبل از جمع انجام می‌شود. پایتون از ترتیب انجام عملیات (order of operations) پیروی می‌کند که احتمالاً در درس ریاضی یاد گرفته‌اید: توان‌رسانی قبل از ضرب و تقسیم انجام می‌شود و ضرب و تقسیم قبل از جمع و تفریق.

در مثال زیر، ضرب قبل از جمع انجام می‌شود.

</div>

```python
12 + 5 * 6
```

<div dir="rtl" style="text-align: right;">

اگر می‌خواهید جمع اول انجام شود، می‌توانید از پرانتز استفاده کنید.

</div>

```python
(12 + 5) * 6
```

<div dir="rtl" style="text-align: right;">

هر عبارت یک **مقدار** (value) دارد. برای مثال، عبارت <code dir="ltr">6 * 7</code> مقدارش <code dir="ltr">42</code> است.

</div>

<div dir="rtl" style="text-align: right;">

## ۱.۳. توابع حسابی

علاوه بر عملگرهای حسابی، پایتون چند **تابع** (function) هم دارد که روی اعداد کار می‌کنند. برای مثال، تابع <code dir="ltr">round</code> یک عدد اعشاری می‌گیرد و آن را به نزدیک‌ترین عدد صحیح گرد می‌کند.

</div>

```python
round(42.4)
```

```python
round(42.6)
```

<div dir="rtl" style="text-align: right;">

تابع <code dir="ltr">abs</code> قدر مطلق (absolute value) یک عدد را حساب می‌کند. برای عدد مثبت، قدر مطلق خود عدد است.

</div>

```python
abs(42)
```

<div dir="rtl" style="text-align: right;">

برای عدد منفی، قدر مطلق مثبت است.

</div>

```python
abs(-42)
```

<div dir="rtl" style="text-align: right;">

وقتی این‌طور از یک تابع استفاده می‌کنیم، می‌گوییم تابع را **فراخوانی** (call) می‌کنیم. عبارتی که یک تابع را فراخوانی می‌کند، یک **فراخوانی تابع** (function call) است.

هنگام فراخوانی تابع، پرانتزها الزامی‌اند. اگر آن‌ها را حذف کنید، پیام خطا می‌گیرید.

**نکته:** سلول بعدی از <code dir="ltr">%%expect</code> استفاده می‌کند، که یک «دستور جادویی» (magic command) در Jupyter است و یعنی انتظار داریم کد این سلول خطا ایجاد کند. برای اطلاعات بیشتر، [مقدمه‌ای بر نوت‌بوک Jupyter](https://colab.research.google.com/github/Alireza-Akhavan/ThinkPython/blob/main/chapters/jupyter_intro.ipynb) (به انگلیسی) را ببینید.

</div>

```python
%%expect SyntaxError

abs 42
```

<div dir="rtl" style="text-align: right;">

می‌توانید خط اول این پیام را نادیده بگیرید؛ اطلاعاتی ندارد که فعلاً برای فهمیدن آن نیاز داشته باشیم. خط دوم، کدی است که خطا دارد و زیرش یک علامت <code dir="ltr">^</code> گذاشته شده تا نشان دهد خطا کجا کشف شده است.

خط آخر نشان می‌دهد که این یک **خطای نحوی** (syntax error) است؛ یعنی ساختار عبارت ایراد دارد. در این مثال، مشکل این است که فراخوانی تابع به پرانتز نیاز دارد.

ببینیم اگر پرانتزها *و* مقدار را حذف کنیم چه می‌شود.

</div>

```python
abs
```

<div dir="rtl" style="text-align: right;">

نام یک تابع، به‌تنهایی، یک عبارت معتبر است و مقدار دارد. وقتی این مقدار نمایش داده می‌شود، نشان می‌دهد که <code dir="ltr">abs</code> یک تابع است و اطلاعات اضافه‌ای هم دارد که بعداً توضیح می‌دهم.

</div>

<div dir="rtl" style="text-align: right;">

## ۱.۴. رشته‌ها

علاوه بر اعداد، پایتون می‌تواند دنباله‌ای از حروف را هم نمایش دهد. به این دنباله‌ها **رشته** (string) می‌گویند، چون حروف مثل مهره‌های یک گردنبند به هم «رشته» شده‌اند. برای نوشتن یک رشته، دنباله‌ای از حروف را داخل نقل‌قول‌های ساده (straight quotes) قرار می‌دهیم.

</div>

```python
'Hello'
```

<div dir="rtl" style="text-align: right;">

استفاده از نقل‌قول دوتایی هم مجاز است.

</div>

```python
"world"
```

<div dir="rtl" style="text-align: right;">

نقل‌قول دوتایی نوشتن رشته‌ای را که آپوستروف دارد آسان می‌کند؛ آپوستروف همان نماد نقل‌قول ساده است.

</div>

```python
"it's a small "
```

<div dir="rtl" style="text-align: right;">

رشته‌ها می‌توانند فاصله، علائم نگارشی و ارقام هم داشته باشند.

</div>

```python
'Well, '
```

<div dir="rtl" style="text-align: right;">

عملگر <code dir="ltr">+</code> روی رشته‌ها هم کار می‌کند؛ دو رشته را به هم می‌چسباند و یک رشته‌ی واحد می‌سازد. به این کار **الحاق** (concatenation) می‌گویند.

</div>

```python
'Well, ' + "it's a small " + 'world.'
```

<div dir="rtl" style="text-align: right;">

عملگر <code dir="ltr">*</code> هم روی رشته‌ها کار می‌کند؛ چند نسخه از یک رشته می‌سازد و آن‌ها را به هم الحاق می‌کند.

</div>

```python
'Spam, ' * 4
```

<div dir="rtl" style="text-align: right;">

بقیه‌ی عملگرهای حسابی روی رشته‌ها کار نمی‌کنند.

پایتون تابعی به نام <code dir="ltr">len</code> دارد که طول یک رشته را حساب می‌کند.

</div>

```python
len('Spam')
```

<div dir="rtl" style="text-align: right;">

دقت کنید که <code dir="ltr">len</code> حروف بین نقل‌قول‌ها را می‌شمارد، نه خود نقل‌قول‌ها را.

هنگام ساختن رشته حتماً از نقل‌قول ساده استفاده کنید. نقل‌قول برعکس (back quote)، که به آن بک‌تیک (backtick) هم می‌گویند، خطای نحوی ایجاد می‌کند.

</div>

```python
%%expect SyntaxError

`Hello`
```

<div dir="rtl" style="text-align: right;">

نقل‌قول‌های هوشمند (smart quotes)، که نقل‌قول‌های خمیده (curly quotes) هم نامیده می‌شوند، نیز مجاز نیستند.

</div>

```python
%%expect SyntaxError

‘Hello’
```

## ۱.۵. مقدار و نوع

تا اینجا سه نوع مقدار دیده‌ایم:

- <code dir="ltr">2</code> یک عدد صحیح است،

- <code dir="ltr">42.0</code> یک عدد اعشاری است، و

- <code dir="ltr">'Hello'</code> یک رشته است.

هر دسته از مقدارها را یک **نوع** (type) می‌نامند. هر مقدار یک نوع دارد؛ یا گاهی می‌گوییم مقدار «به یک نوع تعلق دارد».

پایتون تابعی به نام <code dir="ltr">type</code> دارد که نوع هر مقدار را به شما می‌گوید. نوع عدد صحیح <code dir="ltr">int</code> است.

</div>

```python
type(2)
```

<div dir="rtl" style="text-align: right;">

نوع عدد اعشاری <code dir="ltr">float</code> است.

</div>

```python
type(42.0)
```

<div dir="rtl" style="text-align: right;">

و نوع رشته <code dir="ltr">str</code> است.

</div>

```python
type('Hello, World!')
```

<div dir="rtl" style="text-align: right;">

نوع‌های <code dir="ltr">int</code>، <code dir="ltr">float</code> و <code dir="ltr">str</code> را می‌توان به‌عنوان تابع هم به کار برد. برای مثال، <code dir="ltr">int</code> می‌تواند یک عدد اعشاری را بگیرد و به عدد صحیح تبدیل کند (همیشه به پایین گرد می‌کند).

</div>

```python
int(42.9)
```

<div dir="rtl" style="text-align: right;">

و <code dir="ltr">float</code> می‌تواند یک عدد صحیح را به مقدار اعشاری تبدیل کند.

</div>

```python
float(42)
```

<div dir="rtl" style="text-align: right;">

حالا به چیزی می‌رسیم که ممکن است گیج‌کننده باشد. اگر دنباله‌ای از ارقام را داخل نقل‌قول بگذارید، چه می‌شود؟

</div>

```python
'126'
```

<div dir="rtl" style="text-align: right;">

شبیه عدد است، اما در واقع یک رشته است.

</div>

```python
type('126')
```

<div dir="rtl" style="text-align: right;">

اگر سعی کنید مثل عدد از آن استفاده کنید، ممکن است خطا بگیرید.

</div>

```python
%%expect TypeError

'126' / 3
```

<div dir="rtl" style="text-align: right;">

این مثال یک <code dir="ltr">TypeError</code> تولید می‌کند؛ یعنی مقدارهای موجود در عبارت، که **عملوند** (operand) نامیده می‌شوند، نوع نادرستی دارند. پیام خطا می‌گوید عملگر <code dir="ltr">/</code> از نوع این مقدارها، یعنی <code dir="ltr">str</code> و <code dir="ltr">int</code>، پشتیبانی نمی‌کند.

اگر رشته‌ای دارید که ارقام دارد، می‌توانید با <code dir="ltr">int</code> آن را به عدد صحیح تبدیل کنید.

</div>

```python
int('126') / 3
```

<div dir="rtl" style="text-align: right;">

اگر رشته‌ای دارید که ارقام و ممیز (نقطه‌ی اعشار) دارد، می‌توانید با <code dir="ltr">float</code> آن را به عدد اعشاری تبدیل کنید.

</div>

```python
float('12.6')
```

<div dir="rtl" style="text-align: right;">

وقتی عدد صحیح بزرگی می‌نویسید، ممکن است وسوسه شوید بین گروه‌های رقم‌ها ویرگول بگذارید، مثل <code dir="ltr">1,000,000</code>. این در پایتون عبارتی معتبر است، اما نتیجه‌اش عدد صحیح نیست.

</div>

```python
1,000,000
```

<div dir="rtl" style="text-align: right;">

پایتون <code dir="ltr">1,000,000</code> را دنباله‌ای از اعداد صحیح جداشده با ویرگول تفسیر می‌کند. بعداً درباره‌ی این نوع دنباله بیشتر یاد می‌گیریم.

برای خواناتر شدن اعداد بزرگ می‌توانید از زیرخط (underscore) استفاده کنید.

</div>

```python
1_000_000
```

## ۱.۶. زبان‌های صوری و طبیعی

**زبان‌های طبیعی** (natural languages) زبان‌هایی هستند که مردم با آن‌ها حرف می‌زنند، مثل انگلیسی، اسپانیایی و فرانسوی. این زبان‌ها را کسی طراحی نکرده است؛ به‌طور طبیعی تکامل یافته‌اند.

**زبان‌های صوری** (formal languages) زبان‌هایی هستند که انسان‌ها برای کاربردهای مشخص طراحی کرده‌اند. برای مثال، نمادگذاری‌ای که ریاضی‌دان‌ها به کار می‌برند یک زبان صوری است که در نشان دادن رابطه‌ی میان اعداد و نمادها بسیار خوب عمل می‌کند. به همین ترتیب، زبان‌های برنامه‌نویسی زبان‌های صوری‌ای هستند که برای بیان محاسبات طراحی شده‌اند.

هرچند زبان‌های صوری و طبیعی ویژگی‌های مشترکی دارند، تفاوت‌های مهمی هم میانشان هست:

- ابهام: زبان‌های طبیعی پر از ابهام‌اند و آدم‌ها با کمک سرنخ‌های بافتی (contextual clues) و اطلاعات دیگر با آن کنار می‌آیند. زبان‌های صوری طوری طراحی شده‌اند که تقریباً یا کاملاً بدون ابهام باشند؛ یعنی هر برنامه دقیقاً یک معنی دارد، صرف‌نظر از بافت.

- افزونگی (redundancy): برای جبران ابهام و کم کردن سوءتفاهم، زبان‌های طبیعی از افزونگی استفاده می‌کنند و در نتیجه اغلب پرگو هستند. زبان‌های صوری افزونگی کمتری دارند و فشرده‌ترند.

- تحت‌اللفظی بودن: زبان‌های طبیعی پر از اصطلاح و استعاره‌اند. زبان‌های صوری دقیقاً همان را می‌گویند که نوشته شده است.

چون همه‌ی ما با زبان‌های طبیعی بزرگ می‌شویم، گاهی عادت کردن به زبان‌های صوری سخت است. زبان‌های صوری از زبان‌های طبیعی چگال‌ترند، پس خواندنشان بیشتر طول می‌کشد. ساختار هم مهم است، بنابراین همیشه بهترین کار این نیست که از بالا به پایین و از چپ به راست بخوانید. و سرانجام، جزئیات اهمیت دارند. خطاهای کوچک در املا و نقطه‌گذاری، که در زبان‌های طبیعی می‌توان از کنارشان گذشت، در زبان صوری می‌توانند تفاوت بزرگی ایجاد کنند.

## ۱.۷. اشکال‌زدایی

برنامه‌نویس‌ها اشتباه می‌کنند. به دلایلی شوخ‌طبعانه، خطاهای برنامه‌نویسی را **باگ** (bug) می‌نامند و فرایند ردیابی آن‌ها را **اشکال‌زدایی** (debugging).

برنامه‌نویسی، و به‌ویژه اشکال‌زدایی، گاهی احساسات شدیدی را برمی‌انگیزد. اگر با یک باگ سخت دست‌وپنجه نرم می‌کنید، ممکن است احساس خشم، غم یا شرمندگی کنید.

آماده شدن برای این واکنش‌ها ممکن است به کنار آمدن با آن‌ها کمک کند. یک راه این است که کامپیوتر را مثل کارمندی تصور کنید که نقاط قوتی دارد، مثل سرعت و دقت، و نقاط ضعف مشخصی، مثل نداشتن همدلی و ناتوانی در درک تصویر کلی.

کار شما این است که مدیر خوبی باشید: راه‌هایی پیدا کنید که از نقاط قوت استفاده کنید و نقاط ضعف را کم‌اثر کنید. و راه‌هایی پیدا کنید که از احساساتتان برای درگیر شدن با مسئله استفاده کنید، بدون اینکه اجازه دهید واکنش‌هایتان در توانایی شما برای کار مؤثر اختلال ایجاد کند.

یاد گرفتن اشکال‌زدایی می‌تواند ناامیدکننده باشد، اما مهارتی ارزشمند است که در بسیاری از فعالیت‌ها، فراتر از برنامه‌نویسی، به کار می‌آید. در پایان هر فصل بخشی مثل همین هست که پیشنهادهای من برای اشکال‌زدایی را در بر دارد. امیدوارم کمک‌تان کند!

## ۱.۸. واژه‌نامه

**عملگر حسابی (arithmetic operator):** نمادی مثل <code dir="ltr">+</code> و <code dir="ltr">*</code> که یک عمل حسابی مثل جمع یا ضرب را نشان می‌دهد.

**عدد صحیح (integer):** نوعی که اعداد بدون بخش کسری یا اعشاری را نمایش می‌دهد.

**عدد اعشاری (floating-point):** نوعی که اعداد صحیح و اعداد دارای بخش اعشاری را نمایش می‌دهد.

**تقسیم صحیح (integer division):** عملگری، <code dir="ltr">//</code>، که دو عدد را بر هم تقسیم می‌کند و به پایین، تا یک عدد صحیح، گرد می‌کند.

**عبارت (expression):** ترکیبی از متغیرها، مقدارها و عملگرها.

**مقدار (value):** یک عدد صحیح، یک عدد اعشاری یا یک رشته – یا یکی از انواع دیگر مقدارها که بعداً خواهیم دید.

**تابع (function):** دنباله‌ای نام‌دار از دستورها که عملی مفید انجام می‌دهد. توابع ممکن است آرگومان بگیرند یا نگیرند و ممکن است نتیجه تولید کنند یا نکنند.

**فراخوانی تابع (function call):** عبارتی – یا بخشی از یک عبارت – که یک تابع را اجرا می‌کند. از نام تابع و به دنبال آن فهرست آرگومان‌ها داخل پرانتز تشکیل می‌شود.

**خطای نحوی (syntax error):** خطایی در برنامه که باعث می‌شود تجزیه (parse) آن ناممکن باشد – و در نتیجه اجرای آن هم ناممکن است.

**رشته (string):** نوعی که دنباله‌ای از نویسه‌ها (characters) را نمایش می‌دهد.

**الحاق (concatenation):** چسباندن دو رشته به‌دنبال هم.

**نوع (type):** یک دسته از مقدارها. نوع‌هایی که تا اینجا دیده‌ایم عبارت‌اند از اعداد صحیح (نوع <code dir="ltr">int</code>)، اعداد اعشاری (نوع <code dir="ltr">float</code>) و رشته‌ها (نوع <code dir="ltr">str</code>).

**عملوند (operand):** یکی از مقدارهایی که یک عملگر روی آن‌ها عمل می‌کند.

**زبان طبیعی (natural language):** هر یک از زبان‌هایی که مردم با آن‌ها صحبت می‌کنند و به‌طور طبیعی تکامل یافته‌اند.

**زبان صوری (formal language):** هر یک از زبان‌هایی که انسان‌ها برای مقاصد مشخصی طراحی کرده‌اند، مثل نمایش ایده‌های ریاضی یا برنامه‌های کامپیوتری. همه‌ی زبان‌های برنامه‌نویسی زبان صوری‌اند.

**باگ (bug):** خطا در یک برنامه.

**اشکال‌زدایی (debugging):** فرایند پیدا کردن و رفع خطاها.

### ۱.۹.۱. از یک دستیار مجازی بپرسید

در طول این کتاب چند راه برای استفاده از دستیار مجازی یا چت‌بات برای یادگیری وجود دارد.

- اگر می‌خواهید درباره‌ی موضوعی از فصل بیشتر بدانید، یا چیزی روشن نیست، می‌توانید از او توضیح بخواهید.

- اگر در حل یکی از تمرین‌ها مشکل دارید، می‌توانید از او کمک بخواهید.

در هر فصل تمرین‌هایی پیشنهاد می‌کنم که می‌توانید با دستیار مجازی انجام دهید، اما تشویقتان می‌کنم اول خودتان چیزها را امتحان کنید و ببینید چه چیزی برایتان جواب می‌دهد.

چند موضوع که می‌توانید از دستیار مجازی درباره‌شان بپرسید:

- پیش‌تر از عملگرهای بیتی گفتم، اما توضیح ندادم چرا مقدار <code dir="ltr">7 ^ 2</code> برابر ۵ است. بپرسید «عملگرهای بیتی در پایتون چه هستند؟» (What are the bitwise operators in Python?) یا «مقدار <code dir="ltr">7 XOR 2</code> چیست؟» (What is the value of <code dir="ltr">7 XOR 2</code>?)

- از ترتیب انجام عملیات هم حرف زدم. برای جزئیات بیشتر بپرسید «ترتیب انجام عملیات در پایتون چیست؟» (What is the order of operations in Python?)

- تابع <code dir="ltr">round</code>، که برای گرد کردن عدد اعشاری به نزدیک‌ترین عدد صحیح به کار بردیم، می‌تواند آرگومان دوم هم بگیرد. بپرسید «آرگومان‌های تابع round چیست؟» (What are the arguments of the round function?) یا «چطور عدد پی را تا سه رقم اعشار گرد کنم؟» (How do I round pi off to three decimal places?)

- یک عملگر حسابی دیگر هم هست که اسمش را نبردم؛ بپرسید «عملگر مدول (modulus) در پایتون چیست؟» (What is the modulus operator in Python?)

بیشتر دستیارهای مجازی پایتون را می‌شناسند و به چنین پرسش‌هایی کم‌وبیش قابل‌اعتماد پاسخ می‌دهند. اما یادتان باشد که این ابزارها اشتباه می‌کنند. اگر از چت‌بات کدی گرفتید، آن را آزمایش کنید!
