## ۱۰.۱. دیکشنری یک نگاشت است

**دیکشنری** شبیه فهرست است، اما کلی‌تر. در فهرست، اندیس‌ها باید عدد صحیح باشند؛ در دیکشنری می‌توانند (تقریباً) از هر نوعی باشند. برای مثال، فرض کنید فهرستی از کلمه‌های عددی بسازیم، مثل این.

</div>

```python
lst = ['zero', 'one', 'two']
```

<div dir="rtl" style="text-align: right;">

می‌توانیم از یک عدد صحیح به‌عنوان اندیس برای گرفتن کلمه‌ی متناظر استفاده کنیم.

</div>

```python
lst[1]
```

<div dir="rtl" style="text-align: right;">

اما فرض کنید می‌خواهیم برعکس عمل کنیم و یک کلمه را جست‌وجو کنیم تا عدد صحیح متناظر را بگیریم. با فهرست نمی‌توانیم این کار را بکنیم، اما با دیکشنری می‌توانیم. با ساختن یک دیکشنری تهی و نسبت دادن آن به <code dir="ltr">numbers</code> شروع می‌کنیم.

راه دیگر برای ساختن دیکشنری استفاده از تابع <code dir="ltr">dict</code> است. می‌توانیم دیکشنری تهی را این‌طور بسازیم.

## ۱۰.۳. عملگر <code dir="ltr">in</code>

عملگر <code dir="ltr">in</code> روی دیکشنری‌ها هم کار می‌کند؛ به شما می‌گوید آیا چیزی به‌عنوان *کلید* در دیکشنری هست یا نه.

</div>

```python
'one' in numbers
```

<div dir="rtl" style="text-align: right;">

عملگر <code dir="ltr">in</code> بررسی *نمی‌کند* که چیزی به‌عنوان مقدار در دیکشنری هست یا نه.

</div>

```python
1 in numbers
```

<div dir="rtl" style="text-align: right;">

برای دیدن اینکه چیزی به‌عنوان مقدار در دیکشنری هست یا نه، می‌توانید از متد <code dir="ltr">values</code> استفاده کنید که دنباله‌ای از مقدارها را برمی‌گرداند، و بعد از عملگر <code dir="ltr">in</code> استفاده کنید.

</div>

```python
1 in numbers.values()
```

<div dir="rtl" style="text-align: right;">

موردهای یک دیکشنری پایتون در یک **جدول درهم‌سازی** (hash table) ذخیره می‌شوند، که شیوه‌ای برای سازمان‌دهی داده‌هاست و ویژگی قابل‌توجهی دارد: عملگر <code dir="ltr">in</code> تقریباً همان‌قدر زمان می‌برد، مهم نیست چند مورد در دیکشنری باشد. این باعث می‌شود بتوانیم الگوریتم‌هایی فوق‌العاده کارآمد بنویسیم.

</div>

```python
download('https://raw.githubusercontent.com/Alireza-Akhavan/ThinkPython/main/words.txt');
```

<div dir="rtl" style="text-align: right;">

برای نمایش، دو الگوریتم را برای پیدا کردن جفت کلمه‌هایی که یکی معکوس دیگری است – مثل <code dir="ltr">stressed</code> و <code dir="ltr">desserts</code> – مقایسه می‌کنیم. با خواندن فهرست کلمه‌ها شروع می‌کنیم.

</div>

```python
word_list = open('words.txt').read().split()
len(word_list)
```

<div dir="rtl" style="text-align: right;">

و این <code dir="ltr">reverse_word</code> از فصل قبل است.

</div>

```python
def reverse_word(word):
    return ''.join(reversed(word))
```

<div dir="rtl" style="text-align: right;">

تابع زیر کلمه‌های فهرست را پیمایش می‌کند. برای هر کدام، حرف‌ها را معکوس می‌کند و بعد بررسی می‌کند که کلمه‌ی معکوس‌شده در فهرست کلمه‌ها هست یا نه.

</div>

```python
def too_slow():
    count = 0
    for word in word_list:
        if reverse_word(word) in word_list:
            count += 1
    return count
```

<div dir="rtl" style="text-align: right;">

اجرای این تابع بیش از یک دقیقه طول می‌کشد. مشکل این است که عملگر <code dir="ltr">in</code> کلمه‌های فهرست را یکی‌یکی، از ابتدا، بررسی می‌کند. اگر چیزی را که دنبالش می‌گردد پیدا نکند – که بیشتر اوقات این‌طور است – باید تا انتها جست‌وجو کند.

برای اندازه‌گیری مدت اجرای یک تابع، می‌توانیم از <code dir="ltr">%time</code> استفاده کنیم که یکی از «دستورهای جادویی داخلی» (built-in magic commands) Jupyter است. این دستورها بخشی از زبان پایتون نیستند، پس ممکن است در محیط‌های توسعه‌ی دیگر کار نکنند.

</div>

```python
# %time too_slow()
```

<div dir="rtl" style="text-align: right;">

و عملگر <code dir="ltr">in</code> داخل حلقه است، پس برای هر کلمه یک بار اجرا می‌شود. چون بیش از ۱۰۰٬۰۰۰ کلمه در فهرست هست و برای هر کدام بیش از ۱۰۰٬۰۰۰ کلمه را بررسی می‌کنیم، تعداد کل مقایسه‌ها – تقریباً – مجذور تعداد کلمه‌هاست، که نزدیک به ۱۳ میلیارد است.

</div>

```python
len(word_list)**2
```

<div dir="rtl" style="text-align: right;">

می‌توانیم این تابع را با یک دیکشنری خیلی سریع‌تر کنیم. حلقه‌ی زیر دیکشنری‌ای می‌سازد که کلمه‌ها را به‌عنوان کلید دارد.

</div>

```python
word_dict = {}
for word in word_list:
    word_dict[word] = 1
```

<div dir="rtl" style="text-align: right;">

مقدارهای <code dir="ltr">word_dict</code> همه <code dir="ltr">1</code> هستند، اما می‌توانند هر چیزی باشند، چون هیچ‌وقت جست‌وجویشان نمی‌کنیم – فقط از این دیکشنری برای بررسی اینکه یک کلید وجود دارد یا نه استفاده می‌کنیم.

حالا نسخه‌ای از تابع قبلی که <code dir="ltr">word_list</code> را با <code dir="ltr">word_dict</code> جایگزین می‌کند.

</div>

```python
def much_faster():
    count = 0
    for word in word_dict:
        if reverse_word(word) in word_dict:
            count += 1
    return count
```

<div dir="rtl" style="text-align: right;">

اجرای این تابع کمتر از یک صدم ثانیه طول می‌کشد، پس تقریباً ۱۰٬۰۰۰ برابر سریع‌تر از نسخه‌ی قبلی است.

به‌طور کلی، زمانی که پیدا کردن یک عنصر در فهرست طول می‌کشد متناسب با طول فهرست است. زمانی که پیدا کردن یک کلید در دیکشنری طول می‌کشد تقریباً ثابت است – صرف‌نظر از تعداد موردها.

</div>

```python
%time much_faster()
```

<div dir="rtl" style="text-align: right;">

(نتیجه‌ی تابع ۸۸۵ است، یعنی ۸۸۵ کلمه در فهرست هست که معکوسشان هم در فهرست است؛ هر جفت دوبار شمرده می‌شود و کلمه‌های متقارن هم یک بار حساب می‌شوند. خط مربوط به زمان در خروجی <code dir="ltr">%time</code> به رایانه‌ی شما بستگی دارد.)

</div>

## ۱۰.۴. مجموعه‌ای از شمارنده‌ها

فرض کنید رشته‌ای به شما داده می‌شود و می‌خواهید بشمارید هر حرف چند بار ظاهر می‌شود. دیکشنری ابزار خوبی برای این کار است. با یک دیکشنری تهی شروع می‌کنیم.

</div>

```python
counter = {}
```

<div dir="rtl" style="text-align: right;">

وقتی حرف‌های رشته را پیمایش می‌کنیم، فرض کنید حرف <code dir="ltr">'a'</code> را برای اولین بار می‌بینیم. می‌توانیم آن را این‌طور به دیکشنری اضافه کنیم.

</div>

```python
counter['a'] = 1
```

<div dir="rtl" style="text-align: right;">

مقدار <code dir="ltr">1</code> نشان می‌دهد این حرف را یک بار دیده‌ایم. بعداً، اگر همان حرف را دوباره ببینیم، می‌توانیم شمارنده را این‌طور افزایش دهیم.

</div>

```python
counter['a'] += 1
```

<div dir="rtl" style="text-align: right;">

حالا مقدار وابسته به <code dir="ltr">'a'</code> برابر <code dir="ltr">2</code> است، چون این حرف را دو بار دیده‌ایم.

</div>

```python
counter
```

<div dir="rtl" style="text-align: right;">

تابع زیر از این امکانات برای شمردن تعداد دفعاتی که هر حرف در یک رشته ظاهر می‌شود استفاده می‌کند.

</div>

```python
def value_counts(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

<div dir="rtl" style="text-align: right;">

هر بار که حلقه تکرار می‌شود، اگر <code dir="ltr">letter</code> در دیکشنری نباشد، یک مورد جدید با کلید <code dir="ltr">letter</code> و مقدار <code dir="ltr">1</code> می‌سازیم. اگر <code dir="ltr">letter</code> از قبل در دیکشنری باشد، مقدار وابسته به <code dir="ltr">letter</code> را افزایش می‌دهیم.

این یک مثال است.

</div>

```python
counter = value_counts('brontosaurus')
counter
```

<div dir="rtl" style="text-align: right;">

موردهای <code dir="ltr">counter</code> نشان می‌دهند حرف <code dir="ltr">'b'</code> یک بار، <code dir="ltr">'r'</code> دو بار و به همین ترتیب ظاهر می‌شود.

</div>

<div dir="rtl" style="text-align: right;">

## ۱۰.۵. حلقه و دیکشنری‌ها

اگر از دیکشنری در یک دستور <code dir="ltr">for</code> استفاده کنید، کلیدهای دیکشنری را پیمایش می‌کند. برای نمایش، دیکشنری‌ای می‌سازیم که حرف‌های <code dir="ltr">'banana'</code> را می‌شمارد.

</div>

```python
counter = value_counts('banana')
counter
```

<div dir="rtl" style="text-align: right;">

حلقه‌ی زیر کلیدها، یعنی حرف‌ها، را نمایش می‌دهد.

</div>

```python
for key in counter:
    print(key)
```

<div dir="rtl" style="text-align: right;">

برای نمایش مقدارها می‌توانیم از متد <code dir="ltr">values</code> استفاده کنیم.

</div>

```python
for value in counter.values():
    print(value)
```

<div dir="rtl" style="text-align: right;">

برای نمایش کلیدها و مقدارها، می‌توانیم کلیدها را پیمایش کنیم و مقدارهای متناظر را جست‌وجو کنیم.

</div>

```python
for key in counter:
    value = counter[key]
    print(key, value)
```

<div dir="rtl" style="text-align: right;">

در فصل بعد، راه مختصرتری برای انجام همین کار خواهیم دید.

</div>

<div dir="rtl" style="text-align: right;">

## ۱۰.۶. فهرست‌ها و دیکشنری‌ها

می‌توانید یک فهرست را به‌عنوان مقدار در دیکشنری بگذارید. برای مثال، این دیکشنری‌ای است که عدد <code dir="ltr">4</code> را به فهرستی از چهار حرف نگاشت می‌کند.

</div>

```python
d = {4: ['r', 'o', 'u', 's']}
d
```

<div dir="rtl" style="text-align: right;">

اما نمی‌توانید یک فهرست را به‌عنوان کلید در دیکشنری بگذارید. اگر سعی کنیم این اتفاق می‌افتد.

</div>

```python
%%expect TypeError
letters = list('abcd')
d[letters] = 4
```

<div dir="rtl" style="text-align: right;">

**درهم‌ساز** (hash) تابعی است که مقداری (از هر نوع) را می‌گیرد و یک عدد صحیح برمی‌گرداند. دیکشنری‌ها از این عددهای صحیح، که مقدارهای درهم‌ساز (hash values) نامیده می‌شوند، برای ذخیره و جست‌وجوی کلیدها استفاده می‌کنند.

این سیستم فقط وقتی کار می‌کند که کلید تغییرناپذیر باشد، تا مقدار درهم‌سازش همیشه یکسان بماند. اما اگر کلید تغییرپذیر باشد، مقدار درهم‌سازش ممکن است تغییر کند و دیکشنری درست کار نکند. به همین دلیل کلیدها باید درهم‌پذیر باشند، و به همین دلیل نوع‌های تغییرپذیر مثل فهرست‌ها درهم‌پذیر نیستند.

چون دیکشنری‌ها تغییرپذیرند، به‌عنوان کلید هم نمی‌توانند استفاده شوند. اما *می‌توانند* به‌عنوان مقدار به کار روند.

## ۱۰.۷. انباشتن یک فهرست

در بسیاری از وظایف برنامه‌نویسی، مفید است که یک فهرست یا دیکشنری را پیمایش کنیم و همزمان فهرست دیگری بسازیم. به‌عنوان مثال، کلمه‌های <code dir="ltr">word_dict</code> را پیمایش می‌کنیم و فهرستی از کلمه‌های متقارن می‌سازیم – یعنی کلمه‌هایی که از جلو و عقب یک‌جور نوشته می‌شوند، مثل «noon» و «rotator».

در فصل قبل، یکی از تمرین‌ها از شما خواسته بود تابعی بنویسید که بررسی کند کلمه‌ای متقارن است یا نه. این یک راه‌حل است که از <code dir="ltr">reverse_word</code> استفاده می‌کند.

</div>

```python
def is_palindrome(word):
    """Check if a word is a palindrome."""
    return reverse_word(word) == word
```

<div dir="rtl" style="text-align: right;">

اگر کلمه‌های <code dir="ltr">word_dict</code> را پیمایش کنیم، می‌توانیم تعداد کلمه‌های متقارن را این‌طور بشماریم.

</div>

```python
count = 0

for word in word_dict:
    if is_palindrome(word):
        count +=1
        
count
```

<div dir="rtl" style="text-align: right;">

تا الان این الگو آشناست.

* قبل از حلقه، <code dir="ltr">count</code> با <code dir="ltr">0</code> مقداردهی اولیه می‌شود.

* داخل حلقه، اگر <code dir="ltr">word</code> متقارن باشد، <code dir="ltr">count</code> را افزایش می‌دهیم.

* وقتی حلقه تمام می‌شود، <code dir="ltr">count</code> شامل تعداد کل کلمه‌های متقارن است.

می‌توانیم از الگوی مشابهی برای ساختن فهرستی از کلمه‌های متقارن استفاده کنیم.

</div>

```python
palindromes = []

for word in word_dict:
    if is_palindrome(word):
        palindromes.append(word)

palindromes[:10]
```

این‌طور کار می‌کند:

* قبل از حلقه، <code dir="ltr">palindromes</code> با یک فهرست تهی مقداردهی اولیه می‌شود.

* داخل حلقه، اگر <code dir="ltr">word</code> متقارن باشد، آن را به انتهای <code dir="ltr">palindromes</code> اضافه می‌کنیم.

* وقتی حلقه تمام می‌شود، <code dir="ltr">palindromes</code> فهرستی از کلمه‌های متقارن است.

در این حلقه، <code dir="ltr">palindromes</code> به‌عنوان **انباشتگر** (accumulator) استفاده می‌شود، یعنی متغیری که در طول یک محاسبه داده را جمع یا انباشته می‌کند.

حالا فرض کنید می‌خواهیم فقط کلمه‌های متقارن با هفت حرف یا بیشتر را انتخاب کنیم. می‌توانیم <code dir="ltr">palindromes</code> را پیمایش کنیم و فهرست جدیدی بسازیم که فقط کلمه‌های متقارن بلند را دارد.

## ۱۰.۸. یادداشت‌ها (Memos)

اگر تابع <code dir="ltr">fibonacci</code> را از فصل ۶ اجرا کرده باشید، شاید متوجه شده باشید که هرچه آرگومان بزرگ‌تری بدهید، تابع زمان بیشتری برای اجرا لازم دارد.

</div>

```python
def fibonacci(n):
    if n == 0:
        return 0
    
    if n == 1:
        return 1

    return fibonacci(n-1) + fibonacci(n-2)
```

<div dir="rtl" style="text-align: right;">

علاوه بر این، زمان اجرا به‌سرعت افزایش می‌یابد. برای فهمیدن دلیلش، شکل زیر را در نظر بگیرید که **گراف فراخوانی** (call graph) تابع <code dir="ltr">fibonacci</code> را با <code dir="ltr">n=4</code> نشان می‌دهد:

(در کتاب اصلی اینجا نمودار رسم شده است: درختی از قاب‌ها با <code dir="ltr">n=4</code> در بالا؛ هر قاب <code dir="ltr">n</code> به قاب‌های <code dir="ltr">n-1</code> و <code dir="ltr">n-2</code> خط می‌کشد تا به <code dir="ltr">n=1</code> و <code dir="ltr">n=0</code> برسد.)

گراف فراخوانی مجموعه‌ای از قاب‌های تابع را نشان می‌دهد، با خط‌هایی که هر قاب را به قاب‌های توابعی که فراخوانی می‌کند وصل می‌کنند. در بالای گراف، <code dir="ltr">fibonacci</code> با <code dir="ltr">n=4</code> توابع <code dir="ltr">fibonacci</code> با <code dir="ltr">n=3</code> و <code dir="ltr">n=2</code> را فراخوانی می‌کند. به نوبه‌ی خود، <code dir="ltr">fibonacci</code> با <code dir="ltr">n=3</code> توابع <code dir="ltr">fibonacci</code> با <code dir="ltr">n=2</code> و <code dir="ltr">n=1</code> را فراخوانی می‌کند. و همین‌طور ادامه می‌یابد.

بشمارید <code dir="ltr">fibonacci(0)</code> و <code dir="ltr">fibonacci(1)</code> چند بار فراخوانی می‌شوند. این راه‌حلی ناکارآمد برای مسئله است و هرچه آرگومان بزرگ‌تر شود بدتر می‌شود.

یک راه‌حل این است که مقدارهایی را که از قبل محاسبه شده‌اند با ذخیره کردنشان در یک دیکشنری پیگیری کنیم. مقداری که قبلاً محاسبه شده و برای استفاده‌ی بعدی ذخیره می‌شود **یادداشت** (memo) نامیده می‌شود. این یک نسخه‌ی «یادداشت‌دار» (memoized) از <code dir="ltr">fibonacci</code> است:

</div>

```python
known = {0:0, 1:1}

def fibonacci_memo(n):
    if n in known:
        return known[n]

    res = fibonacci_memo(n-1) + fibonacci_memo(n-2)
    known[n] = res
    return res
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">known</code> دیکشنری‌ای است که عددهای فیبوناچی را که از قبل می‌دانیم پیگیری می‌کند. با دو مورد شروع می‌شود: <code dir="ltr">0</code> به <code dir="ltr">0</code> و <code dir="ltr">1</code> به <code dir="ltr">1</code> نگاشت می‌شود.

هر بار که <code dir="ltr">fibonacci_memo</code> فراخوانی می‌شود، <code dir="ltr">known</code> را بررسی می‌کند. اگر نتیجه از قبل آنجا باشد، می‌تواند بلافاصله برگرداند. در غیر این صورت باید مقدار جدید را محاسبه کند، به دیکشنری اضافه کند و برگرداند.

با مقایسه‌ی این دو تابع، <code dir="ltr">fibonacci(40)</code> حدود ۳۰ ثانیه طول می‌کشد. <code dir="ltr">fibonacci_memo(40)</code> حدود ۳۰ میکروثانیه طول می‌کشد، پس یک میلیون برابر سریع‌تر است. در نوت‌بوک این فصل می‌بینید این اندازه‌گیری‌ها از کجا می‌آیند.

</div>

```python
# %time fibonacci(40)
```

```python
%time fibonacci_memo(40)
```

## ۱۰.۹. اشکال‌زدایی

وقتی با مجموعه‌داده‌های بزرگ‌تر کار می‌کنید، اشکال‌زدایی با نمایش و بررسی دستی خروجی می‌تواند دست‌وپاگیر شود. این‌ها پیشنهادهایی برای اشکال‌زدایی مجموعه‌داده‌های بزرگ‌اند:

1.  **ورودی را کوچک کنید:** اگر ممکن است، اندازه‌ی مجموعه‌داده را کم کنید. برای مثال اگر برنامه یک فایل متنی را می‌خواند، با فقط ۱۰ خط اول یا کوچک‌ترین نمونه‌ای که می‌توانید پیدا کنید شروع کنید. می‌توانید خود فایل‌ها را ویرایش کنید، یا (بهتر) برنامه را طوری تغییر دهید که فقط <code dir="ltr">n</code> خط اول را بخواند.

    اگر خطایی وجود دارد، می‌توانید <code dir="ltr">n</code> را به کوچک‌ترین مقداری که خطا در آن رخ می‌دهد کاهش دهید. همین‌طور که خطاها را پیدا و اصلاح می‌کنید، می‌توانید <code dir="ltr">n</code> را کم‌کم افزایش دهید.

2.  **خلاصه‌ها و نوع‌ها را بررسی کنید:** به‌جای نمایش و بررسی کل مجموعه‌داده، خلاصه‌های داده را نمایش دهید – برای مثال، تعداد موردهای یک دیکشنری یا مجموع فهرستی از عددها.

    یک علت رایج خطاهای زمان اجرا مقداری است که نوع درستی ندارد. برای اشکال‌زدایی این نوع خطا، اغلب کافی است نوع یک مقدار را نمایش دهید.

3.  **خودآزمون بنویسید:** گاهی می‌توانید کدی بنویسید که خطاها را خودکار بررسی کند. برای مثال، اگر میانگین فهرستی از عددها را حساب می‌کنید، می‌توانید بررسی کنید که نتیجه از بزرگ‌ترین عنصر فهرست بزرگ‌تر یا از کوچک‌ترین عنصر کوچک‌تر نیست. این «بررسی سلامت عقل» (sanity check) نامیده می‌شود، چون نتیجه‌های «نامعقول» را تشخیص می‌دهد.

    نوع دیگری از بررسی، نتیجه‌های دو محاسبه‌ی متفاوت را مقایسه می‌کند تا ببیند سازگارند یا نه. این «بررسی سازگاری» (consistency check) نامیده می‌شود.

4.  **خروجی را قالب‌بندی کنید:** قالب‌بندی خروجی اشکال‌زدایی می‌تواند تشخیص خطا را آسان‌تر کند. نمونه‌ای از این را در فصل ۶ دیدیم. ابزار دیگری که ممکن است مفید باشد ماژول <code dir="ltr">pprint</code> است که تابع <code dir="ltr">pprint</code> را فراهم می‌کند و نوع‌های داخلی را با قالبی خواناتر برای انسان نمایش می‌دهد (<code dir="ltr">pprint</code> مخفف «pretty print» به معنی «چاپ زیبا» است).

    باز هم، زمانی که صرف ساختن داربست می‌کنید می‌تواند زمانی را که صرف اشکال‌زدایی می‌کنید کم کند.

## ۱۰.۱۰. واژه‌نامه

**دیکشنری (dictionary):** شیئی که زوج‌های کلید-مقدار را در بر دارد که موردها (items) هم نامیده می‌شوند.

**مورد (item):** در دیکشنری، نام دیگر یک زوج کلید-مقدار.

**کلید (key):** شیئی که در دیکشنری به‌عنوان بخش اول یک زوج کلید-مقدار ظاهر می‌شود.

**مقدار (value):** شیئی که در دیکشنری به‌عنوان بخش دوم یک زوج کلید-مقدار ظاهر می‌شود. این دقیق‌تر از کاربرد قبلی ما از واژه‌ی «مقدار» است.

**نگاشت (mapping):** رابطه‌ای که در آن هر عنصر یک مجموعه با یک عنصر از مجموعه‌ی دیگر متناظر است.

**جدول درهم‌سازی (hash table):** مجموعه‌ای از زوج‌های کلید-مقدار که طوری سازمان‌دهی شده تا بتوانیم یک کلید را جست‌وجو کنیم و مقدارش را کارآمد پیدا کنیم.

**درهم‌پذیر (hashable):** نوع‌های تغییرناپذیر مثل عددهای صحیح، اعشاری و رشته‌ها درهم‌پذیرند. نوع‌های تغییرپذیر مثل فهرست‌ها و دیکشنری‌ها نیستند.

**تابع درهم‌ساز (hash function):** تابعی که یک شیء را می‌گیرد و عدد صحیحی محاسبه می‌کند که برای پیدا کردن مکان یک کلید در جدول درهم‌سازی به کار می‌رود.

**انباشتگر (accumulator):** متغیری که در یک حلقه برای جمع زدن یا انباشتن یک نتیجه به کار می‌رود.

**پالایش (filtering):** پیمایش یک دنباله و انتخاب یا حذف عنصرها.

**گراف فراخوانی (call graph):** نموداری که هر قابی را که در طول اجرای برنامه ساخته می‌شود نشان می‌دهد، با پیکانی از هر فراخواننده به هر فراخوانده.

**یادداشت (memo):** مقداری محاسبه‌شده که برای جلوگیری از محاسبه‌ی غیرضروری در آینده ذخیره می‌شود.
