## ۶.۳. مقدارهای بازگشتی و دستورهای شرطی

اگر پایتون <code dir="ltr">abs</code> را فراهم نکرده بود، می‌توانستیم آن را این‌طور بنویسیم.

</div>

```python
def absolute_value(x):
    if x < 0:
        return -x
    else:
        return x
```

<div dir="rtl" style="text-align: right;">

اگر <code dir="ltr">x</code> منفی باشد، دستور <code dir="ltr">return</code> اول <code dir="ltr">-x</code> را برمی‌گرداند و تابع بلافاصله تمام می‌شود. در غیر این صورت، دستور <code dir="ltr">return</code> دوم <code dir="ltr">x</code> را برمی‌گرداند و تابع تمام می‌شود. پس این تابع درست است.

با این حال، اگر دستورهای <code dir="ltr">return</code> را در یک شرط بگذارید، باید مطمئن شوید که هر مسیر ممکن در برنامه به یک دستور <code dir="ltr">return</code> می‌رسد. برای مثال، این یک نسخه‌ی نادرست از <code dir="ltr">absolute_value</code> است.

</div>

```python
def absolute_value_wrong(x):
    if x < 0:
        return -x
    if x > 0:
        return x
```

<div dir="rtl" style="text-align: right;">

ببینید اگر این تابع را با <code dir="ltr">0</code> به‌عنوان آرگومان فراخوانی کنیم چه می‌شود.

</div>

```python
absolute_value_wrong(0)
```

<div dir="rtl" style="text-align: right;">

چیزی نمی‌گیریم! مشکل این است: وقتی <code dir="ltr">x</code> برابر <code dir="ltr">0</code> است، هیچ‌کدام از شرط‌ها درست نیست و تابع بدون رسیدن به دستور <code dir="ltr">return</code> تمام می‌شود؛ یعنی مقدار بازگشتی <code dir="ltr">None</code> است و به همین دلیل Jupyter چیزی نمایش نمی‌دهد.

به‌عنوان مثالی دیگر، این نسخه‌ای از <code dir="ltr">absolute_value</code> است که یک دستور <code dir="ltr">return</code> اضافه در انتها دارد.

</div>

```python
def absolute_value_extra_return(x):
    if x < 0:
        return -x
    else:
        return x
    
    return 'This is dead code'
```

<div dir="rtl" style="text-align: right;">

اگر <code dir="ltr">x</code> منفی باشد، دستور <code dir="ltr">return</code> اول اجرا می‌شود و تابع تمام می‌شود. در غیر این صورت دستور <code dir="ltr">return</code> دوم اجرا می‌شود و تابع تمام می‌شود. در هر دو حالت هرگز به دستور <code dir="ltr">return</code> سوم نمی‌رسیم – پس هرگز نمی‌تواند اجرا شود.

کدی که هرگز نمی‌تواند اجرا شود **کد مرده** (dead code) نامیده می‌شود. به‌طور کلی کد مرده ضرری ندارد، اما اغلب نشانه‌ی یک سوءتفاهم است و ممکن است برای کسی که می‌خواهد برنامه را بفهمد گیج‌کننده باشد.

</div>

به‌عنوان مثال، فرض کنید می‌خواهید فاصله‌ی بین دو نقطه را که با مختصات $(x_1, y_1)$ و $(x_2, y_2)$ نمایش داده می‌شوند پیدا کنید. طبق قضیه‌ی فیثاغورس، فاصله این است:

$$\mathrm{distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

برای این تابع، ورودی‌ها مختصات نقطه‌ها هستند. مقدار بازگشتی فاصله است. بلافاصله می‌توانید طرح کلی تابع را بنویسید:

</div>

```python
def distance(x1, y1, x2, y2):
    return 0.0
```

<div dir="rtl" style="text-align: right;">

این نسخه هنوز فاصله‌ها را حساب نمی‌کند – همیشه صفر برمی‌گرداند. اما تابعی کامل با یک مقدار بازگشتی است، یعنی می‌توانید قبل از پیچیده‌تر کردنش آن را آزمایش کنید.

برای آزمودن تابع جدید، آن را با آرگومان‌های نمونه فراخوانی می‌کنیم:

</div>

```python
distance(1, 2, 4, 6)
```

در این نقطه تأیید کرده‌ایم که تابع اجرا می‌شود و یک مقدار برمی‌گرداند و می‌توانیم شروع به اضافه کردن کد به بدنه کنیم. قدم بعدی خوب این است که تفاضل‌های <code dir="ltr">x2 - x1</code> و <code dir="ltr">y2 - y1</code> را پیدا کنیم. این نسخه‌ای است که آن مقدارها را در متغیرهای موقت ذخیره می‌کند و نمایش می‌دهد.

</div>

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    print('dx is', dx)
    print('dy is', dy)
    return 0.0
```

<div dir="rtl" style="text-align: right;">

اگر تابع درست کار کند، باید <code dir="ltr">dx is 3</code> و <code dir="ltr">dy is 4</code> را نمایش دهد. اگر این‌طور باشد، می‌دانیم تابع آرگومان‌های درست را می‌گیرد و اولین محاسبه را درست انجام می‌دهد. اگر نه، فقط چند خط برای بررسی هست.

</div>

```python
distance(1, 2, 4, 6)
```

<div dir="rtl" style="text-align: right;">

تا اینجا خوب است. بعد مجموع مربعات <code dir="ltr">dx</code> و <code dir="ltr">dy</code> را حساب می‌کنیم:

</div>

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    print('dsquared is: ', dsquared)
    return 0.0
```

<div dir="rtl" style="text-align: right;">

دوباره می‌توانیم تابع را اجرا کنیم و خروجی را بررسی کنیم که باید <code dir="ltr">25</code> باشد.

</div>

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    result = math.sqrt(dsquared)
    print("result is", result)
```

</div>

دستورهای <code dir="ltr">print</code> که نوشتیم برای اشکال‌زدایی مفیدند، اما وقتی تابع درست کار کرد می‌توانیم آن‌ها را حذف کنیم. به کدی مثل این **داربست** (scaffolding) می‌گویند، چون برای ساختن برنامه مفید است اما بخشی از محصول نهایی نیست.

این مثال توسعه‌ی تدریجی را نشان می‌دهد. جنبه‌های کلیدی این فرایند این‌ها هستند:

1.  با یک برنامه‌ی کارآمد شروع کنید، تغییرات کوچک بدهید و بعد از هر تغییر آزمایش کنید.

2.  از متغیرها برای نگه داشتن مقدارهای میانی استفاده کنید تا بتوانید آن‌ها را نمایش دهید و بررسی کنید.

3.  وقتی برنامه کار کرد، داربست را حذف کنید.

در هر نقطه، اگر خطایی وجود داشته باشد، باید تصور خوبی از جای آن داشته باشید. توسعه‌ی تدریجی می‌تواند زمان زیادی از اشکال‌زدایی شما را نجات دهد.

## ۶.۵. توابع بولی

توابع می‌توانند مقدارهای بولی <code dir="ltr">True</code> و <code dir="ltr">False</code> را برگردانند که اغلب برای کپسوله کردن یک آزمون پیچیده در یک تابع راحت است. برای مثال، <code dir="ltr">is_divisible</code> بررسی می‌کند که آیا <code dir="ltr">x</code> بدون باقی‌مانده بر <code dir="ltr">y</code> بخش‌پذیر است یا نه.

</div>

```python
def is_divisible(x, y):
    if x % y == 0:
        return True
    else:
        return False
```

</div>

```python
is_divisible(6, 4)
```

```python
def is_divisible(x, y):
    return x % y == 0
```

## ۶.۶. بازگشت با مقدارهای بازگشتی

حالا که می‌توانیم توابعی با مقدار بازگشتی بنویسیم، می‌توانیم توابع بازگشتی با مقدار بازگشتی هم بنویسیم، و با این توانایی از یک آستانه‌ی مهم عبور کرده‌ایم – زیرمجموعه‌ای از پایتون که داریم اکنون **تورینگ‌کامل** (Turing complete) است، یعنی می‌توانیم هر محاسبه‌ای را که با یک الگوریتم قابل توصیف است انجام دهیم.

برای نشان دادن بازگشت با مقدارهای بازگشتی، چند تابع ریاضی را که به‌صورت بازگشتی تعریف شده‌اند ارزیابی می‌کنیم. تعریف بازگشتی شبیه تعریف دوری است، به این معنا که تعریف به چیزی که تعریف می‌شود ارجاع می‌دهد. یک تعریف واقعاً دوری چندان مفید نیست:

> vorpal: صفتی که برای توصیف چیزی که vorpal است به کار می‌رود.

اگر این تعریف را در فرهنگ لغت ببینید، شاید عصبانی شوید. از طرف دیگر، اگر تعریف تابع فاکتوریل را که با نماد $!$ نشان داده می‌شود جست‌وجو کنید، ممکن است چیزی شبیه این ببینید:

$$\begin{aligned}
0! &= 1 \\
n! &= n~(n-1)!
\end{aligned}$$

این تعریف می‌گوید فاکتوریل $0$ برابر $1$ است و فاکتوریل هر مقدار دیگر، $n$، برابر $n$ ضرب در فاکتوریل $n-1$ است.

اگر بتوانید یک تعریف بازگشتی از چیزی بنویسید، می‌توانید برنامه‌ی پایتونی برای ارزیابی آن بنویسید. با پیروی از فرایند توسعه‌ی تدریجی، با تابعی شروع می‌کنیم که <code dir="ltr">n</code> را به‌عنوان پارامتر می‌گیرد و همیشه <code dir="ltr">0</code> برمی‌گرداند.

</div>

```python
def factorial(n):
    return 0
```

<div dir="rtl" style="text-align: right;">

حالا بخش اول تعریف را اضافه کنیم – اگر آرگومان <code dir="ltr">0</code> بود، فقط باید <code dir="ltr">1</code> را برگردانیم:

</div>

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return 0
```

<div dir="rtl" style="text-align: right;">

حالا بخش دوم را کامل کنیم – اگر <code dir="ltr">n</code> برابر <code dir="ltr">0</code> نبود، باید یک فراخوانی بازگشتی انجام دهیم تا فاکتوریل <code dir="ltr">n-1</code> را پیدا کنیم و بعد نتیجه را در <code dir="ltr">n</code> ضرب کنیم:

</div>

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        recurse = factorial(n-1)
        return n * recurse
```

<div dir="rtl" style="text-align: right;">

جریان اجرای این برنامه شبیه جریان <code dir="ltr">countdown</code> در فصل ۵ است. اگر <code dir="ltr">factorial</code> را با مقدار <code dir="ltr">3</code> فراخوانی کنیم:

چون <code dir="ltr">3</code> برابر <code dir="ltr">0</code> نیست، شاخه‌ی دوم را می‌گیریم و فاکتوریل <code dir="ltr">n-1</code> را حساب می‌کنیم…

> چون <code dir="ltr">2</code> برابر <code dir="ltr">0</code> نیست، شاخه‌ی دوم را می‌گیریم و فاکتوریل <code dir="ltr">n-1</code> را حساب می‌کنیم…
>
> > چون <code dir="ltr">1</code> برابر <code dir="ltr">0</code> نیست، شاخه‌ی دوم را می‌گیریم و فاکتوریل <code dir="ltr">n-1</code> را حساب می‌کنیم…
> >
> > > چون <code dir="ltr">0</code> برابر <code dir="ltr">0</code> است، شاخه‌ی اول را می‌گیریم و بدون انجام فراخوانی بازگشتی دیگری <code dir="ltr">1</code> را برمی‌گردانیم.
> >
> > مقدار بازگشتی، <code dir="ltr">1</code>، در <code dir="ltr">n</code> که <code dir="ltr">1</code> است ضرب می‌شود و نتیجه برگردانده می‌شود.
>
> مقدار بازگشتی، <code dir="ltr">1</code>، در <code dir="ltr">n</code> که <code dir="ltr">2</code> است ضرب می‌شود و نتیجه برگردانده می‌شود.

مقدار بازگشتی <code dir="ltr">2</code> در <code dir="ltr">n</code> که <code dir="ltr">3</code> است ضرب می‌شود و نتیجه، <code dir="ltr">6</code>، به مقدار بازگشتی فراخوانی تابعی تبدیل می‌شود که کل این فرایند را شروع کرد.

شکل زیر نمودار پشته‌ی این دنباله‌ی فراخوانی‌ها را نشان می‌دهد.

(در کتاب اصلی این‌جا نمودار رسم شده است: قاب <code dir="ltr">__main__</code> و زیرش قاب‌های <code dir="ltr">factorial</code> با <code dir="ltr">n</code> برابر ۳، ۲، ۱ و ۰، که مقدارهای بازگشتی ۶، ۲، ۱ و ۱ کنارشان نوشته شده است.)

مقدارهای بازگشتی در حال برگشتن از پشته نشان داده شده‌اند. در هر قاب، مقدار بازگشتی حاصل‌ضرب <code dir="ltr">n</code> و <code dir="ltr">recurse</code> است.

در آخرین قاب، متغیر محلی <code dir="ltr">recurse</code> وجود ندارد، چون شاخه‌ای که آن را می‌سازد اجرا نمی‌شود.

## ۶.۷. جهش ایمان

دنبال کردن جریان اجرا یک راه برای خواندن برنامه‌هاست، اما می‌تواند به‌سرعت طاقت‌فرسا شود. یک جایگزین چیزی است که من آن را «جهش ایمان» (leap of faith) می‌نامم. وقتی به یک فراخوانی تابع می‌رسید، به‌جای دنبال کردن جریان اجرا، *فرض می‌کنید* که تابع درست کار می‌کند و نتیجه‌ی درست را برمی‌گرداند.

در واقع وقتی از توابع داخلی استفاده می‌کنید، همین جهش ایمان را تمرین می‌کنید. وقتی <code dir="ltr">abs</code> یا <code dir="ltr">math.sqrt</code> را فراخوانی می‌کنید، بدنه‌ی آن توابع را بررسی نمی‌کنید – فقط فرض می‌کنید کار می‌کنند.

درباره‌ی فراخوانی توابع خودتان هم همین‌طور است. برای مثال، پیش‌تر تابعی به نام <code dir="ltr">is_divisible</code> نوشتیم که تعیین می‌کند آیا یک عدد بر عدد دیگر بخش‌پذیر است. وقتی خودمان را قانع کردیم که این تابع درست است، می‌توانیم بدون نگاه دوباره به بدنه‌اش از آن استفاده کنیم.

درباره‌ی برنامه‌های بازگشتی هم همین‌طور است. وقتی به فراخوانی بازگشتی می‌رسید، به‌جای دنبال کردن جریان اجرا، باید فرض کنید فراخوانی بازگشتی کار می‌کند و بعد از خودتان بپرسید: «با فرض اینکه بتوانم فاکتوریل $n-1$ را حساب کنم، آیا می‌توانم فاکتوریل $n$ را حساب کنم؟» تعریف بازگشتی فاکتوریل نشان می‌دهد که می‌توانید، با ضرب در $n$.

البته کمی عجیب است که فرض کنید تابع درست کار می‌کند در حالی که هنوز نوشتنش را تمام نکرده‌اید، اما برای همین به آن «جهش ایمان» می‌گویند!

## ۶.۸. فیبوناچی

بعد از <code dir="ltr">factorial</code>، رایج‌ترین مثال یک تابع بازگشتی <code dir="ltr">fibonacci</code> است که تعریفش این است:

$$\begin{aligned}
\mathrm{fibonacci}(0) &= 0 \\
\mathrm{fibonacci}(1) &= 1 \\
\mathrm{fibonacci}(n) &= \mathrm{fibonacci}(n-1) + \mathrm{fibonacci}(n-2)
\end{aligned}$$

وقتی به پایتون ترجمه می‌شود، این‌طور به نظر می‌رسد:

</div>

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif  n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

<div dir="rtl" style="text-align: right;">

اگر سعی کنید جریان اجرا را این‌جا دنبال کنید، حتی برای مقدارهای کوچک $n$، سرتان منفجر می‌شود. اما طبق جهش ایمان، اگر فرض کنید دو فراخوانی بازگشتی درست کار می‌کنند، می‌توانید مطمئن باشید دستور <code dir="ltr">return</code> آخر درست است.

به‌عنوان نکته‌ی جانبی، این روش محاسبه‌ی اعداد فیبوناچی بسیار ناکارآمد است. در فصل ۱۰ توضیح می‌دهم چرا و راهی برای بهبودش پیشنهاد می‌کنم.

</div>

```python
[fibonacci(i) for i in range(10)]
```

## ۶.۹. بررسی نوع‌ها

اگر <code dir="ltr">factorial</code> را فراخوانی کنیم و <code dir="ltr">1.5</code> را به‌عنوان آرگومان بدهیم چه می‌شود؟

</div>

```python
%%expect RecursionError

factorial(1.5)
```

<div dir="rtl" style="text-align: right;">

شبیه یک بازگشت بی‌نهایت است. چطور ممکن است؟ تابع حالت‌های پایه‌ای دارد، وقتی <code dir="ltr">n == 1</code> یا <code dir="ltr">n == 0</code>. اما اگر <code dir="ltr">n</code> عدد صحیح نباشد، می‌توانیم حالت پایه را *رد کنیم* و برای همیشه بازگشت کنیم.

در این مثال، مقدار اولیه‌ی <code dir="ltr">n</code> برابر <code dir="ltr">1.5</code> است. در اولین فراخوانی بازگشتی مقدار <code dir="ltr">n</code> برابر <code dir="ltr">0.5</code> است. در بعدی <code dir="ltr">-0.5</code> است. از آن‌جا کوچک‌تر (منفی‌تر) می‌شود، اما هرگز <code dir="ltr">0</code> نمی‌شود.

برای جلوگیری از بازگشت بی‌نهایت می‌توانیم از تابع داخلی <code dir="ltr">isinstance</code> برای بررسی نوع آرگومان استفاده کنیم. این‌طور بررسی می‌کنیم که یک مقدار عدد صحیح است یا نه.

</div>

```python
isinstance(3, int)
```

```python
isinstance(1.5, int)
```

<div dir="rtl" style="text-align: right;">

حالا این نسخه‌ای از <code dir="ltr">factorial</code> با بررسی خطاست.

</div>

```python
def factorial(n):
    if not isinstance(n, int):
        print('factorial is only defined for integers.')
        return None
    elif n < 0:
        print('factorial is not defined for negative numbers.')
        return None
    elif n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

<div dir="rtl" style="text-align: right;">

اول بررسی می‌کند که <code dir="ltr">n</code> عدد صحیح است یا نه. اگر نباشد، پیام خطا نمایش می‌دهد و <code dir="ltr">None</code> برمی‌گرداند.

</div>

```python
factorial('hello')
```

<div dir="rtl" style="text-align: right;">

بعد بررسی می‌کند که <code dir="ltr">n</code> منفی است یا نه. اگر باشد، پیام خطا نمایش می‌دهد و <code dir="ltr">None</code> برمی‌گرداند.

</div>

```python
factorial(-2)
```

<div dir="rtl" style="text-align: right;">

اگر از هر دو بررسی بگذریم، می‌دانیم <code dir="ltr">n</code> یک عدد صحیح نامنفی است، پس می‌توانیم مطمئن باشیم بازگشت تمام می‌شود. بررسی پارامترهای یک تابع برای اطمینان از اینکه نوع و مقدار درستی دارند **اعتبارسنجی ورودی** (input validation) نامیده می‌شود.

## ۶.۱۰. اشکال‌زدایی

تقسیم یک برنامه‌ی بزرگ به توابع کوچک‌تر نقاط بازرسی طبیعی برای اشکال‌زدایی می‌سازد. اگر تابعی درست کار نمی‌کند، سه احتمال برای بررسی وجود دارد:

-   مشکلی در آرگومان‌هایی که تابع می‌گیرد وجود دارد – یعنی یک پیش‌شرط نقض شده است.

-   مشکلی در خود تابع وجود دارد – یعنی یک پس‌شرط نقض شده است.

-   فراخواننده با مقدار بازگشتی کار اشتباهی می‌کند.

برای رد کردن احتمال اول، می‌توانید یک دستور <code dir="ltr">print</code> در ابتدای تابع اضافه کنید که مقدار پارامترها (و شاید نوعشان) را نمایش دهد. یا می‌توانید کدی بنویسید که پیش‌شرط‌ها را صریحاً بررسی کند.

اگر پارامترها خوب به نظر می‌رسند، می‌توانید قبل از هر دستور <code dir="ltr">return</code> یک دستور <code dir="ltr">print</code> اضافه کنید و مقدار بازگشتی را نمایش دهید. در صورت امکان، تابع را با آرگومان‌هایی فراخوانی کنید که بررسی نتیجه را آسان می‌کنند.

اگر تابع به نظر درست کار می‌کند، فراخوانی تابع را نگاه کنید تا مطمئن شوید مقدار بازگشتی درست استفاده می‌شود – یا اصلاً استفاده می‌شود!

اضافه کردن دستورهای <code dir="ltr">print</code> در ابتدا و انتهای یک تابع می‌تواند به قابل‌مشاهده‌تر شدن جریان اجرا کمک کند. برای مثال، این نسخه‌ای از <code dir="ltr">factorial</code> با دستورهای print است:

</div>

```python
def factorial(n):
    space = ' ' * (4 * n)
    print(space, 'factorial', n)
    if n == 0:
        print(space, 'returning 1')
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        print(space, 'returning', result)
        return result
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">space</code> رشته‌ای از نویسه‌های فاصله است که تورفتگی خروجی را کنترل می‌کند. این نتیجه‌ی <code dir="ltr">factorial(3)</code> است:

</div>

```python
factorial(3)
```

<div dir="rtl" style="text-align: right;">

اگر درباره‌ی جریان اجرا گیج شده‌اید، این نوع خروجی می‌تواند مفید باشد. مدتی طول می‌کشد تا داربست مؤثر بسازید، اما مقدار کمی داربست می‌تواند خیلی از اشکال‌زدایی را نجات دهد.

</div>

## ۶.۱۱. واژه‌نامه

**مقدار بازگشتی (return value):** نتیجه‌ی یک تابع. اگر فراخوانی تابع به‌عنوان یک عبارت به کار رود، مقدار بازگشتی مقدار آن عبارت است.

**تابع خالص (pure function):** تابعی که چیزی نمایش نمی‌دهد و هیچ اثر دیگری هم جز برگرداندن یک مقدار بازگشتی ندارد.

**کد مرده (dead code):** بخشی از یک برنامه که هرگز نمی‌تواند اجرا شود، اغلب چون بعد از یک دستور <code dir="ltr">return</code> آمده است.

**توسعه‌ی تدریجی (incremental development):** طرح توسعه‌ی برنامه‌ای که قصد دارد با افزودن و آزمودن مقدار کمی کد در هر زمان از اشکال‌زدایی دوری کند.

**داربست (scaffolding):** کدی که در طول توسعه‌ی برنامه به کار می‌رود اما بخشی از نسخه‌ی نهایی نیست.

**تورینگ‌کامل (Turing complete):** یک زبان، یا زیرمجموعه‌ای از یک زبان، تورینگ‌کامل است اگر بتواند هر محاسبه‌ای را که با یک الگوریتم قابل توصیف است انجام دهد.

**اعتبارسنجی ورودی (input validation):** بررسی پارامترهای یک تابع برای اطمینان از اینکه نوع و مقدار درستی دارند.

</div>

در این فصل تابعی نادرست دیدیم که می‌تواند بدون برگرداندن مقدار تمام شود.

</div>

```python
def absolute_value_wrong(x):
    if x < 0:
        return -x
    if x > 0:
        return x
```

<div dir="rtl" style="text-align: right;">

و نسخه‌ای از همان تابع که در انتها کد مرده دارد.

</div>

```python
def absolute_value_extra_return(x):
    if x < 0:
        return -x
    else:
        return x
    
    return 'This is dead code.'
```

<div dir="rtl" style="text-align: right;">

و مثال زیر را دیدیم که درست است اما سبک پایتونی (idiomatic) ندارد.

</div>

```python
def is_divisible(x, y):
    if x % y == 0:
        return True
    else:
        return False
```
