## ۵.۱. تقسیم صحیح و باقی‌مانده

به یاد بیاورید که عملگر تقسیم صحیح، <code dir="ltr">//</code>، دو عدد را بر هم تقسیم می‌کند و به پایین تا یک عدد صحیح گرد می‌کند. برای مثال، فرض کنید مدت یک فیلم ۱۰۵ دقیقه است. شاید بخواهید بدانید این چند ساعت است. تقسیم معمولی یک عدد اعشاری برمی‌گرداند:

</div>

```python
minutes = 105
minutes / 60
```

<div dir="rtl" style="text-align: right;">

اما ما معمولاً ساعت‌ها را با ممیز نمی‌نویسیم. تقسیم صحیح تعداد صحیح ساعت‌ها را، با گرد کردن به پایین، برمی‌گرداند:

</div>

```python
minutes = 105
hours = minutes // 60
hours
```

<div dir="rtl" style="text-align: right;">

برای به دست آوردن باقی‌مانده، می‌توانید یک ساعت را (بر حسب دقیقه) کم کنید:

</div>

```python
remainder = minutes - hours * 60
remainder
```

<div dir="rtl" style="text-align: right;">

یا می‌توانید از **عملگر باقی‌مانده** (modulus operator)، <code dir="ltr">%</code>، استفاده کنید که دو عدد را بر هم تقسیم می‌کند و باقی‌مانده را برمی‌گرداند.

</div>

```python
remainder = minutes % 60
remainder
```

<div dir="rtl" style="text-align: right;">

عملگر باقی‌مانده از آنچه به نظر می‌رسد مفیدتر است. برای مثال، می‌تواند بررسی کند که آیا یک عدد بر عدد دیگر بخش‌پذیر است یا نه – اگر <code dir="ltr">x % y</code> صفر باشد، <code dir="ltr">x</code> بر <code dir="ltr">y</code> بخش‌پذیر است.

همچنین می‌تواند رقم یا رقم‌های سمت راست یک عدد را استخراج کند. برای مثال، <code dir="ltr">x % 10</code> سمت‌راست‌ترین رقم <code dir="ltr">x</code> (در مبنای ۱۰) را می‌دهد. به همین ترتیب، <code dir="ltr">x % 100</code> دو رقم آخر را می‌دهد.

</div>

```python
x = 123
x % 10
```

```python
x % 100
```

<div dir="rtl" style="text-align: right;">

سرانجام، عملگر باقی‌مانده می‌تواند «حساب ساعتی» انجام دهد. برای مثال، اگر رویدادی ساعت ۱۱ صبح شروع شود و سه ساعت طول بکشد، می‌توانیم از عملگر باقی‌مانده برای فهمیدن ساعت پایان آن استفاده کنیم.

</div>

```python
start = 11
duration = 3
end = (start + duration) % 12
end
```

<div dir="rtl" style="text-align: right;">

این رویداد ساعت ۲ بعدازظهر تمام می‌شود.

</div>

## ۵.۲. عبارت‌های بولی

**عبارت بولی** (boolean expression) عبارتی است که یا درست است یا نادرست. برای مثال، عبارت‌های زیر از عملگر مساوی، <code dir="ltr">==</code>، استفاده می‌کنند که دو مقدار را مقایسه می‌کند و اگر برابر باشند <code dir="ltr">True</code> و در غیر این صورت <code dir="ltr">False</code> تولید می‌کند:

</div>

```python
5 == 5
```

```python
5 == 7
```

<div dir="rtl" style="text-align: right;">

یک خطای رایج استفاده از یک علامت مساوی (<code dir="ltr">=</code>) به جای دو علامت مساوی (<code dir="ltr">==</code>) است. یادتان باشد <code dir="ltr">=</code> یک مقدار را به یک متغیر نسبت می‌دهد و <code dir="ltr">==</code> دو مقدار را مقایسه می‌کند.

</div>

```python
x = 5
y = 7
```

```python
x == y
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">True</code> و <code dir="ltr">False</code> مقدارهای خاصی هستند که به نوع <code dir="ltr">bool</code> تعلق دارند؛ رشته نیستند:

</div>

```python
type(True)
```

```python
type(False)
```

<div dir="rtl" style="text-align: right;">

عملگر <code dir="ltr">==</code> یکی از **عملگرهای رابطه‌ای** (relational operators) است؛ بقیه این‌ها هستند:

</div>

```python
x != y               # x is not equal to y
```

```python
x > y                # x is greater than y
```

```python
x < y               # x is less than to y
```

```python
x >= y               # x is greater than or equal to y
```

```python
x <= y               # x is less than or equal to y
```

## ۵.۳. عملگرهای منطقی

برای ترکیب مقدارهای بولی در عبارت‌ها می‌توانیم از **عملگرهای منطقی** (logical operators) استفاده کنیم. رایج‌ترین آن‌ها <code dir="ltr">and</code>، <code dir="ltr">or</code> و <code dir="ltr">not</code> هستند. معنای این عملگرها شبیه معنایشان در زبان انگلیسی است. برای مثال، مقدار عبارت زیر فقط وقتی <code dir="ltr">True</code> است که <code dir="ltr">x</code> بزرگ‌تر از <code dir="ltr">0</code> *و* کوچک‌تر از <code dir="ltr">10</code> باشد.

</div>

```python
x > 0 and x < 10
```

<div dir="rtl" style="text-align: right;">

عبارت زیر اگر *یکی یا هر دو* شرط درست باشند <code dir="ltr">True</code> است، یعنی اگر عدد بر ۲ *یا* ۳ بخش‌پذیر باشد:

</div>

```python
x % 2 == 0 or x % 3 == 0
```

<div dir="rtl" style="text-align: right;">

</div>

```python
not x > y
```

</div>

```python
42 and True
```

## ۵.۴. دستورهای if

برای نوشتن برنامه‌های مفید، تقریباً همیشه به توانایی بررسی شرط‌ها و تغییر رفتار برنامه بر اساس آن‌ها نیاز داریم. **دستورهای شرطی** (conditional statements) این توانایی را به ما می‌دهند. ساده‌ترین شکل آن دستور <code dir="ltr">if</code> است:

</div>

```python
if x > 0:
    print('x is positive')
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">if</code> یک کلمه‌ی کلیدی پایتون است. دستورهای <code dir="ltr">if</code> ساختاری مثل تعریف تابع دارند: یک سرآیند و بعد از آن یک دستور یا دنباله‌ای از دستورها با تورفتگی که **بلوک** (block) نامیده می‌شود.

عبارت بولی بعد از <code dir="ltr">if</code> **شرط** (condition) نامیده می‌شود. اگر درست باشد، دستورهای بلوک با تورفتگی اجرا می‌شوند. اگر نه، اجرا نمی‌شوند.

محدودیتی برای تعداد دستورهایی که در بلوک می‌آیند وجود ندارد، اما حداقل باید یکی باشد. گاهی مفید است بلوکی داشته باشیم که هیچ کاری نمی‌کند – معمولاً به‌عنوان جانگهدار برای کدی که هنوز ننوشته‌اید. در آن صورت می‌توانید از دستور <code dir="ltr">pass</code> استفاده کنید که هیچ کاری نمی‌کند.

</div>

```python
if x < 0:
    pass          # TODO: need to handle negative values!
```

## ۵.۵. بند else

یک دستور <code dir="ltr">if</code> می‌تواند بخش دومی هم داشته باشد که **بند else** نامیده می‌شود. نحو آن این‌طور است:

</div>

```python
if x % 2 == 0:
    print('x is even')
else:
    print('x is odd')
```

<div dir="rtl" style="text-align: right;">

اگر شرط درست باشد، دستور اول با تورفتگی اجرا می‌شود؛ در غیر این صورت دستور دوم اجرا می‌شود.

در این مثال، اگر <code dir="ltr">x</code> زوج باشد، باقی‌مانده‌ی تقسیم <code dir="ltr">x</code> بر <code dir="ltr">2</code> برابر <code dir="ltr">0</code> است، پس شرط درست است و برنامه <code dir="ltr">x is even</code> را نمایش می‌دهد. اگر <code dir="ltr">x</code> فرد باشد، باقی‌مانده <code dir="ltr">1</code> است، پس شرط نادرست است و برنامه <code dir="ltr">x is odd</code> را نمایش می‌دهد.

چون شرط باید درست یا نادرست باشد، دقیقاً یکی از گزینه‌ها اجرا می‌شود. به این گزینه‌ها **شاخه** (branch) می‌گویند.

</div>

<div dir="rtl" style="text-align: right;">

## ۵.۶. شرط‌های زنجیره‌ای

گاهی بیش از دو امکان وجود دارد و به بیش از دو شاخه نیاز داریم. یک راه برای بیان چنین محاسبه‌ای **شرط زنجیره‌ای** (chained conditional) است که شامل بند <code dir="ltr">elif</code> می‌شود.

</div>

```python
if x < y:
    print('x is less than y')
elif x > y:
    print('x is greater than y')
else:
    print('x and y are equal')
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">elif</code> مخفف «else if» است. محدودیتی برای تعداد بندهای <code dir="ltr">elif</code> وجود ندارد. اگر بند <code dir="ltr">else</code> وجود داشته باشد، باید در انتها باشد، اما لازم نیست حتماً وجود داشته باشد.

هر شرط به ترتیب بررسی می‌شود. اگر اولی نادرست باشد، بعدی بررسی می‌شود و همین‌طور تا آخر. اگر یکی از آن‌ها درست باشد، شاخه‌ی متناظر اجرا می‌شود و دستور <code dir="ltr">if</code> تمام می‌شود. حتی اگر بیش از یک شرط درست باشد، فقط اولین شاخه‌ی درست اجرا می‌شود.

</div>

<div dir="rtl" style="text-align: right;">

## ۵.۷. شرط‌های تودرتو

یک دستور شرطی را می‌توان داخل دیگری هم گذاشت. می‌توانستیم مثال بخش قبل را این‌طور بنویسیم:

</div>

```python
if x == y:
    print('x and y are equal')
else:
    if x < y:
        print('x is less than y')
    else:
        print('x is greater than y')
```

<div dir="rtl" style="text-align: right;">

دستور <code dir="ltr">if</code> بیرونی دو شاخه دارد. شاخه‌ی اول یک دستور ساده دارد. شاخه‌ی دوم دستور <code dir="ltr">if</code> دیگری دارد که خودش دو شاخه دارد. هر دو آن شاخه‌ها دستورهای ساده‌اند، هرچند می‌توانستند دستورهای شرطی هم باشند.

هرچند تورفتگی دستورها ساختار را آشکار می‌کند، **شرط‌های تودرتو** (nested conditionals) می‌توانند خواندنشان سخت باشد. پیشنهاد می‌کنم هر وقت می‌توانید از آن‌ها دوری کنید.

عملگرهای منطقی اغلب راهی برای ساده کردن دستورهای شرطی تودرتو فراهم می‌کنند. این مثالی با یک شرط تودرتوست.

</div>

```python
if 0 < x:
    if x < 10:
        print('x is a positive single-digit number.')
```

<div dir="rtl" style="text-align: right;">

دستور <code dir="ltr">print</code> فقط وقتی اجرا می‌شود که از هر دو شرط بگذریم، پس با عملگر <code dir="ltr">and</code> همان اثر را به دست می‌آوریم.

</div>

```python
if 0 < x and x < 10:
    print('x is a positive single-digit number.')
```

<div dir="rtl" style="text-align: right;">

برای این نوع شرط، پایتون گزینه‌ی مختصرتری فراهم می‌کند:

</div>

```python
if 0 < x < 10:
    print('x is a positive single-digit number.')
```

<div dir="rtl" style="text-align: right;">

(با <code dir="ltr">x = 5</code> هر سه نسخه‌ی بالا عبارت <code dir="ltr">x is a positive single-digit number.</code> را چاپ می‌کنند.)

</div>

## ۵.۸. بازگشت

برای یک تابع مجاز است خودش را فراخوانی کند. شاید روشن نباشد که چرا این چیز خوبی است، اما معلوم می‌شود از جادویی‌ترین کارهایی است که یک برنامه می‌تواند بکند. این یک مثال است.

</div>

```python
def countdown(n):
    if n <= 0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n-1)
```

<div dir="rtl" style="text-align: right;">

اگر <code dir="ltr">n</code> برابر ۰ یا منفی باشد، <code dir="ltr">countdown</code> کلمه‌ی «Blastoff!» را چاپ می‌کند. در غیر این صورت <code dir="ltr">n</code> را چاپ می‌کند و بعد خودش را فراخوانی می‌کند و <code dir="ltr">n-1</code> را به‌عنوان آرگومان می‌دهد.

این اتفاقی است که وقتی این تابع را با آرگومان <code dir="ltr">3</code> فراخوانی می‌کنیم می‌افتد.

</div>

```python
countdown(3)
```

<div dir="rtl" style="text-align: right;">

اجرای <code dir="ltr">countdown</code> با <code dir="ltr">n=3</code> شروع می‌شود، و چون <code dir="ltr">n</code> بزرگ‌تر از <code dir="ltr">0</code> است، <code dir="ltr">3</code> را نمایش می‌دهد و بعد خودش را فراخوانی می‌کند…

> اجرای <code dir="ltr">countdown</code> با <code dir="ltr">n=2</code> شروع می‌شود، و چون <code dir="ltr">n</code> بزرگ‌تر از <code dir="ltr">0</code> است، <code dir="ltr">2</code> را نمایش می‌دهد و بعد خودش را فراخوانی می‌کند…
>
> > اجرای <code dir="ltr">countdown</code> با <code dir="ltr">n=1</code> شروع می‌شود، و چون <code dir="ltr">n</code> بزرگ‌تر از <code dir="ltr">0</code> است، <code dir="ltr">1</code> را نمایش می‌دهد و بعد خودش را فراخوانی می‌کند…
> >
> > > اجرای <code dir="ltr">countdown</code> با <code dir="ltr">n=0</code> شروع می‌شود، و چون <code dir="ltr">n</code> بزرگ‌تر از <code dir="ltr">0</code> نیست، «Blastoff!» را نمایش می‌دهد و برمی‌گردد.
> >
> > <code dir="ltr">countdown</code>ای که <code dir="ltr">n=1</code> گرفته بود برمی‌گردد.
>
> <code dir="ltr">countdown</code>ای که <code dir="ltr">n=2</code> گرفته بود برمی‌گردد.

<code dir="ltr">countdown</code>ای که <code dir="ltr">n=3</code> گرفته بود برمی‌گردد.

تابعی که خودش را فراخوانی می‌کند **بازگشتی** (recursive) است. به‌عنوان مثالی دیگر، می‌توانیم تابعی بنویسیم که یک رشته را <code dir="ltr">n</code> بار چاپ می‌کند.

</div>

```python
def print_n_times(string, n):
    if n > 0:
        print(string)
        print_n_times(string, n-1)
```

<div dir="rtl" style="text-align: right;">

اگر <code dir="ltr">n</code> مثبت باشد، <code dir="ltr">print_n_times</code> مقدار <code dir="ltr">string</code> را نمایش می‌دهد و بعد خودش را فراخوانی می‌کند و <code dir="ltr">string</code> و <code dir="ltr">n-1</code> را به‌عنوان آرگومان می‌دهد.

اگر <code dir="ltr">n</code> برابر <code dir="ltr">0</code> یا منفی باشد، شرط نادرست است و <code dir="ltr">print_n_times</code> هیچ کاری نمی‌کند.

## ۵.۱۳. واژه‌نامه

**بازگشت (recursion):** فرایند فراخوانی تابعی که در حال حاضر در حال اجراست.

**عملگر باقی‌مانده (modulus operator):** عملگری، <code dir="ltr">%</code>، که روی اعداد صحیح کار می‌کند و باقی‌مانده‌ی تقسیم یک عدد بر عدد دیگر را برمی‌گرداند.

**عبارت بولی (boolean expression):** عبارتی که مقدارش <code dir="ltr">True</code> یا <code dir="ltr">False</code> است.

**عملگر رابطه‌ای (relational operator):** یکی از عملگرهایی که عملوندهایش را مقایسه می‌کند: <code dir="ltr">==</code>، <code dir="ltr">!=</code>، <code dir="ltr">&gt;</code>، <code dir="ltr">&lt;</code>، <code dir="ltr">&gt;=</code> و <code dir="ltr">&lt;=</code>.

**عملگر منطقی (logical operator):** یکی از عملگرهایی که عبارت‌های بولی را ترکیب می‌کند، از جمله <code dir="ltr">and</code>، <code dir="ltr">or</code> و <code dir="ltr">not</code>.

**دستور شرطی (conditional statement):** دستوری که جریان اجرا را بسته به یک شرط کنترل می‌کند.

**شرط (condition):** عبارت بولی در یک دستور شرطی که تعیین می‌کند کدام شاخه اجرا شود.

**بلوک (block):** یک یا چند دستور با تورفتگی که نشان می‌دهد بخشی از دستور دیگری هستند.

**شاخه (branch):** یکی از دنباله‌های جایگزین دستورها در یک دستور شرطی.

**شرط زنجیره‌ای (chained conditional):** دستور شرطی‌ای با مجموعه‌ای از شاخه‌های جایگزین.

**شرط تودرتو (nested conditional):** دستور شرطی‌ای که در یکی از شاخه‌های دستور شرطی دیگری می‌آید.

**بازگشتی (recursive):** تابعی که خودش را فراخوانی می‌کند بازگشتی است.

**حالت پایه (base case):** شاخه‌ای شرطی در یک تابع بازگشتی که فراخوانی بازگشتی انجام نمی‌دهد.

**بازگشت بی‌نهایت (infinite recursion):** بازگشتی که حالت پایه ندارد یا هرگز به آن نمی‌رسد. بازگشت بی‌نهایت در نهایت باعث خطای زمان اجرا می‌شود.

**خط جدید (newline):** نویسه‌ای که بین دو بخش یک رشته شکست خط ایجاد می‌کند.
