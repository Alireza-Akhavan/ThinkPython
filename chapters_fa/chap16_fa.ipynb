## ۱۶.۱. ساختن یک نقطه (Point)

در گرافیک کامپیوتری، یک مکان روی صفحه اغلب با یک جفت مختصات در صفحه‌ی <code dir="ltr">x</code>-<code dir="ltr">y</code> نشان داده می‌شود. طبق قرارداد، نقطه‌ی <code dir="ltr">(0, 0)</code> معمولاً گوشه‌ی بالا-چپ صفحه است و <code dir="ltr">(x, y)</code> نقطه‌ای است که <code dir="ltr">x</code> واحد به راست و <code dir="ltr">y</code> واحد پایین‌تر از مبدأ است. نسبت به دستگاه مختصات دکارتی که ممکن است در کلاس ریاضی دیده باشید، محور <code dir="ltr">y</code> وارونه است.

چند راه برای نمایش یک نقطه در پایتون وجود دارد:

-   می‌توانیم مختصات را جداگانه در دو متغیر <code dir="ltr">x</code> و <code dir="ltr">y</code> ذخیره کنیم.

-   می‌توانیم مختصات را به‌عنوان عناصر یک فهرست یا چندتایی ذخیره کنیم.

-   می‌توانیم یک نوع جدید بسازیم تا نقطه‌ها را به‌صورت شیء نشان دهد.

در برنامه‌نویسی شیءگرا، ساختن نوع جدید متداول‌ترین راه است. برای این کار با تعریف کلاس <code dir="ltr">Point</code> شروع می‌کنیم.

</div>

```python
class Point:
    """Represents a point in 2-D space."""
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f'Point({self.x}, {self.y})'
```

<div dir="rtl" style="text-align: right;">

متد <code dir="ltr">__init__</code> مختصات را به‌عنوان پارامتر می‌گیرد و به صفت‌های <code dir="ltr">x</code> و <code dir="ltr">y</code> نسبت می‌دهد. متد <code dir="ltr">__str__</code> یک نمایش رشته‌ای از <code dir="ltr">Point</code> برمی‌گرداند.

حالا می‌توانیم یک شیء <code dir="ltr">Point</code> را این‌طور نمونه‌سازی و نمایش دهیم.

</div>

```python
start = Point(0, 0)
print(start)
```

<div dir="rtl" style="text-align: right;">

*(در کتاب اصلی اینجا نمودار حالت شیء رسم شده است: جعبه‌ای با برچسب <code dir="ltr">Point</code> که دو صفت <code dir="ltr">x</code> و <code dir="ltr">y</code> با مقدار ۰ دارد و متغیر <code dir="ltr">start</code> به آن اشاره می‌کند.)*

مثل همیشه، یک نوع تعریف‌شده توسط برنامه‌نویس با جعبه‌ای نشان داده می‌شود که نام نوع بیرون و صفت‌ها داخل آن‌اند.

به‌طور کلی، نوع‌های تعریف‌شده توسط برنامه‌نویس تغییرپذیرند، پس می‌توانیم متدی مثل <code dir="ltr">translate</code> بنویسیم که دو عدد <code dir="ltr">dx</code> و <code dir="ltr">dy</code> را می‌گیرد و به صفت‌های <code dir="ltr">x</code> و <code dir="ltr">y</code> اضافه می‌کند.

</div>

```python
%%add_method_to Point

    def translate(self, dx, dy):
        self.x += dx
        self.y += dy
```

<div dir="rtl" style="text-align: right;">

این تابع <code dir="ltr">Point</code> را از یک مکان در صفحه به مکان دیگر می‌برد (انتقال می‌دهد). اگر نخواهیم <code dir="ltr">Point</code> موجود را تغییر دهیم، می‌توانیم از <code dir="ltr">copy</code> برای کپی کردن شیء اصلی و بعد تغییر دادن کپی استفاده کنیم.

</div>

```python
from copy import copy

end1 = copy(start)
end1.translate(300, 0)
print(end1)
```

<div dir="rtl" style="text-align: right;">

می‌توانیم این مراحل را در متد دیگری به نام <code dir="ltr">translated</code> کپسوله کنیم.

</div>

```python
%%add_method_to Point

    def translated(self, dx=0, dy=0):
        point = copy(self)
        point.translate(dx, dy)
        return point
```

<div dir="rtl" style="text-align: right;">

همان‌طور که متد <code dir="ltr">sort</code> یک فهرست را تغییر می‌دهد و تابع <code dir="ltr">sorted</code> فهرست جدیدی می‌سازد، حالا هم متد <code dir="ltr">translate</code> را داریم که یک <code dir="ltr">Point</code> را تغییر می‌دهد و متد <code dir="ltr">translated</code> که یکی جدید می‌سازد.

این یک مثال است:

</div>

```python
end2 = start.translated(0, 150)
print(end2)
```

<div dir="rtl" style="text-align: right;">

در بخش بعد، از این نقطه‌ها برای تعریف و رسم یک خط استفاده می‌کنیم.

</div>

## ۱۶.۲. ساختن یک خط (Line)

حالا کلاسی تعریف می‌کنیم که پاره‌خط بین دو نقطه را نشان می‌دهد. مثل همیشه با یک متد <code dir="ltr">__init__</code> و یک متد <code dir="ltr">__str__</code> شروع می‌کنیم.

</div>

```python
class Line:
    def __init__(self, p1, p2):
        self.p1 = p1
        self.p2 = p2
        
    def __str__(self):
        return f'Line({self.p1}, {self.p2})'
```

<div dir="rtl" style="text-align: right;">

با این دو متد می‌توانیم یک شیء <code dir="ltr">Line</code> را نمونه‌سازی و نمایش دهیم که برای نمایش محور <code dir="ltr">x</code> به کار می‌بریم.

</div>

```python
line1 = Line(start, end1)
print(line1)
```

<div dir="rtl" style="text-align: right;">

وقتی <code dir="ltr">print</code> را فراخوانی می‌کنیم و <code dir="ltr">line</code> را به‌عنوان پارامتر می‌دهیم، <code dir="ltr">print</code> متد <code dir="ltr">__str__</code> را روی <code dir="ltr">line</code> احضار می‌کند. متد <code dir="ltr">__str__</code> از یک f-string برای ساختن نمایش رشته‌ای <code dir="ltr">line</code> استفاده می‌کند.

f-string دو عبارت داخل آکولاد دارد، <code dir="ltr">self.p1</code> و <code dir="ltr">self.p2</code>. وقتی این عبارت‌ها ارزیابی می‌شوند، نتیجه شیءهای <code dir="ltr">Point</code> است. بعد، وقتی به رشته تبدیل می‌شوند، متد <code dir="ltr">__str__</code> از کلاس <code dir="ltr">Point</code> احضار می‌شود.

به همین دلیل، وقتی یک <code dir="ltr">Line</code> را نمایش می‌دهیم، نتیجه شامل نمایش رشته‌ای شیءهای <code dir="ltr">Point</code> است.

*(نمودار شیء زیر در کتاب اصلی حالت این شیء <code dir="ltr">Line</code> را نشان می‌دهد: جعبه‌ی <code dir="ltr">Line</code> که دو صفت <code dir="ltr">p1</code> و <code dir="ltr">p2</code> دارد، و هر کدام به یک جعبه‌ی <code dir="ltr">Point</code> اشاره می‌کنند؛ اولی <code dir="ltr">x=0, y=0</code> و دومی <code dir="ltr">x=300, y=0</code>.)*

نمایش‌های رشته‌ای و نمودارهای شیء برای اشکال‌زدایی مفیدند، اما هدف این مثال تولید گرافیک است، نه متن! پس از ماژول <code dir="ltr">jupyturtle</code> برای رسم خط‌ها روی صفحه استفاده می‌کنیم.

همان‌طور که در [فصل ۴](https://allendowney.github.io/ThinkPython/chap04.html) کردیم، از <code dir="ltr">make_turtle</code> برای ساختن یک شیء <code dir="ltr">Turtle</code> و یک بوم کوچک استفاده می‌کنیم تا روی آن بکشد. برای کشیدن خط‌ها از دو تابع جدید ماژول <code dir="ltr">jupyturtle</code> استفاده می‌کنیم:

* <code dir="ltr">jumpto</code>، که دو مختصات می‌گیرد و <code dir="ltr">Turtle</code> را بدون کشیدن خط به مکان داده‌شده می‌برد، و

* <code dir="ltr">moveto</code>، که <code dir="ltr">Turtle</code> را از مکان فعلی‌اش به مکان داده‌شده می‌برد و بین آن‌ها یک پاره‌خط می‌کشد.

این‌طور وارد‌شان می‌کنیم.

</div>

## ۱۶.۳. هم‌ارزی و هویت

فرض کنید دو نقطه با مختصات یکسان بسازیم.

</div>

```python
p1 = Point(200, 100)
p2 = Point(200, 100)
```

<div dir="rtl" style="text-align: right;">

اگر از عملگر <code dir="ltr">==</code> برای مقایسه‌شان استفاده کنیم، رفتار پیش‌فرض نوع‌های تعریف‌شده توسط برنامه‌نویس را می‌گیریم – نتیجه فقط وقتی <code dir="ltr">True</code> است که هر دو یک شیء باشند، که نیستند.

</div>

```python
p1 == p2
```

<div dir="rtl" style="text-align: right;">

اگر بخواهیم این رفتار را تغییر دهیم، می‌توانیم متد ویژه‌ای به نام <code dir="ltr">__eq__</code> فراهم کنیم که تعریف می‌کند برابر بودن دو شیء <code dir="ltr">Point</code> یعنی چه.

```python
%%add_method_to Point

def __eq__(self, other):
    return (self.x == other.x) and (self.y == other.y)
```

<div dir="rtl" style="text-align: right;">

این تعریف دو <code dir="ltr">Point</code> را وقتی برابر می‌داند که صفت‌هایشان برابر باشد. حالا وقتی از عملگر <code dir="ltr">==</code> استفاده می‌کنیم، متد <code dir="ltr">__eq__</code> احضار می‌شود، که نشان می‌دهد <code dir="ltr">p1</code> و <code dir="ltr">p2</code> برابر در نظر گرفته می‌شوند.

</div>

```python
p1 == p2
```

<div dir="rtl" style="text-align: right;">

اما عملگر <code dir="ltr">is</code> هنوز نشان می‌دهد که شیءهای متفاوتی هستند.

</div>

```python
p1 is p2
```

<div dir="rtl" style="text-align: right;">

بازنویسی عملگر <code dir="ltr">is</code> ممکن نیست – این عملگر همیشه بررسی می‌کند که شیءها یکسان (identical) هستند یا نه. اما برای نوع‌های تعریف‌شده توسط برنامه‌نویس، می‌توانید عملگر <code dir="ltr">==</code> را بازنویسی کنید تا بررسی کند که شیءها هم‌ارز (equivalent) هستند یا نه. و می‌توانید تعریف کنید که هم‌ارز یعنی چه.

## ۱۶.۴. ساختن یک مستطیل (Rectangle)

حالا کلاسی تعریف می‌کنیم که مستطیل‌ها را نشان می‌دهد و می‌کشد. برای ساده ماندن، فرض می‌کنیم مستطیل‌ها یا عمودی‌اند یا افقی، نه کج. به نظرتان چه صفت‌هایی را برای مشخص کردن مکان و اندازه‌ی یک مستطیل باید به کار ببریم؟

دست‌کم دو امکان وجود دارد:

-   می‌توانید عرض و ارتفاع مستطیل و مکان یک گوشه را مشخص کنید.

-   می‌توانید دو گوشه‌ی مقابل را مشخص کنید.

در این مرحله سخت است بگوییم کدام بهتر است، پس اولی را پیاده‌سازی می‌کنیم. این هم تعریف کلاس.

</div>

```python
class Rectangle:
    """Represents a rectangle. 

    attributes: width, height, corner.
    """
    def __init__(self, width, height, corner):
        self.width = width
        self.height = height
        self.corner = corner
        
    def __str__(self):
        return f'Rectangle({self.width}, {self.height}, {self.corner})'
```

<div dir="rtl" style="text-align: right;">

مثل همیشه، متد <code dir="ltr">__init__</code> پارامترها را به صفت‌ها نسبت می‌دهد و <code dir="ltr">__str__</code> یک نمایش رشته‌ای از شیء برمی‌گرداند. حالا می‌توانیم یک شیء <code dir="ltr">Rectangle</code> نمونه‌سازی کنیم و یک <code dir="ltr">Point</code> را به‌عنوان مکان گوشه‌ی بالا-چپ به کار ببریم.

</div>

```python
corner = Point(30, 20)
box1 = Rectangle(100, 50, corner)
print(box1)
```

<div dir="rtl" style="text-align: right;">

*(نمودار کتاب حالت این شیء را نشان می‌دهد: جعبه‌ی <code dir="ltr">Rectangle</code> با صفت‌های <code dir="ltr">width=100</code> و <code dir="ltr">height=50</code>، و صفت <code dir="ltr">corner</code> که به یک جعبه‌ی <code dir="ltr">Point</code> با <code dir="ltr">x=30</code> و <code dir="ltr">y=20</code> اشاره می‌کند.)*

برای کشیدن یک مستطیل، از متد زیر استفاده می‌کنیم که چهار شیء <code dir="ltr">Point</code> برای گوشه‌ها می‌سازد.

</div>

```python
%%add_method_to Rectangle

    def make_points(self):
        p1 = self.corner
        p2 = p1.translated(self.width, 0)
        p3 = p2.translated(0, self.height)
        p4 = p3.translated(-self.width, 0)
        return p1, p2, p3, p4
```

<div dir="rtl" style="text-align: right;">

بعد چهار شیء <code dir="ltr">Line</code> برای ضلع‌ها می‌سازیم.

</div>

```python
%%add_method_to Rectangle

    def make_lines(self):
        p1, p2, p3, p4 = self.make_points()
        return Line(p1, p2), Line(p2, p3), Line(p3, p4), Line(p4, p1)
```

<div dir="rtl" style="text-align: right;">

و بعد ضلع‌ها را می‌کشیم.

</div>

```python
%%add_method_to Rectangle

    def draw(self):
        lines = self.make_lines()
        for line in lines:
            line.draw()
```

این یک مثال است.

</div>

## ۱۶.۶. کپی عمیق (Deep copy)

وقتی از <code dir="ltr">copy</code> برای تکثیر <code dir="ltr">box1</code> استفاده می‌کنیم، شیء <code dir="ltr">Rectangle</code> را کپی می‌کند اما شیء <code dir="ltr">Point</code> داخل آن را نه. پس <code dir="ltr">box1</code> و <code dir="ltr">box2</code> شیءهای متفاوتی هستند، همان‌طور که می‌خواستیم.

</div>

```python
box1 is box2
```

<div dir="rtl" style="text-align: right;">

اما صفت‌های <code dir="ltr">corner</code> آن‌ها به یک شیء اشاره می‌کنند.

```python
box1.corner is box2.corner
```

<div dir="rtl" style="text-align: right;">

*(نمودار کتاب حالت این شیءها را نشان می‌دهد: دو جعبه‌ی <code dir="ltr">Rectangle</code> مجزا، هر کدام با <code dir="ltr">width</code> و <code dir="ltr">height</code> خودش، اما صفت <code dir="ltr">corner</code> هر دو به همان یک جعبه‌ی <code dir="ltr">Point</code> اشاره می‌کند.)*

کاری که <code dir="ltr">copy</code> می‌کند **کپی سطحی** (shallow copy) نامیده می‌شود چون شیء را کپی می‌کند اما شیءهایی را که در بر دارد نه. در نتیجه، تغییر <code dir="ltr">width</code> یا <code dir="ltr">height</code> یک <code dir="ltr">Rectangle</code> روی دیگری اثر نمی‌گذارد، اما تغییر صفت‌های <code dir="ltr">Point</code> مشترک روی هر دو اثر می‌گذارد! این رفتار گیج‌کننده و مستعد خطا است.

خوشبختانه ماژول <code dir="ltr">copy</code> تابع دیگری به نام <code dir="ltr">deepcopy</code> دارد که نه‌تنها شیء، بلکه شیءهایی را که به آن‌ها اشاره می‌کند، و شیءهایی را که *آن‌ها* به آن‌ها اشاره می‌کنند و همین‌طور الی آخر، کپی می‌کند. این عملیات **کپی عمیق** (deep copy) نامیده می‌شود.

برای نمایش، با یک <code dir="ltr">Rectangle</code> تازه شروع می‌کنیم که یک <code dir="ltr">Point</code> تازه دارد.

## ۱۶.۷. چندریختی (Polymorphism)

در مثال قبل، متد <code dir="ltr">draw</code> را روی دو شیء <code dir="ltr">Line</code> و دو شیء <code dir="ltr">Rectangle</code> احضار کردیم. می‌توانیم همین کار را با ساختن فهرستی از شیءها مختصرتر انجام دهیم.

</div>

```python
shapes = [line1, line2, box3, box4]
```

<div dir="rtl" style="text-align: right;">

عناصر این فهرست از نوع‌های متفاوتی‌اند، اما همه متد <code dir="ltr">draw</code> دارند، پس می‌توانیم فهرست را پیمایش و روی هر یک <code dir="ltr">draw</code> را احضار کنیم.

<div dir="rtl" style="text-align: right;">

بار اول و دوم که حلقه اجرا می‌شود، <code dir="ltr">shape</code> به یک شیء <code dir="ltr">Line</code> اشاره می‌کند، پس وقتی <code dir="ltr">draw</code> احضار می‌شود، متدی که اجرا می‌شود همانی است که در کلاس <code dir="ltr">Line</code> تعریف شده است.

بار سوم و چهارم، <code dir="ltr">shape</code> به یک شیء <code dir="ltr">Rectangle</code> اشاره می‌کند، پس وقتی <code dir="ltr">draw</code> احضار می‌شود، متدی که اجرا می‌شود همانی است که در کلاس <code dir="ltr">Rectangle</code> تعریف شده است.

به یک معنا، هر شیء می‌داند چطور خودش را بکشد. این ویژگی **چندریختی** (polymorphism) نامیده می‌شود. این واژه از ریشه‌های یونانی‌ای می‌آید که به معنای «دارای شکل‌های بسیار» است. در برنامه‌نویسی شیءگرا، چندریختی توانایی نوع‌های متفاوت برای ارائه‌ی متدهای یکسان است، که انجام محاسبه‌های بسیار – مثل کشیدن شکل‌ها – را با احضار یک متد یکسان روی نوع‌های متفاوتی از شیءها ممکن می‌کند.

در تمرینی در پایان این فصل، کلاس جدیدی تعریف می‌کنید که یک دایره را نشان می‌دهد و متد <code dir="ltr">draw</code> دارد. بعد می‌توانید با چندریختی خط‌ها، مستطیل‌ها و دایره‌ها را بکشید.

</div>

## ۱۶.۸. اشکال‌زدایی

در این فصل با باگ ظریفی روبه‌رو شدیم که چون یک <code dir="ltr">Point</code> ساختیم که بین دو شیء <code dir="ltr">Rectangle</code> مشترک بود و بعد <code dir="ltr">Point</code> را تغییر دادیم رخ داد. به‌طور کلی دو راه برای پرهیز از مشکل‌هایی مثل این وجود دارد: می‌توانید از به اشتراک گذاشتن شیءها پرهیز کنید، یا از تغییر دادن آن‌ها.

برای پرهیز از اشتراک شیءها، می‌توانید از کپی عمیق استفاده کنید، همان‌طور که در این فصل کردیم.

برای پرهیز از تغییر شیءها، می‌توانید به جای توابع ناخالص مثل <code dir="ltr">translate</code> از توابع خالص مثل <code dir="ltr">translated</code> استفاده کنید. برای مثال، این نسخه‌ای از <code dir="ltr">translated</code> است که یک <code dir="ltr">Point</code> جدید می‌سازد و هرگز صفت‌هایش را تغییر نمی‌دهد.

</div>

```python
    def translated(self, dx=0, dy=0):
        x = self.x + dx
        y = self.y + dy
        return Point(x, y)
```

<div dir="rtl" style="text-align: right;">

پایتون امکاناتی دارد که پرهیز از تغییر دادن شیءها را آسان‌تر می‌کند. این امکانات خارج از حوصله‌ی این کتاب‌اند، اما اگر کنجکاو هستید، از یک دستیار مجازی بپرسید: "How do I make a Python object immutable?" (چطور یک شیء پایتون را تغییرناپذیر کنم؟)

ساختن شیء جدید زمان بیشتری از تغییر شیء موجود می‌برد، اما این تفاوت به‌ندرت در عمل اهمیت دارد. برنامه‌هایی که از شیءهای مشترک و توابع ناخالص پرهیز می‌کنند اغلب توسعه، آزمون و اشکال‌زدایی‌شان آسان‌تر است – و بهترین نوع اشکال‌زدایی همان است که مجبور نباشید انجامش دهید.

</div>

## ۱۶.۹. واژه‌نامه

**کپی سطحی (shallow copy):** عملیات کپی که شیءهای تودرتو را کپی نمی‌کند.

**کپی عمیق (deep copy):** عملیات کپی که شیءهای تودرتو را هم کپی می‌کند.

**چندریختی (polymorphism):** توانایی یک متد یا عملگر برای کار با چند نوع شیء.

می‌توانید از تابع زیر استفاده کنید، که نسخه‌ای از تابع <code dir="ltr">circle</code> است که در فصل ۴ نوشتیم.

</div>

```python
from jupyturtle import make_turtle, forward, left, right
import math
    
def draw_circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    angle = 360 / n
    left(angle / 2)
    for i in range(n):
        forward(length)
        left(angle)
```
