</div>

````python
n = 17
````

<div dir="rtl" style="text-align: right;">

</div>

<div dir="rtl" style="text-align: right;">

</div>

````python
message = 'And now for something completely different'
````

## ۲.۲. نمودارهای حالت

یک راه رایج برای نمایش متغیرها روی کاغذ این است که نام متغیر را بنویسیم و یک فلش از آن به مقدارش بکشیم.

(در کتاب اصلی این‌جا شکلی رسم شده است: سه متغیر <code dir="ltr">n</code> با مقدار ۱۷، <code dir="ltr">pi</code> با مقدار ۳.۱۴۱۵۹… و <code dir="ltr">message</code> با مقدار رشته‌ی بالا، که هرکدام با یک فلش به مقدارشان وصل شده‌اند.)

به این نوع شکل **نمودار حالت** (state diagram) می‌گویند، چون نشان می‌دهد هر متغیر در چه «حالتی» است (آن را «حالت ذهنی» متغیر در نظر بگیرید). در سراسر کتاب از نمودارهای حالت استفاده می‌کنیم تا مدلی از چگونگی ذخیره شدن متغیرها و مقدارهایشان در پایتون را نشان دهیم.

</div>

## ۲.۳. نام متغیرها

نام متغیرها می‌تواند هر قدر که دوست دارید بلند باشد. می‌توانند هم حرف داشته باشند و هم عدد، اما نمی‌توانند با عدد شروع شوند. استفاده از حروف بزرگ مجاز است، اما قرارداد این است که برای نام متغیرها فقط از حروف کوچک استفاده شود.

تنها علامت نگارشی‌ای که در نام متغیر می‌تواند بیاید زیرخط (underscore)، <code dir="ltr">_</code>، است. این علامت اغلب در نام‌هایی که چند کلمه دارند به کار می‌رود، مثل <code dir="ltr">your_name</code> یا <code dir="ltr">airspeed_of_unladen_swallow</code>.

اگر به یک متغیر نام غیرمجاز بدهید، خطای نحوی می‌گیرید. نام <code dir="ltr">million!</code> غیرمجاز است چون علامت نگارشی دارد.

</div>

````python
%%expect SyntaxError

million! = 1000000
````

<div dir="rtl" style="text-align: right;">

<code dir="ltr">76trombones</code> غیرمجاز است چون با عدد شروع می‌شود.

</div>

````python
%%expect SyntaxError

76trombones = 'big parade'
````

<div dir="rtl" style="text-align: right;">

<code dir="ltr">class</code> هم غیرمجاز است، اما شاید دلیلش بدیهی نباشد.

</div>

````python
%%expect SyntaxError

class = 'Self-Defence Against Fresh Fruit'
````

<div dir="rtl" style="text-align: right;">

معلوم می‌شود که <code dir="ltr">class</code> یک **کلمه‌ی کلیدی** (keyword) است؛ یعنی کلمه‌ای خاص که برای مشخص کردن ساختار برنامه به کار می‌رود. کلمه‌های کلیدی را نمی‌توان به‌عنوان نام متغیر به کار برد.

فهرست کامل کلمه‌های کلیدی پایتون این است:


````python
from keyword import kwlist

len(kwlist)
````

<div dir="rtl" style="text-align: right;">

لازم نیست این فهرست را حفظ کنید. در بیشتر محیط‌های توسعه، کلمه‌های کلیدی با رنگ متفاوتی نمایش داده می‌شوند؛ اگر سعی کنید از یکی‌شان به‌عنوان نام متغیر استفاده کنید، متوجه می‌شوید.

</div>

## ۲.۴. دستور import

برای استفاده از بعضی قابلیت‌های پایتون باید آن‌ها را **وارد** (import) کنید. برای مثال، دستور زیر ماژول <code dir="ltr">math</code> را وارد می‌کند.

</div>

````python
import math
````

<div dir="rtl" style="text-align: right;">

**ماژول** (module) مجموعه‌ای از متغیرها و توابع است. ماژول math متغیری به نام <code dir="ltr">pi</code> دارد که مقدار ثابت ریاضی π را در خود دارد. می‌توانیم مقدارش را این‌طور نمایش دهیم.

</div>

````python
math.pi
````

<div dir="rtl" style="text-align: right;">

برای استفاده از یک متغیر در یک ماژول، باید بین نام ماژول و نام متغیر از **عملگر نقطه** (dot operator)، <code dir="ltr">.</code>، استفاده کنید.

ماژول math توابعی هم دارد. برای مثال، <code dir="ltr">sqrt</code> ریشه‌ی دوم را حساب می‌کند.

</div>

````python
math.sqrt(25)
````

<div dir="rtl" style="text-align: right;">

و <code dir="ltr">pow</code> عدد اول را به توان عدد دوم می‌رساند.

</div>

````python
math.pow(5, 2)
````

<div dir="rtl" style="text-align: right;">

تا اینجا دو راه برای به توان رساندن یک عدد دیده‌ایم: می‌توانیم از تابع <code dir="ltr">math.pow</code> یا از عملگر توان‌رسانی، <code dir="ltr">**</code>، استفاده کنیم. هر دو خوب‌اند، اما عملگر بیشتر به کار می‌رود.

</div>

</div>

هرچه برنامه‌ها بزرگ‌تر و پیچیده‌تر شوند، خواندنشان دشوارتر می‌شود. زبان‌های صوری چگال‌اند و اغلب سخت است به یک تکه کد نگاه کنیم و بفهمیم چه می‌کند و چرا.

به همین دلیل، ایده‌ی خوبی است که به برنامه‌هایتان یادداشت‌هایی اضافه کنید تا با زبان طبیعی توضیح بدهند برنامه چه می‌کند. به این یادداشت‌ها **توضیح** یا **کامنت** (comment) می‌گویند و با نماد <code dir="ltr">#</code> شروع می‌شوند.

</div>

````python
# number of seconds in 42:42
seconds = 42 * 60 + 42
````

<div dir="rtl" style="text-align: right;">

</div>

````python
miles = 10 / 1.61     # 10 kilometers in miles
````

<div dir="rtl" style="text-align: right;">

کامنت‌ها وقتی بیشترین فایده را دارند که ویژگی‌های غیرواضح کد را مستند کنند. منطقی است فرض کنیم خواننده می‌تواند بفهمد کد *چه* می‌کند؛ مفیدتر این است که توضیح بدهیم *چرا*.

این کامنت با خود کد افزونه است و بی‌فایده:

</div>

````python
v = 8     # assign 8 to v
````

</div>

## ۲.۹. اشکال‌زدایی

در یک برنامه سه نوع خطا ممکن است رخ دهد: خطای نحوی، خطای زمان اجرا و خطای معنایی. تشخیص فرق آن‌ها مفید است تا سریع‌تر ردیابی‌شان کنید.

* **خطای نحوی** (syntax error): «نحو» به ساختار برنامه و قواعد مربوط به آن ساختار اشاره دارد. اگر در هر جای برنامه‌ی شما خطای نحوی باشد، پایتون برنامه را اجرا نمی‌کند. بلافاصله پیام خطا نمایش می‌دهد.

* **خطای زمان اجرا** (runtime error): اگر برنامه‌ی شما خطای نحوی نداشته باشد، می‌تواند شروع به اجرا کند. اما اگر چیزی اشتباه پیش برود، پایتون پیام خطا نمایش می‌دهد و متوقف می‌شود. به این نوع خطا خطای زمان اجرا می‌گویند. به آن **استثنا** (exception) هم می‌گویند، چون نشان می‌دهد اتفاقی استثنایی رخ داده است.

* **خطای معنایی** (semantic error): نوع سوم خطا «معنایی» است؛ یعنی مربوط به معنا. اگر برنامه‌ی شما خطای معنایی داشته باشد، بدون تولید هیچ پیام خطایی اجرا می‌شود، اما کاری را که قصدتان بود انجام نمی‌دهد. شناسایی خطاهای معنایی می‌تواند حیله‌گرانه باشد، چون لازم است به‌صورت معکوس کار کنید: به خروجی برنامه نگاه کنید و سعی کنید بفهمید برنامه چه می‌کند.

همان‌طور که دیدیم، نام غیرمجاز برای متغیر یک خطای نحوی است.

</div>

````python
%%expect SyntaxError

million! = 1000000
````

## ۲.۱۰. واژه‌نامه

**متغیر (variable):** نامی که به یک مقدار اشاره می‌کند.

**دستور انتساب (assignment statement):** دستوری که یک مقدار را به یک متغیر نسبت می‌دهد.

**نمودار حالت (state diagram):** نمایشی گرافیکی از مجموعه‌ای از متغیرها و مقدارهایی که به آن‌ها اشاره می‌کنند.

**کلمه‌ی کلیدی (keyword):** کلمه‌ای خاص که برای مشخص کردن ساختار برنامه به کار می‌رود.

**دستور import (import statement):** دستوری که فایل یک ماژول را می‌خواند تا بتوانیم از متغیرها و توابعی که دارد استفاده کنیم.

**ماژول (module):** فایلی که کد پایتون دارد، از جمله تعریف توابع و گاهی دستورهای دیگر.

**عملگر نقطه (dot operator):** عملگر <code dir="ltr">.</code> که برای دسترسی به یک تابع در ماژولی دیگر به کار می‌رود؛ با نوشتن نام ماژول، سپس یک نقطه و بعد نام تابع.

**ارزیابی کردن (evaluate):** انجام عملیات یک عبارت برای محاسبه‌ی یک مقدار.

**دستور (statement):** یک یا چند خط کد که یک فرمان یا عمل را نشان می‌دهند.

**اجرا کردن (execute):** اجرای یک دستور و انجام آنچه می‌گوید.

**آرگومان (argument):** مقداری که هنگام فراخوانی یک تابع به آن داده می‌شود.

**توضیح یا کامنت (comment):** متنی که در برنامه می‌آید و درباره‌ی برنامه اطلاعات می‌دهد اما روی اجرای آن اثری ندارد.

**خطای زمان اجرا (runtime error):** خطایی که باعث می‌شود برنامه پیام خطا نمایش دهد و خارج شود.

**استثنا (exception):** خطایی که هنگام اجرای برنامه شناسایی می‌شود.

**خطای معنایی (semantic error):** خطایی که باعث می‌شود برنامه کار اشتباهی انجام دهد، اما پیام خطایی نمایش ندهد.
