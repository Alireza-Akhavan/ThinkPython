## ۱۴.۱. نوع‌های تعریف‌شده توسط برنامه‌نویس

از نوع‌های داخلی زیادی در پایتون استفاده کرده‌ایم – حالا نوع جدیدی تعریف می‌کنیم. به‌عنوان اولین مثال، نوعی به نام <code dir="ltr">Time</code> می‌سازیم که یک زمان از روز را نشان می‌دهد. نوع تعریف‌شده توسط برنامه‌نویس **کلاس** (class) هم نامیده می‌شود. تعریف کلاس این‌طور به نظر می‌رسد:

</div>

```python
class Time:
    """Represents a time of day."""
```

<div dir="rtl" style="text-align: right;">

سرآیند نشان می‌دهد کلاس جدید <code dir="ltr">Time</code> نام دارد. بدنه یک docstring است که توضیح می‌دهد کلاس برای چیست. تعریف یک کلاس یک **شیء کلاس** (class object) می‌سازد.

شیء کلاس مثل کارخانه‌ای برای ساختن شیءهاست. برای ساختن یک شیء <code dir="ltr">Time</code>، <code dir="ltr">Time</code> را مثل یک تابع فراخوانی می‌کنید.

</div>

```python
lunch = Time()
```

<div dir="rtl" style="text-align: right;">

نتیجه یک شیء جدید است که نوعش <code dir="ltr">__main__.Time</code> است، که <code dir="ltr">__main__</code> نام ماژولی است که <code dir="ltr">Time</code> در آن تعریف شده.

</div>

```python
type(lunch)
```

<div dir="rtl" style="text-align: right;">

وقتی یک شیء را چاپ می‌کنید، پایتون به شما می‌گوید از چه نوعی است و کجای حافظه ذخیره شده (پیشوند <code dir="ltr">0x</code> یعنی عدد بعدی در مبنای شانزده است).

</div>

```python
print(lunch)
```

<div dir="rtl" style="text-align: right;">

</div>

## ۱۴.۲. صفت‌ها (Attributes)

یک شیء می‌تواند متغیرهایی داشته باشد که **صفت** (attribute) نامیده می‌شوند (در انگلیسی تأکید روی هجای اول تلفظ می‌شود، مثل "AT-trib-ute"، نه هجای دوم، مثل "a-TRIB-ute"). می‌توانیم صفت‌ها را با نماد نقطه بسازیم.

</div>

```python
lunch.hour = 11
lunch.minute = 59
lunch.second = 1
```

<div dir="rtl" style="text-align: right;">

این مثال صفت‌هایی به نام <code dir="ltr">hour</code>، <code dir="ltr">minute</code> و <code dir="ltr">second</code> می‌سازد که ساعت، دقیقه و ثانیه‌ی زمان <code dir="ltr">11:59:01</code> را دارند، که تا جایی که به من مربوط است، وقت ناهار است.

نمودار زیر وضعیت <code dir="ltr">lunch</code> و صفت‌هایش را بعد از این نسبت‌دهی‌ها نشان می‌دهد.

(در کتاب اصلی اینجا نمودار رسم شده است: متغیر <code dir="ltr">lunch</code> که به یک شیء <code dir="ltr">Time</code> اشاره می‌کند؛ این شیء سه صفت <code dir="ltr">hour</code> با مقدار ۱۱، <code dir="ltr">minute</code> با مقدار ۵۹ و <code dir="ltr">second</code> با مقدار ۱ دارد.)

متغیر <code dir="ltr">lunch</code> به یک شیء <code dir="ltr">Time</code> اشاره می‌کند که سه صفت دارد. هر صفت به یک عدد صحیح اشاره می‌کند. نمودار وضعیتی مثل این – که یک شیء و صفت‌هایش را نشان می‌دهد – **نمودار شیء** (object diagram) نامیده می‌شود.

می‌توانید مقدار یک صفت را با عملگر نقطه بخوانید.

</div>

```python
lunch.hour
```

<div dir="rtl" style="text-align: right;">

می‌توانید یک صفت را به‌عنوان بخشی از هر عبارتی به کار ببرید.

</div>

```python
total_minutes = lunch.hour * 60 + lunch.minute
total_minutes
```

<div dir="rtl" style="text-align: right;">

و می‌توانید عملگر نقطه را در یک عبارت داخل f-string به کار ببرید.

</div>

```python
f'{lunch.hour}:{lunch.minute}:{lunch.second}'
```

<div dir="rtl" style="text-align: right;">

اما دقت کنید که مثال قبل در قالب استاندارد نیست. برای درست کردنش، باید صفت‌های <code dir="ltr">minute</code> و <code dir="ltr">second</code> را با صفر پیشرو چاپ کنیم. می‌توانیم این کار را با گسترش عبارت‌های داخل آکولاد با یک **مشخص‌کننده‌ی قالب** (format specifier) انجام دهیم. در مثال زیر، مشخص‌کننده‌های قالب نشان می‌دهند <code dir="ltr">minute</code> و <code dir="ltr">second</code> باید با دست‌کم دو رقم و در صورت لزوم با یک صفر پیشرو نمایش داده شوند.

</div>

```python
f'{lunch.hour}:{lunch.minute:02d}:{lunch.second:02d}'
```

<div dir="rtl" style="text-align: right;">

از این f-string برای نوشتن تابعی استفاده می‌کنیم که مقدار یک شیء <code dir="ltr">Time</code> را نمایش می‌دهد. می‌توانید یک شیء را به روش معمول به‌عنوان آرگومان بدهید. برای مثال، تابع زیر یک شیء <code dir="ltr">Time</code> را به‌عنوان آرگومان می‌گیرد.

</div>

```python
def print_time(time):
    s = f'{time.hour:02d}:{time.minute:02d}:{time.second:02d}'
    print(s)
```

<div dir="rtl" style="text-align: right;">

وقتی آن را فراخوانی می‌کنیم، می‌توانیم <code dir="ltr">lunch</code> را به‌عنوان آرگومان بدهیم.

</div>

```python
print_time(lunch)
```

## ۱۴.۳. شیءها به‌عنوان مقدار بازگشتی

توابع می‌توانند شیء برگردانند. برای مثال، <code dir="ltr">make_time</code> پارامترهایی به نام <code dir="ltr">hour</code>، <code dir="ltr">minute</code> و <code dir="ltr">second</code> می‌گیرد، آن‌ها را به‌عنوان صفت در یک شیء <code dir="ltr">Time</code> ذخیره می‌کند و شیء جدید را برمی‌گرداند.

</div>

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

<div dir="rtl" style="text-align: right;">

شاید تعجب‌آور باشد که پارامترها همان نام صفت‌ها را دارند، اما این روش رایجی برای نوشتن چنین تابعی است. این‌طور از <code dir="ltr">make_time</code> برای ساختن یک شیء <code dir="ltr">Time</code> استفاده می‌کنیم.

</div>

```python
time = make_time(11, 59, 1)
print_time(time)
```

<div dir="rtl" style="text-align: right;">

## ۱۴.۴. شیءها تغییرپذیرند

فرض کنید به سانس یک فیلم می‌روید که ساعت <code dir="ltr">9:20 PM</code> شروع می‌شود و <code dir="ltr">92</code> دقیقه، یعنی یک ساعت و <code dir="ltr">32</code> دقیقه، طول می‌کشد. فیلم چه ساعتی تمام می‌شود؟

اول شیء <code dir="ltr">Time</code>‌ای می‌سازیم که زمان شروع را نشان می‌دهد.

</div>

```python
start = make_time(9, 20, 0)
print_time(start)
```

<div dir="rtl" style="text-align: right;">

برای پیدا کردن زمان پایان، می‌توانیم صفت‌های شیء <code dir="ltr">Time</code> را با اضافه کردن مدت فیلم تغییر دهیم.

</div>

```python
start.hour += 1
start.minute += 32
print_time(start)
```

<div dir="rtl" style="text-align: right;">

فیلم ساعت 10:52 PM تمام می‌شود.

این محاسبه را در یک تابع کپسوله می‌کنیم و آن را تعمیم می‌دهیم تا مدت فیلم را در سه پارامتر <code dir="ltr">hours</code>، <code dir="ltr">minutes</code> و <code dir="ltr">seconds</code> بگیرد.

</div>

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
```

<div dir="rtl" style="text-align: right;">

این مثالی است که اثرش را نشان می‌دهد.

</div>

```python
start = make_time(9, 20, 0)
increment_time(start, 1, 32, 0)
print_time(start)
```

<div dir="rtl" style="text-align: right;">

نمودار پشته‌ی زیر وضعیت برنامه را درست قبل از اینکه <code dir="ltr">increment_time</code> شیء را تغییر دهد نشان می‌دهد.

(در کتاب اصلی اینجا نمودار رسم شده است: قاب <code dir="ltr">__main__</code> با متغیر <code dir="ltr">start</code> و قاب <code dir="ltr">increment_time</code> با پارامترهای <code dir="ltr">time</code>، <code dir="ltr">hours</code> با مقدار ۱، <code dir="ltr">minutes</code> با مقدار ۳۲ و <code dir="ltr">seconds</code> با مقدار ۰؛ هر دوی <code dir="ltr">start</code> و <code dir="ltr">time</code> به یک شیء <code dir="ltr">Time</code> با صفت‌های ۹، ۲۰ و ۰ اشاره می‌کنند.)

داخل تابع، <code dir="ltr">time</code> نام مستعار <code dir="ltr">start</code> است، پس وقتی <code dir="ltr">time</code> تغییر می‌کند، <code dir="ltr">start</code> هم تغییر می‌کند.

این تابع کار می‌کند، اما بعد از اجرایش، متغیری به نام <code dir="ltr">start</code> برایمان می‌ماند که به شیئی اشاره می‌کند که زمان *پایان* را نشان می‌دهد، و دیگر شیئی که زمان شروع را نشان دهد نداریم. بهتر است <code dir="ltr">start</code> را بدون تغییر بگذاریم و شیء جدیدی برای نشان دادن زمان پایان بسازیم. می‌توانیم این کار را با کپی کردن <code dir="ltr">start</code> و تغییر کپی انجام دهیم.

</div>

<div dir="rtl" style="text-align: right;">

## ۱۴.۵. کپی کردن

ماژول <code dir="ltr">copy</code> تابعی به نام <code dir="ltr">copy</code> دارد که می‌تواند هر شیئی را تکثیر کند. می‌توانیم آن را این‌طور وارد کنیم.

</div>

```python
from copy import copy
```

<div dir="rtl" style="text-align: right;">

برای دیدن اینکه چطور کار می‌کند، با یک شیء <code dir="ltr">Time</code> جدید که زمان شروع فیلم را نشان می‌دهد شروع می‌کنیم.

</div>

```python
start = make_time(9, 20, 0)
```

<div dir="rtl" style="text-align: right;">

و یک کپی می‌سازیم.

</div>

```python
end = copy(start)
```

<div dir="rtl" style="text-align: right;">

حالا <code dir="ltr">start</code> و <code dir="ltr">end</code> داده‌ی یکسانی دارند.

</div>

```python
print_time(start)
print_time(end)
```

<div dir="rtl" style="text-align: right;">

اما عملگر <code dir="ltr">is</code> تأیید می‌کند که آن‌ها یک شیء نیستند.

</div>

```python
start is end
```

<div dir="rtl" style="text-align: right;">

ببینیم عملگر <code dir="ltr">==</code> چه می‌کند.

</div>

```python
start == end
```

<div dir="rtl" style="text-align: right;">

شاید انتظار داشته باشید <code dir="ltr">==</code> مقدار <code dir="ltr">True</code> بدهد چون شیءها داده‌ی یکسانی دارند. اما برای کلاس‌های تعریف‌شده توسط برنامه‌نویس، رفتار پیش‌فرض عملگر <code dir="ltr">==</code> همان رفتار عملگر <code dir="ltr">is</code> است – یکسانی (identity) را بررسی می‌کند، نه هم‌ارزی (equivalence) را.

</div>

<div dir="rtl" style="text-align: right;">

## ۱۴.۶. توابع خالص

می‌توانیم از <code dir="ltr">copy</code> برای نوشتن توابع خالصی استفاده کنیم که پارامترهایشان را تغییر نمی‌دهند. برای مثال، این تابعی است که یک شیء <code dir="ltr">Time</code> و یک مدت‌زمان به ساعت، دقیقه و ثانیه می‌گیرد. از شیء اصلی کپی می‌گیرد، با <code dir="ltr">increment_time</code> کپی را تغییر می‌دهد و آن را برمی‌گرداند.

</div>

```python
def add_time(time, hours, minutes, seconds):
    total = copy(time)
    increment_time(total, hours, minutes, seconds)
    return total
```

<div dir="rtl" style="text-align: right;">

این‌طور از آن استفاده می‌کنیم.

</div>

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

<div dir="rtl" style="text-align: right;">

مقدار بازگشتی شیء جدیدی است که زمان پایان فیلم را نشان می‌دهد. و می‌توانیم تأیید کنیم که <code dir="ltr">start</code> بدون تغییر مانده است.

</div>

```python
print_time(start)
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">add_time</code> یک **تابع خالص** (pure function) است چون هیچ‌یک از شیءهایی را که به‌عنوان آرگومان به آن داده شده تغییر نمی‌دهد و تنها اثرش برگرداندن یک مقدار است.

هر کاری که با توابع ناخالص انجام شود با توابع خالص هم قابل انجام است. در واقع، بعضی زبان‌های برنامه‌نویسی فقط توابع خالص را مجاز می‌دانند. برنامه‌هایی که از توابع خالص استفاده می‌کنند ممکن است کمتر مستعد خطا باشند، اما توابع ناخالص گاهی راحت‌اند و می‌توانند کارآمدتر باشند.

به‌طور کلی، پیشنهاد می‌کنم هر جا معقول است توابع خالص بنویسید و فقط اگر مزیت قانع‌کننده‌ای هست به توابع ناخالص متوسل شوید. این رویکرد **سبک برنامه‌نویسی تابعی** (functional programming style) نامیده می‌شود.

</div>

## ۱۴.۷. نمونه‌ی اولیه و وصله (Prototype and patch)

در مثال قبل، <code dir="ltr">increment_time</code> و <code dir="ltr">add_time</code> به نظر کار می‌کنند، اما اگر مثال دیگری را امتحان کنیم، می‌بینیم که کاملاً درست نیستند.

فرض کنید به سالن می‌رسید و متوجه می‌شوید فیلم ساعت <code dir="ltr">9:40</code> شروع می‌شود، نه <code dir="ltr">9:20</code>. این چیزی است که هنگام محاسبه‌ی زمان پایان به‌روزشده پیش می‌آید.

</div>

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

<div dir="rtl" style="text-align: right;">

نتیجه زمان معتبری نیست. مشکل این است که <code dir="ltr">increment_time</code> حالت‌هایی را که تعداد ثانیه‌ها یا دقیقه‌ها بیش از <code dir="ltr">60</code> می‌شود مدیریت نمی‌کند.

این نسخه‌ی بهبودیافته‌ای است که بررسی می‌کند آیا <code dir="ltr">second</code> از <code dir="ltr">60</code> بیشتر یا مساوی آن است – اگر بله، <code dir="ltr">minute</code> را افزایش می‌دهد – بعد بررسی می‌کند آیا <code dir="ltr">minute</code> از <code dir="ltr">60</code> بیشتر یا مساوی آن است – اگر بله، <code dir="ltr">hour</code> را افزایش می‌دهد.

</div>

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds

    if time.second >= 60:
        time.second -= 60
        time.minute += 1

    if time.minute >= 60:
        time.minute -= 60
        time.hour += 1
```

<div dir="rtl" style="text-align: right;">

درست کردن <code dir="ltr">increment_time</code> <code dir="ltr">add_time</code> را هم که از آن استفاده می‌کند درست می‌کند. پس حالا مثال قبل درست کار می‌کند.

</div>

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

<div dir="rtl" style="text-align: right;">

اما این تابع هنوز درست نیست، چون آرگومان‌ها ممکن است بزرگ‌تر از <code dir="ltr">60</code> باشند. برای مثال، فرض کنید مدت اجرا به‌جای <code dir="ltr">1</code> ساعت و <code dir="ltr">32</code> دقیقه، <code dir="ltr">92</code> دقیقه داده شود. ممکن است <code dir="ltr">add_time</code> را این‌طور فراخوانی کنیم.

<div dir="rtl" style="text-align: right;">

نتیجه زمان معتبری نیست. پس رویکرد متفاوتی را با استفاده از تابع <code dir="ltr">divmod</code> امتحان می‌کنیم. از <code dir="ltr">start</code> کپی می‌گیریم و با افزایش صفت <code dir="ltr">minute</code> آن را تغییر می‌دهیم.

حالا <code dir="ltr">minute</code> درست است و می‌توانیم ساعت‌ها را به <code dir="ltr">hour</code> اضافه کنیم.

<div dir="rtl" style="text-align: right;">

نتیجه زمان معتبری است. می‌توانیم همین کار را با <code dir="ltr">hour</code> و <code dir="ltr">second</code> هم بکنیم و کل فرایند را در یک تابع کپسوله کنیم.

</div>

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
    
    carry, time.second = divmod(time.second, 60)
    carry, time.minute = divmod(time.minute + carry, 60)
    carry, time.hour = divmod(time.hour + carry, 24)
```

## ۱۴.۸. توسعه‌ی طراحی‌محور (Design-first development)

یک طرح جایگزین **توسعه‌ی طراحی‌محور** (design-first development) است، که قبل از ساختن نمونه‌ی اولیه برنامه‌ریزی بیشتری لازم دارد. در فرایند طراحی‌محور، گاهی بینشی سطح بالا درباره‌ی مسئله برنامه‌نویسی را خیلی آسان‌تر می‌کند.

در این مورد، بینش این است که می‌توانیم یک شیء <code dir="ltr">Time</code> را عددی سه‌رقمی در مبنای ۶۰ – که مبنای شصتگانی (sexagesimal) هم نامیده می‌شود – در نظر بگیریم. صفت <code dir="ltr">second</code> «ستون یکان»، صفت <code dir="ltr">minute</code> «ستون شصت‌ها» و صفت <code dir="ltr">hour</code> «ستون سی‌وشش‌صدها» است. وقتی <code dir="ltr">increment_time</code> را نوشتیم، در واقع در مبنای ۶۰ جمع می‌کردیم، به همین دلیل بود که باید از یک ستون به ستون بعدی رقم نقلی (carry) می‌بردیم.

این مشاهده رویکرد دیگری برای کل مسئله پیشنهاد می‌کند – می‌توانیم شیءهای <code dir="ltr">Time</code> را به عدد صحیح تبدیل کنیم و از این واقعیت که پایتون می‌داند چطور حساب عددهای صحیح را انجام دهد بهره ببریم.

این تابعی است که یک <code dir="ltr">Time</code> را به عدد صحیح تبدیل می‌کند.

</div>

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

<div dir="rtl" style="text-align: right;">

نتیجه تعداد ثانیه‌ها از ابتدای روز است. برای مثال، <code dir="ltr">01:01:01</code> یعنی <code dir="ltr">1</code> ساعت، <code dir="ltr">1</code> دقیقه و <code dir="ltr">1</code> ثانیه از ابتدای روز، که مجموع <code dir="ltr">3600</code> ثانیه، <code dir="ltr">60</code> ثانیه و <code dir="ltr">1</code> ثانیه است.

</div>

```python
time = make_time(1, 1, 1)
print_time(time)
time_to_int(time)
```

<div dir="rtl" style="text-align: right;">

و این تابعی است که در جهت مخالف می‌رود – تبدیل یک عدد صحیح به شیء <code dir="ltr">Time</code> – با استفاده از تابع <code dir="ltr">divmod</code>.

</div>

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

<div dir="rtl" style="text-align: right;">

می‌توانیم آن را با تبدیل مثال قبل به <code dir="ltr">Time</code> آزمایش کنیم.

</div>

```python
time = int_to_time(3661)
print_time(time)
```

<div dir="rtl" style="text-align: right;">

با استفاده از این توابع، می‌توانیم نسخه‌ی مختصرتری از <code dir="ltr">add_time</code> بنویسیم.

</div>

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

<div dir="rtl" style="text-align: right;">

خط اول آرگومان‌ها را به یک شیء <code dir="ltr">Time</code> به نام <code dir="ltr">duration</code> تبدیل می‌کند. خط دوم <code dir="ltr">time</code> و <code dir="ltr">duration</code> را به ثانیه تبدیل می‌کند و جمعشان می‌زند. خط سوم مجموع را به شیء <code dir="ltr">Time</code> تبدیل می‌کند و برمی‌گرداند.

این‌طور کار می‌کند.

</div>

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

<div dir="rtl" style="text-align: right;">

از بعضی جهات، تبدیل از مبنای ۶۰ به مبنای ۱۰ و برگشت دشوارتر از سروکار داشتن مستقیم با زمان‌هاست. تبدیل مبنا انتزاعی‌تر است؛ شهود ما برای کار با مقدارهای زمان بهتر است.

اما اگر بینش لازم را داشته باشیم که زمان‌ها را مثل عددهای مبنای ۶۰ در نظر بگیریم – و برای نوشتن توابع تبدیل <code dir="ltr">time_to_int</code> و <code dir="ltr">int_to_time</code> تلاش کنیم – برنامه‌ای می‌گیریم که کوتاه‌تر، خواناتر، اشکال‌زدایی‌اش آسان‌تر و قابل‌اعتمادتر است.

افزودن ویژگی‌ها در آینده هم آسان‌تر است. برای مثال، تصور کنید دو شیء <code dir="ltr">Time</code> را از هم کم کنید تا مدت بین آن‌ها را پیدا کنید. رویکرد ساده‌لوحانه پیاده‌سازی تفریق با قرض گرفتن است. استفاده از توابع تبدیل آسان‌تر و با احتمال بیشتر درست است.

## ۱۴.۱۰. واژه‌نامه

**برنامه‌نویسی شیءگرا (object-oriented programming):** سبکی از برنامه‌نویسی که از شیءها برای سازمان‌دهی کد و داده استفاده می‌کند.

**کلاس (class):** نوع تعریف‌شده توسط برنامه‌نویس. تعریف کلاس یک شیء کلاس جدید می‌سازد.

**شیء کلاس (class object):** شیئی که یک کلاس را نشان می‌دهد – نتیجه‌ی تعریف کلاس است.

**نمونه‌سازی (instantiation):** فرایند ساختن شیئی که به یک کلاس تعلق دارد.

**نمونه (instance):** شیئی که به یک کلاس تعلق دارد.

**صفت (attribute):** متغیری وابسته به یک شیء، که متغیر نمونه (instance variable) هم نامیده می‌شود.

**نمودار شیء (object diagram):** نمایش گرافیکی یک شیء، صفت‌هایش و مقدارهای آن‌ها.

**مشخص‌کننده‌ی قالب (format specifier):** در f-string، تعیین می‌کند مقدار چگونه به رشته تبدیل شود.

**تابع خالص (pure function):** تابعی که پارامترهایش را تغییر نمی‌دهد و اثری جز برگرداندن مقدار ندارد.

**سبک برنامه‌نویسی تابعی (functional programming style):** شیوه‌ای از برنامه‌نویسی که تا حد ممکن از توابع خالص استفاده می‌کند.

**نمونه‌ی اولیه و وصله (prototype and patch):** شیوه‌ای برای توسعه‌ی برنامه با شروع از یک پیش‌نویس خام و افزودن تدریجی ویژگی‌ها و رفع باگ‌ها.

**توسعه‌ی طراحی‌محور (design-first development):** شیوه‌ای برای توسعه‌ی برنامه با برنامه‌ریزی دقیق‌تر از نمونه‌ی اولیه و وصله.
