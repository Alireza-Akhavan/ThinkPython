## ۱۵.۱. تعریف متدها

در فصل قبل کلاسی به نام <code dir="ltr">Time</code> تعریف کردیم و تابعی به نام <code dir="ltr">print_time</code> نوشتیم که یک زمان از روز را نمایش می‌دهد.

</div>

```python
class Time:
    """Represents the time of day."""

def print_time(time):
    s = f'{time.hour:02d}:{time.minute:02d}:{time.second:02d}'
    print(s)
```

<div dir="rtl" style="text-align: right;">

برای تبدیل <code dir="ltr">print_time</code> به متد، فقط کافی است تعریف تابع را داخل تعریف کلاس ببریم. به تغییر تورفتگی دقت کنید.

همزمان، نام پارامتر را از <code dir="ltr">time</code> به <code dir="ltr">self</code> تغییر می‌دهیم. این تغییر ضروری نیست، اما قراردادی است که اولین پارامتر یک متد <code dir="ltr">self</code> نامیده شود.

</div>

```python
class Time:
    """Represents the time of day."""    

    def print_time(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        print(s)
```

<div dir="rtl" style="text-align: right;">

برای فراخوانی این متد، باید یک شیء <code dir="ltr">Time</code> را به‌عنوان آرگومان بدهید. این تابعی است که برای ساختن یک شیء <code dir="ltr">Time</code> استفاده می‌کنیم.

</div>

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

<div dir="rtl" style="text-align: right;">

و این یک نمونه‌ی <code dir="ltr">Time</code> است.

</div>

```python
start = make_time(9, 40, 0)
```

<div dir="rtl" style="text-align: right;">

حالا دو راه برای فراخوانی <code dir="ltr">print_time</code> وجود دارد. راه اول (و کمتر رایج) استفاده از نحو تابعی است.

</div>

```python
Time.print_time(start)
```

<div dir="rtl" style="text-align: right;">

در این نسخه، <code dir="ltr">Time</code> نام کلاس، <code dir="ltr">print_time</code> نام متد و <code dir="ltr">start</code> به‌عنوان پارامتر داده شده است. راه دوم (و idiomatic‌تر) استفاده از نحو متدی است:

</div>

```python
start.print_time()
```

<div dir="rtl" style="text-align: right;">

در این نسخه، <code dir="ltr">start</code> شیئی است که متد روی آن احضار می‌شود، که **گیرنده** (receiver) نامیده می‌شود، بر اساس این قیاس که احضار یک متد مثل فرستادن پیام به یک شیء است.

صرف‌نظر از نحو، رفتار متد یکسان است. گیرنده به اولین پارامتر نسبت داده می‌شود، پس داخل متد، <code dir="ltr">self</code> به همان شیئی اشاره می‌کند که <code dir="ltr">start</code> اشاره می‌کند.

</div>

<div dir="rtl" style="text-align: right;">

## ۱۵.۲. متد دیگر

این تابع <code dir="ltr">time_to_int</code> از فصل قبل است.

</div>

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

<div dir="rtl" style="text-align: right;">

و این نسخه‌ای است که به‌صورت متد بازنویسی شده.

</div>

```python
%%add_method_to Time

    def time_to_int(self):
        minutes = self.hour * 60 + self.minute
        seconds = minutes * 60 + self.second
        return seconds
```

<div dir="rtl" style="text-align: right;">

خط اول از دستور ویژه‌ی <code dir="ltr">add_method_to</code> استفاده می‌کند که متدی را به کلاسی که قبلاً تعریف شده اضافه می‌کند. این دستور در نوت‌بوک Jupyter کار می‌کند، اما بخشی از پایتون نیست، پس در محیط‌های دیگر کار نمی‌کند. معمولاً همه‌ی متدهای یک کلاس داخل تعریف کلاس هستند، پس همزمان با کلاس تعریف می‌شوند. اما برای این کتاب مفید است که متدها را یکی‌یکی تعریف کنیم.

مثل مثال قبل، تعریف متد تورفته است و نام پارامتر <code dir="ltr">self</code> است. جدا از این، متد با تابع یکسان است. این‌طور احضارش می‌کنیم.

</div>

```python
start.time_to_int()
```

<div dir="rtl" style="text-align: right;">

معمولاً می‌گویند یک تابع را «فراخوانی» (call) و یک متد را «احضار» (invoke) می‌کنیم، اما معنایشان یکی است.

</div>

<div dir="rtl" style="text-align: right;">

## ۱۵.۳. متدهای ایستا (Static methods)

به‌عنوان مثالی دیگر، تابع <code dir="ltr">int_to_time</code> را در نظر بگیرید. این نسخه‌ی فصل قبل است.

</div>

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

<div dir="rtl" style="text-align: right;">

این تابع <code dir="ltr">seconds</code> را به‌عنوان پارامتر می‌گیرد و یک شیء <code dir="ltr">Time</code> جدید برمی‌گرداند. اگر آن را به متد کلاس <code dir="ltr">Time</code> تبدیل کنیم، باید آن را روی یک شیء <code dir="ltr">Time</code> احضار کنیم. اما اگر می‌خواهیم یک شیء <code dir="ltr">Time</code> جدید بسازیم، قرار است آن را روی چه چیزی احضار کنیم؟

می‌توانیم این مسئله‌ی «مرغ و تخم‌مرغ» را با **متد ایستا** (static method) حل کنیم، که متدی است که برای احضار شدن نیازی به نمونه‌ای از کلاس ندارد. این‌طور این تابع را به‌صورت متد ایستا بازنویسی می‌کنیم.

</div>

```python
%%add_method_to Time

    def int_to_time(seconds):
        minute, second = divmod(seconds, 60)
        hour, minute = divmod(minute, 60)
        return make_time(hour, minute, second)
```

<div dir="rtl" style="text-align: right;">

چون متد ایستا است، <code dir="ltr">self</code> را به‌عنوان پارامتر ندارد. برای احضارش از <code dir="ltr">Time</code>، یعنی شیء کلاس، استفاده می‌کنیم.

</div>

```python
start = Time.int_to_time(34800)
```

<div dir="rtl" style="text-align: right;">

نتیجه شیء جدیدی است که ۹:۴۰ را نشان می‌دهد.

</div>

```python
start.print_time()
```

<div dir="rtl" style="text-align: right;">

حالا که <code dir="ltr">Time.int_to_time</code> را داریم، می‌توانیم از آن برای نوشتن <code dir="ltr">add_time</code> به‌صورت متد استفاده کنیم. این تابع فصل قبل است.

</div>

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

<div dir="rtl" style="text-align: right;">

و این نسخه‌ای است که به‌صورت متد بازنویسی شده.

</div>

```python
%%add_method_to Time

    def add_time(self, hours, minutes, seconds):
        duration = make_time(hours, minutes, seconds)
        seconds = Time.time_to_int(self) + Time.time_to_int(duration)
        return Time.int_to_time(seconds)
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">add_time</code> پارامتر <code dir="ltr">self</code> دارد چون متد ایستا نیست. متد معمولی است – که **متد نمونه** (instance method) هم نامیده می‌شود. برای احضارش به یک نمونه‌ی <code dir="ltr">Time</code> نیاز داریم.

</div>

```python
end = start.add_time(1, 32, 0)
end.print_time()
```

<div dir="rtl" style="text-align: right;">

## ۱۵.۴. مقایسه‌ی شیءهای Time

به‌عنوان یک مثال دیگر، <code dir="ltr">is_after</code> را به‌صورت متد می‌نویسیم. این تابع <code dir="ltr">is_after</code> است، که راه‌حل یکی از تمرین‌های فصل قبل است.

</div>

```python
def is_after(t1, t2):
    return time_to_int(t1) > time_to_int(t2)
```

<div dir="rtl" style="text-align: right;">

و این به‌صورت متد است.

</div>

```python
%%add_method_to Time

    def is_after(self, other):
        return self.time_to_int() > other.time_to_int()
```

<div dir="rtl" style="text-align: right;">

چون دو شیء را مقایسه می‌کنیم و اولین پارامتر <code dir="ltr">self</code> است، پارامتر دوم را <code dir="ltr">other</code> می‌نامیم. برای استفاده از این متد، باید آن را روی یک شیء احضار کنیم و دیگری را به‌عنوان آرگومان بدهیم.

</div>

```python
end.is_after(start)
```

<div dir="rtl" style="text-align: right;">

یک چیز خوب در این نحو این است که تقریباً مثل یک پرسش خوانده می‌شود: «<code dir="ltr">end</code> بعد از <code dir="ltr">start</code> است؟»

</div>

## ۱۵.۶. متد init

ویژه‌ترین متد از میان متدهای ویژه <code dir="ltr">__init__</code> است، که به این دلیل چنین نامیده می‌شود که صفت‌های یک شیء جدید را مقداردهی اولیه (initialize) می‌کند. متد <code dir="ltr">__init__</code> برای کلاس <code dir="ltr">Time</code> ممکن است این‌طور باشد:

</div>

```python
%%add_method_to Time

    def __init__(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second
```

<div dir="rtl" style="text-align: right;">

حالا وقتی یک شیء <code dir="ltr">Time</code> نمونه‌سازی می‌کنیم، پایتون <code dir="ltr">__init__</code> را احضار می‌کند و آرگومان‌ها را به آن می‌دهد. پس می‌توانیم همزمان شیء بسازیم و صفت‌ها را مقداردهی اولیه کنیم.

</div>

```python
time = Time(9, 40, 0)
print(time)
```

<div dir="rtl" style="text-align: right;">

در این مثال، پارامترها اختیاری‌اند، پس اگر <code dir="ltr">Time</code> را بدون آرگومان فراخوانی کنید، مقدارهای پیش‌فرض را می‌گیرید.

</div>

```python
time = Time()
print(time)
```

<div dir="rtl" style="text-align: right;">

اگر یک آرگومان بدهید، <code dir="ltr">hour</code> را بازنویسی می‌کند:

```python
time = Time(9)
print(time)
```

<div dir="rtl" style="text-align: right;">

اگر دو آرگومان بدهید، <code dir="ltr">hour</code> و <code dir="ltr">minute</code> را بازنویسی می‌کنند.

</div>

```python
time = Time(9, 45)
print(time)
```

<div dir="rtl" style="text-align: right;">

و اگر سه آرگومان بدهید، هر سه مقدار پیش‌فرض را بازنویسی می‌کنند.

وقتی کلاس جدیدی می‌نویسم، تقریباً همیشه با نوشتن <code dir="ltr">__init__</code> شروع می‌کنم، که ساختن شیءها را آسان‌تر می‌کند، و <code dir="ltr">__str__</code>، که برای اشکال‌زدایی مفید است.

</div>

## ۱۵.۷. سربارگذاری عملگر (Operator overloading)

با تعریف متدهای ویژه‌ی دیگر، می‌توانید رفتار عملگرها را روی نوع‌های تعریف‌شده توسط برنامه‌نویس مشخص کنید. برای مثال، اگر متدی به نام <code dir="ltr">__add__</code> برای کلاس <code dir="ltr">Time</code> تعریف کنید، می‌توانید عملگر <code dir="ltr">+</code> را روی شیءهای Time به کار ببرید.

این یک متد <code dir="ltr">__add__</code> است.

</div>

```python
%%add_method_to Time

    def __add__(self, other):
        seconds = self.time_to_int() + other.time_to_int()
        return Time.int_to_time(seconds)
```

<div dir="rtl" style="text-align: right;">

می‌توانیم آن را این‌طور به کار ببریم.

</div>

```python
duration = Time(1, 32)
end = start + duration
print(end)
```

<div dir="rtl" style="text-align: right;">

وقتی این سه خط کد را اجرا می‌کنیم، اتفاق‌های زیادی می‌افتد:

* وقتی یک شیء <code dir="ltr">Time</code> را نمونه‌سازی می‌کنیم، متد <code dir="ltr">__init__</code> احضار می‌شود.

* وقتی عملگر <code dir="ltr">+</code> را با یک شیء <code dir="ltr">Time</code> به کار می‌بریم، متد <code dir="ltr">__add__</code> آن احضار می‌شود.

* و وقتی یک شیء <code dir="ltr">Time</code> را چاپ می‌کنیم، متد <code dir="ltr">__str__</code> آن احضار می‌شود.

تغییر رفتار یک عملگر طوری که با نوع‌های تعریف‌شده توسط برنامه‌نویس کار کند **سربارگذاری عملگر** (operator overloading) نامیده می‌شود. برای هر عملگر، مثل <code dir="ltr">+</code>، یک متد ویژه‌ی متناظر وجود دارد، مثل <code dir="ltr">__add__</code>.

</div>

## ۱۵.۸. اشکال‌زدایی

یک شیء <code dir="ltr">Time</code> معتبر است اگر مقدارهای <code dir="ltr">minute</code> و <code dir="ltr">second</code> بین <code dir="ltr">0</code> و <code dir="ltr">60</code> باشند – با احتساب <code dir="ltr">0</code> اما بدون احتساب <code dir="ltr">60</code> – و اگر <code dir="ltr">hour</code> مثبت باشد. همچنین <code dir="ltr">hour</code> و <code dir="ltr">minute</code> باید مقدار صحیح باشند، اما ممکن است اجازه دهیم <code dir="ltr">second</code> بخش کسری داشته باشد. نیازمندی‌هایی مثل این **پایا** (invariant) نامیده می‌شوند چون باید همیشه درست باشند. به عبارت دیگر، اگر درست نباشند، چیزی اشتباه شده است.

نوشتن کد برای بررسی پایاها می‌تواند به شناسایی خطاها و پیدا کردن علتشان کمک کند. برای مثال، ممکن است متدی مثل <code dir="ltr">is_valid</code> داشته باشید که یک شیء Time را بگیرد و اگر پایایی را نقض کند <code dir="ltr">False</code> برگرداند.

</div>

```python
%%add_method_to Time

    def is_valid(self):
        if self.hour < 0 or self.minute < 0 or self.second < 0:
            return False
        if self.minute >= 60 or self.second >= 60:
            return False
        if not isinstance(self.hour, int):
            return False
        if not isinstance(self.minute, int):
            return False
        return True
```

<div dir="rtl" style="text-align: right;">

بعد، در ابتدای هر متد می‌توانید آرگومان‌ها را بررسی کنید تا مطمئن شوید معتبرند.

</div>

```python
%%add_method_to Time

    def is_after(self, other):
        assert self.is_valid(), 'self is not a valid Time'
        assert other.is_valid(), 'other is not a valid Time'
        return self.time_to_int() > other.time_to_int()
```

<div dir="rtl" style="text-align: right;">

دستور <code dir="ltr">assert</code> عبارت بعد از خود را ارزیابی می‌کند. اگر نتیجه <code dir="ltr">True</code> باشد، کاری نمی‌کند؛ اگر نتیجه <code dir="ltr">False</code> باشد، باعث <code dir="ltr">AssertionError</code> می‌شود. این یک مثال است.

</div>

```python
duration = Time(minute=132)
print(duration)
```

## ۱۵.۹. واژه‌نامه

**زبان شیءگرا (object-oriented language):** زبانی که امکاناتی برای پشتیبانی از برنامه‌نویسی شیءگرا دارد، به‌ویژه نوع‌های تعریف‌شده توسط کاربر.

**متد (method):** تابعی که داخل تعریف یک کلاس تعریف می‌شود و روی نمونه‌های آن کلاس احضار می‌شود.

**گیرنده (receiver):** شیئی که یک متد روی آن احضار می‌شود.

**متد ایستا (static method):** متدی که می‌توان آن را بدون شیء به‌عنوان گیرنده احضار کرد.

**متد نمونه (instance method):** متدی که باید با یک شیء به‌عنوان گیرنده احضار شود.

**متد ویژه (special method):** متدی که شیوه‌ی کار عملگرها و بعضی توابع را با یک شیء تغییر می‌دهد.

**سربارگذاری عملگر (operator overloading):** فرایند استفاده از متدهای ویژه برای تغییر شیوه‌ی کار عملگرها با نوع‌های تعریف‌شده توسط کاربر.

**پایا (invariant):** شرطی که باید در طول اجرای برنامه همیشه درست باشد.

</div>
