## ۴.۳. کپسوله‌سازی و تعمیم

بیایید کد رسم مربع از بخش قبل را بگیریم و در تابعی به نام <code dir="ltr">square</code> بگذاریم.

</div>

````python
def square():
    for i in range(4):
        forward(50)
        left(90)
````

<div dir="rtl" style="text-align: right;">

حالا می‌توانیم تابع را این‌طور فراخوانی کنیم.

</div>

````python
make_turtle()
square()
````

<div dir="rtl" style="text-align: right;">

پیچیدن یک تکه کد در یک تابع **کپسوله‌سازی** (encapsulation) نامیده می‌شود. یکی از مزایای کپسوله‌سازی این است که یک نام به کد می‌دهد و این نام نوعی مستندسازی است. مزیت دیگر این است که اگر بخواهید دوباره از کد استفاده کنید، دو بار فراخوانی یک تابع مختصرتر از کپی و پیست کردن بدنه است!

در نسخه‌ی فعلی، اندازه‌ی مربع همیشه <code dir="ltr">50</code> است. اگر بخواهیم مربع‌هایی با اندازه‌های مختلف بکشیم، می‌توانیم طول ضلع‌ها را به‌عنوان پارامتر بگیریم.

</div>

````python
def square(length):
    for i in range(4):
        forward(length)
        left(90)
````

<div dir="rtl" style="text-align: right;">

حالا می‌توانیم مربع‌هایی با اندازه‌های مختلف بکشیم.

</div>

````python
make_turtle()
square(30)
square(60)
````

<div dir="rtl" style="text-align: right;">

*(خروجی: دو مربع تودرتو که گوشه‌ی پایین‌چپ‌شان مشترک است؛ ضلع‌ها ۳۰ و ۶۰.)*

افزودن یک پارامتر به تابع **تعمیم** (generalization) نامیده می‌شود، چون تابع را کلی‌تر می‌کند: با نسخه‌ی قبلی مربع همیشه هم‌اندازه است؛ با این نسخه می‌تواند هر اندازه‌ای داشته باشد.

اگر پارامتر دیگری اضافه کنیم، می‌توانیم آن را باز هم کلی‌تر کنیم. تابع زیر چندضلعی‌های منتظم با تعداد ضلع داده‌شده را می‌کشد.

</div>

````python
def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)
````

<div dir="rtl" style="text-align: right;">

در یک چندضلعی منتظم با <code dir="ltr">n</code> ضلع، زاویه‌ی بین ضلع‌های مجاور <code dir="ltr">360 / n</code> درجه است.

مثال زیر یک چندضلعی <code dir="ltr">7</code> ضلعی با طول ضلع <code dir="ltr">30</code> می‌کشد.

</div>

````python
make_turtle()
polygon(7, 30)
````

<div dir="rtl" style="text-align: right;">

وقتی تابعی بیش از چند آرگومان عددی دارد، به‌راحتی فراموش می‌شود که آن‌ها چه بودند یا به چه ترتیبی باید باشند. ایده‌ی خوبی است که نام پارامترها را در فهرست آرگومان‌ها بنویسید.

</div>

````python
make_turtle()
polygon(n=7, length=30)
````

<div dir="rtl" style="text-align: right;">

به این‌ها گاهی «آرگومان‌های نام‌دار» می‌گویند چون نام پارامترها را دارند. اما در پایتون بیشتر به آن‌ها **آرگومان کلیدواژه‌ای** (keyword argument) می‌گویند (با کلمه‌های کلیدی پایتون مثل <code dir="ltr">for</code> و <code dir="ltr">def</code> اشتباه نگیرید).

این استفاده از عملگر انتساب، <code dir="ltr">=</code>، یادآوری می‌کند که آرگومان‌ها و پارامترها چگونه کار می‌کنند – وقتی تابعی را فراخوانی می‌کنید، آرگومان‌ها به پارامترها نسبت داده می‌شوند.

</div>

<div dir="rtl" style="text-align: right;">

## ۴.۴. تقریب یک دایره

حالا فرض کنید می‌خواهیم یک دایره بکشیم. می‌توانیم این کار را، به‌صورت تقریبی، با کشیدن یک چندضلعی با تعداد زیادی ضلع انجام دهیم، به‌طوری که هر ضلع آن‌قدر کوچک باشد که دیدنش سخت شود. این تابعی است که از <code dir="ltr">polygon</code> برای کشیدن یک چندضلعی <code dir="ltr">30</code> ضلعی که دایره را تقریب می‌زند استفاده می‌کند.

</div>

````python
import math

def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

<div dir="rtl" style="text-align: right;">

<code dir="ltr">circle</code> شعاع دایره را به‌عنوان پارامتر می‌گیرد. <code dir="ltr">circumference</code> را حساب می‌کند که محیط دایره‌ای با شعاع داده‌شده است. <code dir="ltr">n</code> تعداد ضلع‌هاست، پس <code dir="ltr">circumference / n</code> طول هر ضلع است.

اجرای این تابع ممکن است زمان زیادی ببرد. می‌توانیم با فراخوانی <code dir="ltr">make_turtle</code> با یک آرگومان کلیدواژه‌ای به نام <code dir="ltr">delay</code> آن را سریع‌تر کنیم؛ این آرگومان زمانی را، بر حسب ثانیه، که لاک‌پشت بعد از هر قدم صبر می‌کند تعیین می‌کند. مقدار پیش‌فرض <code dir="ltr">0.2</code> ثانیه است – اگر آن را <code dir="ltr">0.02</code> بگذاریم حدود ۱۰ برابر سریع‌تر اجرا می‌شود.

</div>

````python
make_turtle(delay=0.02)
circle(30)
````

<div dir="rtl" style="text-align: right;">

*(خروجی: یک دایره‌ی تقریبی با شعاع ۳۰.)*

محدودیت این راه‌حل این است که <code dir="ltr">n</code> یک مقدار ثابت است؛ یعنی برای دایره‌های بسیار بزرگ ضلع‌ها خیلی بلند می‌شوند و برای دایره‌های کوچک وقت را برای کشیدن ضلع‌های خیلی کوتاه تلف می‌کنیم. یک راه این است که تابع را با گرفتن <code dir="ltr">n</code> به‌عنوان پارامتر تعمیم دهیم. اما فعلاً ساده نگهش می‌داریم.

</div>

<div dir="rtl" style="text-align: right;">

## ۴.۵. بازآرایی (Refactoring)

حالا نسخه‌ی کلی‌تری از <code dir="ltr">circle</code> بنویسیم، به نام <code dir="ltr">arc</code>، که پارامتر دومی به نام <code dir="ltr">angle</code> می‌گیرد و کمانی از یک دایره را می‌کشد که زاویه‌ی داده‌شده را می‌پوشاند. برای مثال، اگر <code dir="ltr">angle</code> برابر <code dir="ltr">360</code> درجه باشد، یک دایره‌ی کامل می‌کشد. اگر <code dir="ltr">angle</code> برابر <code dir="ltr">180</code> درجه باشد، نیم‌دایره می‌کشد.

برای نوشتن <code dir="ltr">circle</code> توانستیم از <code dir="ltr">polygon</code> دوباره استفاده کنیم، چون یک چندضلعی با ضلع‌های زیاد تقریب خوبی از دایره است. اما نمی‌توانیم از <code dir="ltr">polygon</code> برای نوشتن <code dir="ltr">arc</code> استفاده کنیم.

در عوض، نسخه‌ی کلی‌تر <code dir="ltr">polygon</code> را می‌سازیم که <code dir="ltr">polyline</code> نام دارد.

</div>

````python
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)
````

<div dir="rtl" style="text-align: right;">

<code dir="ltr">polyline</code> این‌ها را به‌عنوان پارامتر می‌گیرد: تعداد پاره‌خط‌هایی که باید کشیده شوند، <code dir="ltr">n</code>، طول پاره‌خط‌ها، <code dir="ltr">length</code>، و زاویه‌ی بین آن‌ها، <code dir="ltr">angle</code>.

حالا می‌توانیم <code dir="ltr">polygon</code> را طوری بازنویسی کنیم که از <code dir="ltr">polyline</code> استفاده کند.

</div>

````python
def polygon(n, length):
    angle = 360.0 / n
    polyline(n, length, angle)
````

<div dir="rtl" style="text-align: right;">

و می‌توانیم از <code dir="ltr">polyline</code> برای نوشتن <code dir="ltr">arc</code> استفاده کنیم.

</div>

````python
def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)
````

<div dir="rtl" style="text-align: right;">

<code dir="ltr">arc</code> شبیه <code dir="ltr">circle</code> است، با این تفاوت که <code dir="ltr">arc_length</code> را حساب می‌کند که کسری از محیط یک دایره است.

سرانجام، می‌توانیم <code dir="ltr">circle</code> را طوری بازنویسی کنیم که از <code dir="ltr">arc</code> استفاده کند.

</div>

````python
def circle(radius):
    arc(radius,  360)
````

<div dir="rtl" style="text-align: right;">

برای بررسی اینکه این توابع همان‌طور که انتظار می‌رود کار می‌کنند، از آن‌ها برای کشیدن چیزی شبیه حلزون استفاده می‌کنیم. با <code dir="ltr">delay=0</code> لاک‌پشت تا حد امکان سریع اجرا می‌شود.

</div>

````python
make_turtle(delay=0)
polygon(n=20, length=9)
arc(radius=70, angle=70)
circle(radius=10)
````

<div dir="rtl" style="text-align: right;">

*(خروجی: یک چندضلعی ۲۰ ضلعی، یک کمان ۷۰ درجه‌ای که از آن بیرون می‌آید و یک دایره‌ی کوچک در انتهای کمان؛ شبیه حلزون.)*

در این مثال با کدی که کار می‌کرد شروع کردیم و آن را با توابع مختلف دوباره سازمان‌دهی کردیم. تغییراتی مثل این، که کد را بهبود می‌دهند بدون اینکه رفتارش را تغییر دهند، **بازآرایی** (refactoring) نامیده می‌شوند.

اگر از قبل برنامه‌ریزی کرده بودیم، شاید ابتدا <code dir="ltr">polyline</code> را می‌نوشتیم و از بازآرایی دوری می‌کردیم، اما اغلب در آغاز یک پروژه آن‌قدر نمی‌دانید که همه‌ی توابع را طراحی کنید. وقتی شروع به کدنویسی می‌کنید، مسئله را بهتر می‌فهمید. گاهی بازآرایی نشانه‌ی این است که چیزی یاد گرفته‌اید.

</div>

## ۴.۶. نمودار پشته

وقتی <code dir="ltr">circle</code> را فراخوانی می‌کنیم، <code dir="ltr">arc</code> را فراخوانی می‌کند، که <code dir="ltr">polyline</code> را فراخوانی می‌کند. می‌توانیم از یک نمودار پشته برای نشان دادن این دنباله‌ی فراخوانی‌ها و پارامترهای هر یک استفاده کنیم.

(در کتاب اصلی این‌جا نمودار پشته رسم شده است: سه قاب، <code dir="ltr">circle</code> با <code dir="ltr">radius=30</code>، <code dir="ltr">arc</code> با <code dir="ltr">radius=30</code> و <code dir="ltr">angle=360</code> و <code dir="ltr">polyline</code> با <code dir="ltr">n=60</code> و <code dir="ltr">length=3.04</code> و <code dir="ltr">angle=5.8</code>.)

دقت کنید که مقدار <code dir="ltr">angle</code> در <code dir="ltr">polyline</code> با مقدار <code dir="ltr">angle</code> در <code dir="ltr">arc</code> فرق دارد. پارامترها محلی‌اند، یعنی می‌توانید همان نام پارامتر را در توابع مختلف به کار ببرید؛ در هر تابع یک متغیر متفاوت است و می‌تواند به مقدار متفاوتی اشاره کند.

## ۴.۷. یک طرح توسعه

**طرح توسعه** (development plan) فرایندی برای نوشتن برنامه‌هاست. فرایندی که در این فصل به کار بردیم «کپسوله‌سازی و تعمیم» است. مراحل این فرایند عبارت‌اند از:

1.  با نوشتن یک برنامه‌ی کوچک بدون تعریف تابع شروع کنید.

2.  وقتی برنامه کار کرد، یک بخش منسجم از آن را مشخص کنید، آن بخش را در یک تابع کپسوله کنید و نامی به آن بدهید.

3.  با افزودن پارامترهای مناسب تابع را تعمیم دهید.

4.  مراحل ۱ تا ۳ را تکرار کنید تا مجموعه‌ای از توابع کارآمد داشته باشید.

5.  به دنبال فرصت‌هایی برای بهبود برنامه با بازآرایی بگردید. برای مثال، اگر در چند جا کد مشابه دارید، فکر کنید آن را به یک تابع به‌اندازه‌ی کافی کلی تبدیل کنید.

این فرایند معایبی دارد – بعداً جایگزین‌هایی را می‌بینیم – اما اگر از قبل نمی‌دانید چطور برنامه را به توابع تقسیم کنید می‌تواند مفید باشد. این رویکرد به شما اجازه می‌دهد در حین پیشرفت طراحی کنید.

طراحی یک تابع دو بخش دارد:

* **رابط** (interface) نحوه‌ی استفاده از تابع است، شامل نام، پارامترهایی که می‌گیرد و کاری که قرار است تابع انجام دهد.

* **پیاده‌سازی** (implementation) نحوه‌ی انجام کاری است که تابع قرار است انجام دهد.

برای مثال، این اولین نسخه‌ی <code dir="ltr">circle</code> است که نوشتیم و از <code dir="ltr">polygon</code> استفاده می‌کند.

</div>

````python
def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

## ۴.۸. رشته‌های مستندساز (Docstring)

**docstring** رشته‌ای است در ابتدای یک تابع که رابط را توضیح می‌دهد («doc» مخفف «documentation» یعنی مستندات است). این یک مثال است:

</div>

````python
def polyline(n, length, angle):
    """Draws line segments with the given length and angle between them.
    
    n: integer number of line segments
    length: length of the line segments
    angle: angle between segments (in degrees)
    """    
    for i in range(n):
        forward(length)
        left(angle)
````

<div dir="rtl" style="text-align: right;">

طبق قرارداد، docstringها رشته‌هایی با سه نقل‌قول هستند که **رشته‌ی چندخطی** (multiline string) هم نامیده می‌شوند، چون سه نقل‌قول اجازه می‌دهند رشته بیش از یک خط را بگیرد.

یک docstring باید:

* به‌طور مختصر توضیح دهد که تابع چه می‌کند، بدون اینکه وارد جزئیات نحوه‌ی کارش شود،

* توضیح دهد هر پارامتر چه تأثیری بر رفتار تابع دارد، و

* اگر بدیهی نیست، نشان دهد هر پارامتر باید از چه نوعی باشد.

نوشتن این نوع مستندات بخش مهمی از طراحی رابط است. رابطی که خوب طراحی شده باشد باید توضیح دادنش ساده باشد؛ اگر توضیح دادن یکی از توابعتان برایتان سخت است، شاید رابط را بتوان بهتر کرد.

</div>

<div dir="rtl" style="text-align: right;">

## ۴.۹. اشکال‌زدایی

رابط مثل یک قرارداد بین یک تابع و فراخواننده‌اش است. فراخواننده توافق می‌کند آرگومان‌های مشخصی را فراهم کند و تابع توافق می‌کند کار مشخصی را انجام دهد.

برای مثال، <code dir="ltr">polyline</code> سه آرگومان لازم دارد: <code dir="ltr">n</code> باید عدد صحیح باشد؛ <code dir="ltr">length</code> باید عددی مثبت باشد؛ و <code dir="ltr">angle</code> باید عدد باشد و در نظر گرفته می‌شود که بر حسب درجه است.

این الزام‌ها **پیش‌شرط** (precondition) نامیده می‌شوند، چون قرار است پیش از شروع اجرای تابع برقرار باشند. در مقابل، شرط‌های پایان تابع **پس‌شرط** (postcondition) هستند. پس‌شرط‌ها شامل اثر مورد نظر تابع (مثل کشیدن پاره‌خط‌ها) و هر اثر جانبی (مثل حرکت دادن لاک‌پشت یا ایجاد تغییرات دیگر) هستند.

پیش‌شرط‌ها مسئولیت فراخواننده‌اند. اگر فراخواننده پیش‌شرطی را نقض کند و تابع درست کار نکند، باگ در فراخواننده است، نه در تابع.

اگر پیش‌شرط‌ها برقرار باشند و پس‌شرط‌ها نه، باگ در تابع است. اگر پیش‌شرط‌ها و پس‌شرط‌هایتان روشن باشند، می‌توانند به اشکال‌زدایی کمک کنند.

</div>

## ۴.۱۰. واژه‌نامه

**طراحی رابط (interface design):** فرایندی برای طراحی رابط یک تابع، که شامل پارامترهایی است که باید بگیرد.

**بوم (canvas):** پنجره‌ای که برای نمایش عناصر گرافیکی از جمله خط، دایره، مستطیل و شکل‌های دیگر به کار می‌رود.

**کپسوله‌سازی (encapsulation):** فرایند تبدیل دنباله‌ای از دستورها به یک تعریف تابع.

**تعمیم (generalization):** فرایند جایگزین کردن چیزی که بی‌جهت خاص است (مثل یک عدد) با چیزی که به‌اندازه‌ی کافی کلی است (مثل یک متغیر یا پارامتر).

**آرگومان کلیدواژه‌ای (keyword argument):** آرگومانی که نام پارامتر را هم دارد.

**بازآرایی (refactoring):** فرایند تغییر یک برنامه‌ی کارآمد برای بهبود رابط توابع و ویژگی‌های دیگر کد.

**طرح توسعه (development plan):** فرایندی برای نوشتن برنامه‌ها.

**docstring:** رشته‌ای که در بالای تعریف یک تابع می‌آید تا رابط تابع را مستند کند.

**رشته‌ی چندخطی (multiline string):** رشته‌ای که داخل سه نقل‌قول قرار دارد و می‌تواند بیش از یک خط از برنامه را بگیرد.

**پیش‌شرط (precondition):** الزامی که فراخواننده باید پیش از شروع تابع برآورده کند.

**پس‌شرط (postcondition):** الزامی که تابع باید پیش از پایانش برآورده کند.

</div>
