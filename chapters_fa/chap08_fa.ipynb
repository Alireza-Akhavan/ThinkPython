## ۸.۱. رشته یک دنباله است

رشته دنباله‌ای از نویسه‌هاست. **نویسه** (character) می‌تواند یک حرف (تقریباً در هر الفبایی)، یک رقم، یک علامت نگارشی یا یک فاصله باشد.

با عملگر کروشه می‌توانید یک نویسه را از رشته انتخاب کنید. این دستور نمونه نویسه‌ی شماره‌ی ۱ را از <code dir="ltr">fruit</code> انتخاب می‌کند و به <code dir="ltr">letter</code> نسبت می‌دهد:

</div>

```python
fruit = 'banana'
letter = fruit[1]
```

<div dir="rtl" style="text-align: right;">

عبارت داخل کروشه یک **اندیس** (index) است که به این دلیل چنین نامیده می‌شود که *نشان می‌دهد* کدام نویسه از دنباله انتخاب شود. اما نتیجه ممکن است آن‌طور که انتظار دارید نباشد.

</div>

```python
letter
```

<div dir="rtl" style="text-align: right;">

حرفی که اندیس <code dir="ltr">1</code> دارد در واقع حرف دوم رشته است. اندیس فاصله از ابتدای رشته است، پس فاصله‌ی حرف اول <code dir="ltr">0</code> است.

</div>

```python
fruit[0]
```

<div dir="rtl" style="text-align: right;">

می‌توانید <code dir="ltr">'b'</code> را حرف «صفرم» <code dir="ltr">'banana'</code> در نظر بگیرید (در انگلیسی: "zero-eth").

اندیس داخل کروشه می‌تواند یک متغیر باشد.

</div>

```python
i = 1
fruit[i]
```

<div dir="rtl" style="text-align: right;">

یا عبارتی که شامل متغیرها و عملگرهاست.

</div>

```python
fruit[i+1]
```

<div dir="rtl" style="text-align: right;">

اما مقدار اندیس باید عدد صحیح باشد – در غیر این صورت <code dir="ltr">TypeError</code> می‌گیرید.

</div>

```python
%%expect TypeError

fruit[1.5]
```

<div dir="rtl" style="text-align: right;">

همان‌طور که در فصل ۱ دیدیم، می‌توانیم از تابع داخلی <code dir="ltr">len</code> برای گرفتن طول رشته استفاده کنیم.

</div>

```python
n = len(fruit)
n
```

<div dir="rtl" style="text-align: right;">

برای گرفتن آخرین حرف رشته ممکن است وسوسه شوید این را بنویسید:

</div>

```python
%%expect IndexError

fruit[n]
```

<div dir="rtl" style="text-align: right;">

اما این باعث <code dir="ltr">IndexError</code> می‌شود، چون در <code dir="ltr">'banana'</code> حرفی با اندیس ۶ وجود ندارد. چون شمارش را از <code dir="ltr">0</code> شروع کردیم، شش حرف از <code dir="ltr">0</code> تا <code dir="ltr">5</code> شماره‌گذاری شده‌اند. برای گرفتن آخرین نویسه باید <code dir="ltr">1</code> را از <code dir="ltr">n</code> کم کنید:

</div>

```python
fruit[n-1]
```

<div dir="rtl" style="text-align: right;">

اما راه آسان‌تری هم هست. برای گرفتن آخرین حرف یک رشته می‌توانید از اندیس منفی استفاده کنید که از انتها به عقب می‌شمارد.

</div>

```python
fruit[-1]
```

<div dir="rtl" style="text-align: right;">

اندیس <code dir="ltr">-1</code> آخرین حرف را انتخاب می‌کند، <code dir="ltr">-2</code> یکی مانده به آخر را، و به همین ترتیب.

</div>

## ۸.۲. برش‌های رشته

یک قطعه از رشته **برش** (slice) نامیده می‌شود. انتخاب یک برش شبیه انتخاب یک نویسه است.

</div>

```python
fruit = 'banana'
fruit[0:3]
```

<div dir="rtl" style="text-align: right;">

عملگر <code dir="ltr">[n:m]</code> بخشی از رشته را از نویسه‌ی <code dir="ltr">n</code>ام تا نویسه‌ی <code dir="ltr">m</code>ام برمی‌گرداند، با احتساب اولی و بدون احتساب دومی. این رفتار برخلاف شهود است، اما شاید کمک کند تصور کنید اندیس‌ها *بین* نویسه‌ها اشاره می‌کنند، مثل شکل زیر:

(در کتاب اصلی اینجا نمودار رسم شده است: رشته‌ی <code dir="ltr">' b a n a n a '</code> با اندیس‌های <code dir="ltr">0</code> تا <code dir="ltr">6</code> که با خط‌چین به مرز بین نویسه‌ها اشاره می‌کنند.)

برای مثال، برش <code dir="ltr">[3:6]</code> حرف‌های <code dir="ltr">ana</code> را انتخاب می‌کند، یعنی <code dir="ltr">6</code> به‌عنوان بخشی از یک برش مجاز است، اما به‌عنوان اندیس مجاز نیست.

اگر اندیس اول را حذف کنید، برش از ابتدای رشته شروع می‌شود.

</div>

```python
fruit[:3]
```

<div dir="rtl" style="text-align: right;">

اگر اندیس دوم را حذف کنید، برش تا انتهای رشته ادامه می‌یابد:

</div>

```python
fruit[3:]
```

<div dir="rtl" style="text-align: right;">

اگر اندیس اول بزرگ‌تر یا مساوی اندیس دوم باشد، نتیجه یک **رشته‌ی تهی** (empty string) است که با دو علامت نقل‌قول نشان داده می‌شود:

</div>

```python
fruit[3:3]
```

<div dir="rtl" style="text-align: right;">

رشته‌ی تهی هیچ نویسه‌ای ندارد و طولش ۰ است.

در ادامه‌ی همین مثال، فکر می‌کنید <code dir="ltr">fruit[:]</code> چه معنایی دارد؟ امتحان کنید و ببینید.

</div>

```python
fruit[:]
```

<div dir="rtl" style="text-align: right;">

## ۸.۳. رشته‌ها تغییرناپذیرند

وسوسه‌انگیز است که عملگر <code dir="ltr">[]</code> را در سمت چپ یک نسبت‌دهی به کار ببریم، با این قصد که یک نویسه از رشته را تغییر دهیم، مثل این:

</div>

```python
%%expect TypeError

greeting = 'Hello, world!'
greeting[0] = 'J'
```

<div dir="rtl" style="text-align: right;">

نتیجه یک <code dir="ltr">TypeError</code> است. در پیام خطا، «object» همان رشته و «item» نویسه‌ای است که سعی کردیم نسبت دهیم. فعلاً **شیء** (object) همان مقدار است، اما این تعریف را بعداً دقیق‌تر می‌کنیم.

دلیل این خطا این است که رشته‌ها **تغییرناپذیر** (immutable) هستند، یعنی نمی‌توانید رشته‌ای را که وجود دارد تغییر دهید. بهترین کاری که می‌توانید بکنید ساختن یک رشته‌ی جدید است که گونه‌ای از رشته‌ی اصلی باشد.

</div>

```python
new_greeting = 'J' + greeting[1:]
new_greeting
```

<div dir="rtl" style="text-align: right;">

این مثال یک حرف اول جدید را به یک برش از <code dir="ltr">greeting</code> الحاق می‌کند. روی رشته‌ی اصلی هیچ اثری ندارد.

</div>

```python
greeting
```

<div dir="rtl" style="text-align: right;">

## ۸.۴. مقایسه‌ی رشته‌ها

عملگرهای رابطه‌ای روی رشته‌ها هم کار می‌کنند. برای اینکه ببینیم دو رشته برابرند، می‌توانیم از عملگر <code dir="ltr">==</code> استفاده کنیم.

</div>

```python
word = 'banana'

if word == 'banana':
    print('All right, banana.')
```

<div dir="rtl" style="text-align: right;">

عملیات رابطه‌ای دیگر برای مرتب کردن کلمه‌ها به ترتیب الفبایی مفیدند:

</div>

```python
def compare_word(word):
    if word < 'banana':
        print(word, 'comes before banana.')
    elif word > 'banana':
        print(word, 'comes after banana.')
    else:
        print('All right, banana.')
```

```python
compare_word('apple')
```

<div dir="rtl" style="text-align: right;">

پایتون حروف بزرگ و کوچک را مثل آدم‌ها در نظر نمی‌گیرد. همه‌ی حروف بزرگ قبل از همه‌ی حروف کوچک می‌آیند، پس:

</div>

```python
compare_word('Pineapple')
```

<div dir="rtl" style="text-align: right;">

برای حل این مشکل می‌توانیم رشته‌ها را قبل از مقایسه به یک قالب استاندارد، مثلاً همه حروف کوچک، تبدیل کنیم. اگر باید از خودتان در برابر مردی که مسلح به یک آناناس است دفاع کنید، این را به خاطر داشته باشید.

</div>

<div dir="rtl" style="text-align: right;">

## ۸.۵. متدهای رشته

رشته‌ها متدهایی دارند که عملیات مفید گوناگونی را انجام می‌دهند. متد شبیه تابع است – آرگومان می‌گیرد و مقداری برمی‌گرداند – اما نحوه‌ی نوشتنش فرق دارد. برای مثال، متد <code dir="ltr">upper</code> یک رشته را می‌گیرد و رشته‌ی جدیدی با همه‌ی حروف بزرگ برمی‌گرداند.

به‌جای نحو تابعی <code dir="ltr">upper(word)</code>، از نحو متدی <code dir="ltr">word.upper()</code> استفاده می‌کند.

</div>

```python
word = 'banana'
new_word = word.upper()
new_word
```

<div dir="rtl" style="text-align: right;">

این استفاده از عملگر نقطه نام متد، یعنی <code dir="ltr">upper</code>، و نام رشته‌ای را که متد روی آن اعمال می‌شود، یعنی <code dir="ltr">word</code>، مشخص می‌کند. پرانتزهای خالی نشان می‌دهند این متد آرگومانی نمی‌گیرد.

فراخوانی یک متد **احضار** (invocation) نامیده می‌شود؛ در این مورد می‌گوییم <code dir="ltr">upper</code> را روی <code dir="ltr">word</code> احضار می‌کنیم.

</div>

<div dir="rtl" style="text-align: right;">

## ۸.۶. نوشتن فایل‌ها

عملگرها و متدهای رشته برای خواندن و نوشتن فایل‌های متنی مفیدند. به‌عنوان مثال با متن رمان *Dracula* نوشته‌ی برم استوکر کار می‌کنیم که از پروژه‌ی گوتنبرگ (<https://www.gutenberg.org/ebooks/345>) در دسترس است.

</div>

```python
download('https://raw.githubusercontent.com/Alireza-Akhavan/ThinkPython/main/pg345.txt');
```

<div dir="rtl" style="text-align: right;">

کتاب را در یک فایل متنی ساده به نام <code dir="ltr">pg345.txt</code> دانلود کرده‌ام که می‌توانیم این‌طور برای خواندن بازش کنیم:

آرگومان کلیدواژه‌ای <code dir="ltr">encoding='utf-8'</code> در کتاب چاپی نیست. فقط در ویندوز لازم است.

</div>

```python
reader = open('pg345.txt', encoding='utf-8')
```

<div dir="rtl" style="text-align: right;">

این فایل علاوه بر متن کتاب، بخشی در ابتدا با اطلاعاتی درباره‌ی کتاب و بخشی در انتها با اطلاعاتی درباره‌ی مجوز دارد. قبل از پردازش متن، می‌توانیم با پیدا کردن خط‌های ویژه‌ی ابتدا و انتها که با <code dir="ltr">'***'</code> شروع می‌شوند، این مطالب اضافه را حذف کنیم.

تابع زیر یک خط را می‌گیرد و بررسی می‌کند که آیا یکی از خط‌های ویژه است یا نه. از متد <code dir="ltr">startswith</code> استفاده می‌کند که بررسی می‌کند رشته با دنباله‌ی مشخصی از نویسه‌ها شروع می‌شود یا نه.

</div>

```python
def is_special_line(line):
    return line.startswith('*** ')
```

<div dir="rtl" style="text-align: right;">

می‌توانیم از این تابع برای پیمایش خط‌های فایل و نمایش فقط خط‌های ویژه استفاده کنیم.

</div>

```python
for line in reader:
    if is_special_line(line):
        print(line.strip())
```

<div dir="rtl" style="text-align: right;">

حالا فایل جدیدی به نام <code dir="ltr">pg345_cleaned.txt</code> می‌سازیم که فقط متن کتاب را دارد. برای اینکه دوباره کتاب را پیمایش کنیم، باید دوباره برای خواندن بازش کنیم. و برای نوشتن یک فایل جدید می‌توانیم آن را برای نوشتن باز کنیم.

</div>

```python
reader = open('pg345.txt', encoding='utf-8')
writer = open('pg345_cleaned.txt', 'w', encoding='utf-8')
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">open</code> یک پارامتر اختیاری می‌گیرد که «حالت» (mode) را مشخص می‌کند – در این مثال <code dir="ltr">'w'</code> نشان می‌دهد فایل را برای نوشتن باز می‌کنیم. اگر فایل وجود نداشته باشد ساخته می‌شود؛ اگر از قبل وجود داشته باشد، محتویاتش جایگزین می‌شود.

در قدم اول، فایل را پیمایش می‌کنیم تا اولین خط ویژه را پیدا کنیم.

</div>

```python
for line in reader:
    if is_special_line(line):
        break
```

<div dir="rtl" style="text-align: right;">

دستور <code dir="ltr">break</code> از حلقه «بیرون می‌پرد» – یعنی باعث می‌شود حلقه بلافاصله، قبل از رسیدن به انتهای فایل، تمام شود.

وقتی حلقه خارج می‌شود، <code dir="ltr">line</code> شامل خط ویژه‌ای است که شرط را درست کرد.

</div>

```python
line
```

<div dir="rtl" style="text-align: right;">

چون <code dir="ltr">reader</code> مکان خود در فایل را نگه می‌دارد، می‌توانیم از یک حلقه‌ی دوم برای ادامه از همان‌جایی که مانده‌ایم استفاده کنیم.

حلقه‌ی زیر بقیه‌ی فایل را، خط به خط، می‌خواند. وقتی خط ویژه‌ای را که پایان متن را نشان می‌دهد پیدا کند، از حلقه بیرون می‌پرد. در غیر این صورت خط را در فایل خروجی می‌نویسد.

</div>

```python
for line in reader:
    if is_special_line(line):
        break
    writer.write(line)
```

<div dir="rtl" style="text-align: right;">

وقتی این حلقه خارج می‌شود، <code dir="ltr">line</code> شامل خط ویژه‌ی دوم است.

</div>

```python
line
```

<div dir="rtl" style="text-align: right;">

در این نقطه <code dir="ltr">reader</code> و <code dir="ltr">writer</code> هنوز باز هستند، یعنی می‌توانیم همچنان از <code dir="ltr">reader</code> خط بخوانیم یا در <code dir="ltr">writer</code> خط بنویسیم. برای اینکه نشان دهیم کارمان تمام شده، می‌توانیم با احضار متد <code dir="ltr">close</code> هر دو فایل را ببندیم.

</div>

```python
reader.close()
writer.close()
```

<div dir="rtl" style="text-align: right;">

برای بررسی اینکه این فرایند موفق بوده، می‌توانیم چند خط اول فایل جدیدی را که ساختیم بخوانیم.

</div>

```python
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    line = line.strip()
    if len(line) > 0:
        print(line)
    if line.endswith('Stoker'):
        break
```

<div dir="rtl" style="text-align: right;">

متد <code dir="ltr">endswith</code> بررسی می‌کند رشته با دنباله‌ی مشخصی از نویسه‌ها تمام می‌شود یا نه.

</div>

## ۸.۷. جست‌وجو و جایگزینی

در ترجمه‌ی ایسلندیِ *Dracula* در سال ۱۹۰۱، نام یکی از شخصیت‌ها از «Jonathan» به «Thomas» تغییر داده شد. برای انجام همین تغییر در نسخه‌ی انگلیسی، می‌توانیم کتاب را پیمایش کنیم، از متد <code dir="ltr">replace</code> برای جایگزین کردن یک نام با نام دیگر استفاده کنیم و نتیجه را در یک فایل جدید بنویسیم.

با شمردن خط‌های نسخه‌ی پاک‌شده‌ی فایل شروع می‌کنیم.

</div>

```python
total = 0
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    total += 1
    
total
```

<div dir="rtl" style="text-align: right;">

برای دیدن اینکه آیا یک خط شامل «Jonathan» است، می‌توانیم از عملگر <code dir="ltr">in</code> استفاده کنیم که بررسی می‌کند این دنباله‌ی نویسه‌ها در هر جای خط وجود دارد یا نه.

</div>

```python
total = 0
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    if 'Jonathan' in line:
        total += 1

total
```

<div dir="rtl" style="text-align: right;">

۱۹۹ خط وجود دارد که این نام را دارند، اما این دقیقاً کل دفعاتی که ظاهر می‌شود نیست، چون ممکن است در یک خط بیش از یک بار بیاید. برای گرفتن مجموع می‌توانیم از متد <code dir="ltr">count</code> استفاده کنیم که تعداد دفعاتی را که یک دنباله در رشته ظاهر می‌شود برمی‌گرداند.

</div>

```python
total = 0
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    total += line.count('Jonathan')

total
```

```python
writer = open('pg345_replaced.txt', 'w', encoding='utf-8')

for line in open('pg345_cleaned.txt', encoding='utf-8'):
    line = line.replace('Jonathan', 'Thomas')
    writer.write(line)
```

```python
total = 0
for line in open('pg345_replaced.txt', encoding='utf-8'):
    total += line.count('Thomas')

total
```

## ۸.۸. عبارت‌های منظم

اگر دقیقاً بدانیم دنبال چه دنباله‌ای از نویسه‌ها هستیم، می‌توانیم با عملگر <code dir="ltr">in</code> پیدایش کنیم و با متد <code dir="ltr">replace</code> جایگزینش کنیم. اما ابزار دیگری به نام **عبارت منظم** (regular expression) وجود دارد که همین عملیات – و خیلی بیشتر – را انجام می‌دهد.

برای نمایش، با یک مثال ساده شروع می‌کنیم و کم‌کم بالا می‌رویم. فرض کنید باز هم می‌خواهیم همه‌ی خط‌هایی را که یک کلمه‌ی خاص دارند پیدا کنیم. برای تنوع، دنبال اشاره به شخصیت اصلی کتاب، کنت دراکولا، می‌گردیم. این خطی است که به او اشاره می‌کند.

</div>

```python
text = "I am Dracula; and I bid you welcome, Mr. Harker, to my house."
```

<div dir="rtl" style="text-align: right;">

و این **الگویی** (pattern) است که برای جست‌وجو استفاده می‌کنیم.

</div>

```python
pattern = 'Dracula'
```

<div dir="rtl" style="text-align: right;">

ماژولی به نام <code dir="ltr">re</code> توابع مربوط به عبارت‌های منظم را فراهم می‌کند. می‌توانیم آن را این‌طور وارد کنیم و با تابع <code dir="ltr">search</code> بررسی کنیم که الگو در متن هست یا نه.

</div>

```python
import re

result = re.search(pattern, text)
result
```

<div dir="rtl" style="text-align: right;">

اگر الگو در متن ظاهر شود، <code dir="ltr">search</code> یک شیء <code dir="ltr">Match</code> برمی‌گرداند که نتایج جست‌وجو را دارد. در میان اطلاعات دیگر، متغیری به نام <code dir="ltr">string</code> دارد که متنِ جست‌وجوشده را دارد.

</div>

```python
result.string
```

<div dir="rtl" style="text-align: right;">

متدی به نام <code dir="ltr">group</code> هم دارد که بخشی از متن را که با الگو تطابق داشته برمی‌گرداند.

</div>

```python
result.group()
```

<div dir="rtl" style="text-align: right;">

و متدی به نام <code dir="ltr">span</code> دارد که اندیس‌های شروع و پایان الگو را در متن برمی‌گرداند.

</div>

```python
result.span()
```

<div dir="rtl" style="text-align: right;">

اگر الگو در متن نباشد، مقدار بازگشتی <code dir="ltr">search</code> برابر <code dir="ltr">None</code> است.

</div>

```python
result = re.search('Count', text)
print(result)
```

<div dir="rtl" style="text-align: right;">

پس می‌توانیم با بررسی اینکه نتیجه <code dir="ltr">None</code> است یا نه، تشخیص دهیم جست‌وجو موفق بوده یا نه.

</div>

```python
result == None
```

<div dir="rtl" style="text-align: right;">

با کنار هم گذاشتن همه‌ی این‌ها، این تابعی است که خط‌های کتاب را پیمایش می‌کند تا به خطی برسد که با الگوی داده‌شده تطابق دارد و شیء <code dir="ltr">Match</code> را برمی‌گرداند.

</div>

```python
def find_first(pattern):
    for line in open('pg345_cleaned.txt', encoding='utf-8'):
        result = re.search(pattern, line)
        if result != None:
            return result
```

<div dir="rtl" style="text-align: right;">

می‌توانیم از آن برای پیدا کردن اولین اشاره به یک شخصیت استفاده کنیم.

</div>

```python
result = find_first('Harker')
result.string
```

<div dir="rtl" style="text-align: right;">

در این مثال لازم نبود از عبارت‌های منظم استفاده کنیم – می‌توانستیم همین کار را با عملگر <code dir="ltr">in</code> راحت‌تر انجام دهیم. اما عبارت‌های منظم کارهایی می‌کنند که عملگر <code dir="ltr">in</code> نمی‌تواند.

برای مثال، اگر الگو شامل نویسه‌ی خط عمودی <code dir="ltr">'|'</code> باشد، می‌تواند با دنباله‌ی سمت چپ یا دنباله‌ی سمت راست تطابق پیدا کند. فرض کنید می‌خواهیم اولین اشاره به Mina Murray را در کتاب پیدا کنیم، اما مطمئن نیستیم با نام کوچک به او اشاره شده یا با نام خانوادگی. می‌توانیم از الگوی زیر استفاده کنیم که با هر دو نام تطابق دارد.

</div>

```python
pattern = 'Mina|Murray'
result = find_first(pattern)
result.string
```

<div dir="rtl" style="text-align: right;">

می‌توانیم از الگویی مثل این برای دیدن اینکه یک شخصیت چند بار با هر کدام از دو نامش آمده استفاده کنیم. این تابعی است که کتاب را پیمایش می‌کند و تعداد خط‌هایی را که با الگوی داده‌شده تطابق دارند می‌شمارد.

</div>

```python
def count_matches(pattern):
    count = 0
    for line in open('pg345_cleaned.txt', encoding='utf-8'):
        result = re.search(pattern, line)
        if result != None:
            count += 1
    return count
```

<div dir="rtl" style="text-align: right;">

حالا ببینیم Mina چند بار ذکر شده است.

</div>

```python
count_matches('Mina|Murray')
```

<div dir="rtl" style="text-align: right;">

نویسه‌ی ویژه‌ی <code dir="ltr">'^'</code> با ابتدای رشته تطابق دارد، پس می‌توانیم خطی را که با یک الگوی مشخص شروع می‌شود پیدا کنیم.

</div>

```python
result = find_first('^Dracula')
result.string
```

<div dir="rtl" style="text-align: right;">

و نویسه‌ی ویژه‌ی <code dir="ltr">'$'</code> با انتهای رشته تطابق دارد، پس می‌توانیم خطی را که با یک الگوی مشخص تمام می‌شود پیدا کنیم (با نادیده گرفتن خط جدید انتها).

</div>

```python
result = find_first('Harker$')
result.string
```

## ۸.۹. جایگزینی در رشته

برم استوکر در ایرلند به دنیا آمد و وقتی *Dracula* در ۱۸۹۷ منتشر شد، در انگلستان زندگی می‌کرد. پس انتظار داریم از املای بریتانیایی کلماتی مثل «centre» و «colour» استفاده کند. برای بررسی، می‌توانیم از الگوی زیر استفاده کنیم که با «centre» یا املای آمریکایی «center» تطابق دارد.

</div>

```python
pattern = 'cent(er|re)'
```

<div dir="rtl" style="text-align: right;">

در این الگو، پرانتزها بخشی از الگو را که خط عمودی روی آن اعمال می‌شود محصور می‌کنند. پس این الگو با دنباله‌ای که با <code dir="ltr">'cent'</code> شروع می‌شود و به <code dir="ltr">'er'</code> یا <code dir="ltr">'re'</code> ختم می‌شود تطابق دارد.

می‌توانیم بررسی کنیم که آیا از املای بریتانیایی «colour» هم استفاده کرده است. الگوی زیر از نویسه‌ی ویژه‌ی <code dir="ltr">'?'</code> استفاده می‌کند که یعنی نویسه‌ی قبلی اختیاری است.

```python
pattern = 'colou?r'
```

<div dir="rtl" style="text-align: right;">

این الگو با «colour» که <code dir="ltr">'u'</code> دارد یا «color» بدون آن تطابق دارد.

</div>

```python
result = find_first(pattern)
line = result.string
line
```

<div dir="rtl" style="text-align: right;">

باز هم، همان‌طور که انتظار می‌رفت، از املای بریتانیایی استفاده کرده است.

حالا فرض کنید می‌خواهیم نسخه‌ای از کتاب با املای آمریکایی تولید کنیم. می‌توانیم از تابع <code dir="ltr">sub</code> در ماژول <code dir="ltr">re</code> استفاده کنیم که **جایگزینی رشته** (string substitution) انجام می‌دهد.

</div>

```python
re.sub(pattern, 'color', line)
```

<div dir="rtl" style="text-align: right;">

آرگومان اول الگویی است که می‌خواهیم پیدا و جایگزین کنیم، دومی چیزی است که می‌خواهیم با آن جایگزین شود، و سومی رشته‌ای است که می‌خواهیم در آن جست‌وجو کنیم. در نتیجه می‌بینید که «colour» با «color» جایگزین شده است.

</div>

```python
# I used this function to search for lines to use as examples

def all_matches(pattern):
    for line in open('pg345_cleaned.txt', encoding='utf-8'):
        result = re.search(pattern, line)
        if result:
            print(line.strip())
```

## ۸.۱۱. واژه‌نامه

**دنباله (sequence):** مجموعه‌ای مرتب از مقدارها که هر مقدار با یک اندیس صحیح مشخص می‌شود.

**نویسه (character):** عنصری از یک رشته، از جمله حرف، عدد و نماد.

**اندیس (index):** مقدار صحیحی که برای انتخاب یک عنصر در یک دنباله، مثل یک نویسه در رشته، به کار می‌رود. در پایتون اندیس‌ها از <code dir="ltr">0</code> شروع می‌شوند.

**برش (slice):** بخشی از رشته که با یک بازه از اندیس‌ها مشخص می‌شود.

**رشته‌ی تهی (empty string):** رشته‌ای که هیچ نویسه‌ای ندارد و طولش <code dir="ltr">0</code> است.

**شیء (object):** چیزی که یک متغیر می‌تواند به آن اشاره کند. یک شیء نوع و مقدار دارد.

**تغییرناپذیر (immutable):** اگر عناصر یک شیء قابل تغییر نباشند، شیء تغییرناپذیر است.

**احضار (invocation):** یک عبارت – یا بخشی از یک عبارت – که یک متد را فراخوانی می‌کند.

**عبارت منظم (regular expression):** دنباله‌ای از نویسه‌ها که یک الگوی جست‌وجو را تعریف می‌کند.

**الگو (pattern):** قاعده‌ای که شرط‌هایی را که یک رشته باید برآورده کند تا تطابق داشته باشد مشخص می‌کند.

**جایگزینی رشته (string substitution):** جایگزین کردن یک رشته، یا بخشی از یک رشته، با رشته‌ای دیگر.

**دستور پوسته (shell command):** عبارتی در یک زبان پوسته، که زبانی برای تعامل با سیستم‌عامل است.

</div>

عبارت منظم اغلب حرف <code dir="ltr">r</code> قبل از علامت نقل‌قول دارد که نشان می‌دهد «رشته‌ی خام» (raw string) است. برای اطلاعات بیشتر از یک دستیار مجازی بپرسید: «What is a raw string in Python?»

</div>

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```
