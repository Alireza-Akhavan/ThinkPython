## ۳.۱. تعریف توابع جدید

**تعریف تابع** (function definition) نام یک تابع جدید و دنباله‌ی دستورهایی را که هنگام فراخوانی تابع اجرا می‌شوند مشخص می‌کند. این یک مثال است:

</div>

````python
def print_message():
    print("Functions let us name a group of statements.")
    print("Then we can run them whenever we like.")
````

<div dir="rtl" style="text-align: right;">

<code dir="ltr">def</code> کلمه‌ی کلیدی‌ای است که نشان می‌دهد این یک تعریف تابع است. نام تابع <code dir="ltr">print_message</code> است. هر چیزی که نام مجاز برای متغیر باشد، نام مجاز برای تابع هم هست.

پرانتزهای خالی بعد از نام نشان می‌دهند که این تابع آرگومانی نمی‌گیرد.

خط اول تعریف تابع **سرآیند** (header) نامیده می‌شود – بقیه‌ی آن **بدنه** (body) است. سرآیند باید با دونقطه (colon) تمام شود و بدنه باید تورفتگی (indentation) داشته باشد. طبق قرارداد، تورفتگی همیشه چهار فاصله است. بدنه‌ی این تابع دو دستور <code dir="ltr">print</code> است؛ به‌طور کلی بدنه‌ی یک تابع می‌تواند هر تعداد دستور از هر نوع داشته باشد.

تعریف یک تابع یک **شیء تابع** (function object) می‌سازد که می‌توانیم آن را این‌طور نمایش دهیم.

</div>

````python
print_message
````

<div dir="rtl" style="text-align: right;">

خروجی نشان می‌دهد که <code dir="ltr">print_message</code> تابعی است که آرگومانی نمی‌گیرد. <code dir="ltr">__main__</code> نام ماژولی است که <code dir="ltr">print_message</code> را در خود دارد.

حالا که یک تابع تعریف کرده‌ایم، می‌توانیم آن را به همان روشی که توابع داخلی را فراخوانی می‌کنیم فراخوانی کنیم.

</div>

````python
print_message()
````

<div dir="rtl" style="text-align: right;">

وقتی تابع اجرا می‌شود، دستورهای بدنه را اجرا می‌کند که این دو خط متن را نمایش می‌دهند.

</div>

<div dir="rtl" style="text-align: right;">

## ۳.۲. پارامترها

بعضی از توابعی که دیده‌ایم آرگومان لازم دارند؛ برای مثال، وقتی <code dir="ltr">abs</code> را فراخوانی می‌کنید یک عدد به‌عنوان آرگومان می‌دهید. بعضی توابع بیش از یک آرگومان می‌گیرند؛ برای مثال <code dir="ltr">math.pow</code> دو آرگومان می‌گیرد، پایه و توان.

این تعریف تابعی است که یک آرگومان می‌گیرد.

</div>

````python
def print_twice(string):
    print(string)
    print(string)
````

<div dir="rtl" style="text-align: right;">

نام متغیر داخل پرانتز یک **پارامتر** (parameter) است. وقتی تابع فراخوانی می‌شود، مقدار آرگومان به پارامتر نسبت داده می‌شود. برای مثال، می‌توانیم <code dir="ltr">print_twice</code> را این‌طور فراخوانی کنیم.

</div>

````python
print_twice('Hello, ')
````

<div dir="rtl" style="text-align: right;">

اجرای این تابع همان اثری را دارد که نسبت دادن آرگومان به پارامتر و سپس اجرای بدنه‌ی تابع، مثل این:

</div>

````python
string = 'Hello, '
print(string)
print(string)
````

<div dir="rtl" style="text-align: right;">

می‌توانید از یک متغیر هم به‌عنوان آرگومان استفاده کنید.

</div>

````python
line = 'Hello, '
print_twice(line)
````

<div dir="rtl" style="text-align: right;">

در این مثال، مقدار <code dir="ltr">line</code> به پارامتر <code dir="ltr">string</code> نسبت داده می‌شود.

</div>

<div dir="rtl" style="text-align: right;">

## ۳.۳. فراخوانی توابع

وقتی تابعی را تعریف کردید، می‌توانید از آن داخل تابع دیگری استفاده کنید. برای نشان دادن این، توابعی می‌نویسیم که یک «بند» متنی را چاپ می‌کنند. بند ما این است: چهار بار <code dir="ltr">Spam, </code> در خط اول، چهار بار در خط دوم، دو بار در خط سوم، یک جمله‌ی کوتاه در خط چهارم و دوباره دو بار <code dir="ltr">Spam, </code> در خط پنجم.

با تابع زیر شروع می‌کنیم که دو پارامتر می‌گیرد.

</div>

````python
def repeat(word, n):
    print(word * n)
````

<div dir="rtl" style="text-align: right;">

می‌توانیم از این تابع برای چاپ خط اول استفاده کنیم، این‌طور.

</div>

````python
spam = 'Spam, '
repeat(spam, 4)
````

<div dir="rtl" style="text-align: right;">

برای نمایش دو خط اول، می‌توانیم تابع جدیدی تعریف کنیم که از <code dir="ltr">repeat</code> استفاده می‌کند.

</div>

````python
def first_two_lines():
    repeat(spam, 4)
    repeat(spam, 4)
````

<div dir="rtl" style="text-align: right;">

و بعد آن را این‌طور فراخوانی کنیم.

</div>

````python
first_two_lines()
````

<div dir="rtl" style="text-align: right;">

برای نمایش سه خط آخر، می‌توانیم تابع دیگری تعریف کنیم که آن هم از <code dir="ltr">repeat</code> استفاده می‌کند.

</div>

````python
def last_three_lines():
    repeat(spam, 2)
    print('(Hooray for Spam!)')
    repeat(spam, 2)
````

````python
last_three_lines()
````

<div dir="rtl" style="text-align: right;">

سرانجام، می‌توانیم همه را با یک تابع که کل بند را چاپ می‌کند کنار هم بیاوریم.

</div>

````python
def print_verse():
    first_two_lines()
    last_three_lines()
````

````python
print_verse()
````

<div dir="rtl" style="text-align: right;">

وقتی <code dir="ltr">print_verse</code> را اجرا می‌کنیم، <code dir="ltr">first_two_lines</code> را فراخوانی می‌کند، که <code dir="ltr">repeat</code> را فراخوانی می‌کند، که <code dir="ltr">print</code> را فراخوانی می‌کند. این تعداد زیادی تابع است.

البته می‌شد همین کار را با توابع کمتری انجام داد، اما هدف این مثال نشان دادن این است که توابع چگونه می‌توانند با هم کار کنند.

</div>

<div dir="rtl" style="text-align: right;">

## ۳.۴. تکرار

اگر بخواهیم بیش از یک بند را نمایش دهیم، می‌توانیم از دستور <code dir="ltr">for</code> استفاده کنیم. این یک مثال ساده است.

</div>

````python
for i in range(2):
    print(i)
````

<div dir="rtl" style="text-align: right;">

خط اول یک سرآیند است که به دونقطه ختم می‌شود. خط دوم بدنه است که باید تورفتگی داشته باشد.

سرآیند با کلمه‌ی کلیدی <code dir="ltr">for</code>، یک متغیر جدید به نام <code dir="ltr">i</code> و کلمه‌ی کلیدی دیگری، <code dir="ltr">in</code>، شروع می‌شود. از تابع <code dir="ltr">range</code> برای ساختن دنباله‌ای از دو مقدار استفاده می‌کند که <code dir="ltr">0</code> و <code dir="ltr">1</code> هستند. در پایتون، وقتی شروع به شمردن می‌کنیم، معمولاً از <code dir="ltr">0</code> شروع می‌کنیم.

وقتی دستور <code dir="ltr">for</code> اجرا می‌شود، اولین مقدار از <code dir="ltr">range</code> را به <code dir="ltr">i</code> نسبت می‌دهد و بعد تابع <code dir="ltr">print</code> را در بدنه اجرا می‌کند که <code dir="ltr">0</code> را نمایش می‌دهد.

وقتی به انتهای بدنه می‌رسد، دوباره به سرآیند برمی‌گردد؛ به همین دلیل به این دستور **حلقه** (loop) می‌گویند. بار دوم که حلقه اجرا می‌شود، مقدار بعدی را از <code dir="ltr">range</code> به <code dir="ltr">i</code> نسبت می‌دهد و آن را نمایش می‌دهد. سپس، چون آن آخرین مقدار از <code dir="ltr">range</code> است، حلقه تمام می‌شود.

این‌طور می‌توانیم از یک حلقه‌ی <code dir="ltr">for</code> برای چاپ دو بند استفاده کنیم.

</div>

````python
for i in range(2):
    print("Verse", i)
    print_verse()
    print()
````

<div dir="rtl" style="text-align: right;">

می‌توانید یک حلقه‌ی <code dir="ltr">for</code> را داخل یک تابع بگذارید. برای مثال، <code dir="ltr">print_n_verses</code> پارامتری به نام <code dir="ltr">n</code> می‌گیرد که باید عدد صحیح باشد و تعداد بندهای داده‌شده را نمایش می‌دهد.

</div>

````python
def print_n_verses(n):
    for i in range(n):
        print_verse()
        print()
````

<div dir="rtl" style="text-align: right;">

در این مثال از <code dir="ltr">i</code> در بدنه‌ی حلقه استفاده نمی‌کنیم، اما به هر حال باید یک نام متغیر در سرآیند باشد.

</div>

<div dir="rtl" style="text-align: right;">

## ۳.۵. متغیرها و پارامترها محلی‌اند

وقتی داخل یک تابع متغیری می‌سازید، آن متغیر **محلی** (local) است؛ یعنی فقط داخل همان تابع وجود دارد. برای مثال، تابع زیر دو آرگومان می‌گیرد، آن‌ها را الحاق می‌کند و نتیجه را دو بار چاپ می‌کند.

</div>

````python
def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
````

<div dir="rtl" style="text-align: right;">

این هم مثالی که از آن استفاده می‌کند:

</div>

````python
line1 = 'Python is '
line2 = 'fun to learn.'
cat_twice(line1, line2)
````

<div dir="rtl" style="text-align: right;">

وقتی <code dir="ltr">cat_twice</code> اجرا می‌شود، یک متغیر محلی به نام <code dir="ltr">cat</code> می‌سازد که وقتی تابع تمام می‌شود از بین می‌رود. اگر سعی کنیم آن را نمایش دهیم، <code dir="ltr">NameError</code> می‌گیریم:

</div>

````python
%%expect NameError

print(cat)
````

<div dir="rtl" style="text-align: right;">

بیرون از تابع، <code dir="ltr">cat</code> تعریف نشده است.

پارامترها هم محلی‌اند. برای مثال، بیرون از <code dir="ltr">cat_twice</code> چیزی به نام <code dir="ltr">part1</code> یا <code dir="ltr">part2</code> وجود ندارد.

</div>

<div dir="rtl" style="text-align: right;">

## ۳.۶. نمودارهای پشته

برای دنبال کردن اینکه کدام متغیرها را کجا می‌توان به کار برد، گاهی مفید است یک **نمودار پشته** (stack diagram) رسم کنیم. مثل نمودارهای حالت، نمودارهای پشته مقدار هر متغیر را نشان می‌دهند، اما تابعی را که هر متغیر به آن تعلق دارد هم نشان می‌دهند.

هر تابع با یک **قاب** (frame) نمایش داده می‌شود. قاب جعبه‌ای است که نام تابع بیرون آن و پارامترها و متغیرهای محلی تابع داخل آن نوشته شده‌اند.

(در کتاب اصلی این‌جا نمودار پشته‌ی مثال قبل رسم شده است: چهار قاب روی هم، از پایین <code dir="ltr">__main__</code> با متغیرهای <code dir="ltr">line1</code> و <code dir="ltr">line2</code>، سپس <code dir="ltr">cat_twice</code> با <code dir="ltr">part1</code>، <code dir="ltr">part2</code> و <code dir="ltr">cat</code>، سپس <code dir="ltr">print_twice</code> با <code dir="ltr">string</code>، و بالای همه <code dir="ltr">print</code>.)

قاب‌ها در یک پشته چیده شده‌اند که نشان می‌دهد کدام تابع کدام را فراخوانی کرده و همین‌طور تا آخر. اگر از پایین بخوانید، <code dir="ltr">print</code> را <code dir="ltr">print_twice</code> فراخوانی کرده، که <code dir="ltr">cat_twice</code> آن را فراخوانی کرده، که <code dir="ltr">__main__</code> آن را فراخوانی کرده – <code dir="ltr">__main__</code> نام خاصی برای بالاترین قاب است. وقتی بیرون از هر تابعی یک متغیر می‌سازید، آن متغیر به <code dir="ltr">__main__</code> تعلق دارد.

در قاب <code dir="ltr">print</code>، علامت سؤال نشان می‌دهد که نام پارامتر را نمی‌دانیم. اگر کنجکاوید، از یک دستیار مجازی بپرسید «پارامترهای تابع print در پایتون چیست؟» (What are the parameters of the Python print function?)

</div>

## ۳.۷. ردیابی خطا (Traceback)

وقتی در یک تابع خطای زمان اجرا رخ می‌دهد، پایتون نام تابعی را که در حال اجرا بوده، نام تابعی که آن را فراخوانی کرده و همین‌طور بالاتر در پشته نمایش می‌دهد. برای دیدن یک مثال، نسخه‌ای از <code dir="ltr">print_twice</code> را تعریف می‌کنم که خطا دارد – سعی می‌کند <code dir="ltr">cat</code> را چاپ کند که یک متغیر محلی در تابع دیگری است.

</div>

````python
def print_twice(string):
    print(cat)            # NameError
    print(cat)
````

<div dir="rtl" style="text-align: right;">

</div>

````python
# This cell tells Jupyter to provide detailed debugging information
# when a runtime error occurs, including a traceback.

%xmode Verbose
````

````python
%%expect NameError

cat_twice(line1, line2)
````

## ۳.۸. چرا توابع؟

شاید هنوز روشن نباشد که چرا ارزش دارد زحمت تقسیم یک برنامه به توابع را بکشیم. چند دلیل وجود دارد:

- ساختن یک تابع جدید به شما فرصت می‌دهد گروهی از دستورها را نام‌گذاری کنید که برنامه‌تان را برای خواندن و اشکال‌زدایی آسان‌تر می‌کند.

- توابع می‌توانند با حذف کد تکراری برنامه را کوچک‌تر کنند. بعداً اگر تغییری بدهید، فقط باید آن را در یک جا انجام دهید.

- تقسیم یک برنامه‌ی بلند به توابع به شما اجازه می‌دهد بخش‌ها را یکی‌یکی اشکال‌زدایی کنید و بعد آن‌ها را به یک کل کارآمد وصل کنید.

- توابعی که خوب طراحی شده باشند اغلب برای برنامه‌های زیادی مفیدند. وقتی یکی را نوشتید و اشکال‌زدایی کردید، می‌توانید دوباره از آن استفاده کنید.

## ۳.۹. اشکال‌زدایی

اشکال‌زدایی می‌تواند ناامیدکننده باشد، اما چالش‌برانگیز، جالب و گاهی حتی سرگرم‌کننده هم هست. و یکی از مهم‌ترین مهارت‌هایی است که می‌توانید یاد بگیرید.

از جهاتی اشکال‌زدایی شبیه کار کارآگاهی است. سرنخ‌هایی به شما داده می‌شود و باید رویدادهایی را که به نتایجی که می‌بینید منجر شده‌اند استنباط کنید.

اشکال‌زدایی همچنین شبیه علم تجربی است. وقتی ایده‌ای درباره‌ی اینکه چه چیزی اشتباه پیش می‌رود پیدا کردید، برنامه‌تان را تغییر می‌دهید و دوباره امتحان می‌کنید. اگر فرضیه‌تان درست بود، می‌توانید نتیجه‌ی تغییر را پیش‌بینی کنید و یک قدم به یک برنامه‌ی کارآمد نزدیک‌تر می‌شوید. اگر فرضیه‌تان اشتباه بود، باید فرضیه‌ی جدیدی بسازید.

برای بعضی‌ها برنامه‌نویسی و اشکال‌زدایی یکی هستند؛ یعنی برنامه‌نویسی فرایند تدریجیِ اشکال‌زدایی یک برنامه است تا زمانی که کاری را که می‌خواهید انجام دهد. ایده این است که باید با یک برنامه‌ی کارآمد شروع کنید و تغییرات کوچک بدهید و در حین کار اشکال‌زدایی‌شان کنید.

اگر می‌بینید زمان زیادی را صرف اشکال‌زدایی می‌کنید، این اغلب نشانه‌ی آن است که پیش از شروع آزمایش، کد زیادی می‌نویسید. اگر قدم‌های کوچک‌تری بردارید، ممکن است ببینید سریع‌تر پیش می‌روید.

## ۳.۱۰. واژه‌نامه

**تعریف تابع (function definition):** دستوری که یک تابع می‌سازد.

**سرآیند (header):** خط اول یک تعریف تابع.

**بدنه (body):** دنباله‌ی دستورهایی که داخل یک تعریف تابع هستند.

**شیء تابع (function object):** مقداری که یک تعریف تابع می‌سازد. نام تابع متغیری است که به یک شیء تابع اشاره می‌کند.

**پارامتر (parameter):** نامی که داخل یک تابع برای اشاره به مقداری که به‌عنوان آرگومان داده شده به کار می‌رود.

**حلقه (loop):** دستوری که یک یا چند دستور را، اغلب به‌صورت تکراری، اجرا می‌کند.

**متغیر محلی (local variable):** متغیری که داخل یک تابع تعریف شده و فقط داخل همان تابع قابل دسترسی است.

**نمودار پشته (stack diagram):** نمایشی گرافیکی از پشته‌ای از توابع، متغیرهایشان و مقدارهایی که به آن‌ها اشاره می‌کنند.

**قاب (frame):** جعبه‌ای در نمودار پشته که یک فراخوانی تابع را نشان می‌دهد. متغیرهای محلی و پارامترهای تابع را در خود دارد.

**ردیابی (traceback):** فهرستی از توابعی که در حال اجرا هستند و هنگام رخ دادن یک استثنا چاپ می‌شود.
