بعضی از تمرین‌های این کتاب را می‌توان با مجموعه‌ها مختصر و کارآمد انجام داد. برای مثال، این راه‌حل یکی از تمرین‌های فصل ۱۱ است که از یک دیکشنری برای بررسی اینکه آیا عنصر تکراری در یک دنباله وجود دارد استفاده می‌کند.

</div>

```python
def has_duplicates(t):
    d = {}
    for x in t:
        d[x] = True
    return len(d) < len(t)
```

<div dir="rtl" style="text-align: right;">

این نسخه عنصرهای <code dir="ltr">t</code> را به‌عنوان کلید به یک دیکشنری اضافه می‌کند و بعد بررسی می‌کند که آیا کلیدها از عنصرها کمترند. با مجموعه‌ها می‌توانیم همین تابع را این‌طور بنویسیم.

</div>

```python
def has_duplicates(t):
    s = set(t)
    return len(s) < len(t)
```

در فصل ۱۲ از دیکشنری‌ها برای پیدا کردن کلمه‌هایی که در یک سند هستند اما در فهرست کلمه‌ها نیستند استفاده کردیم. از تابع زیر استفاده کردیم که دو دیکشنری می‌گیرد و دیکشنری جدیدی برمی‌گرداند که فقط کلیدهایی از اولی را دارد که در دومی نیستند.

</div>

```python
def subtract(d1, d2):
    res = {}
    for key in d1:
        if key not in d2:
            res[key] = d1[key]
    return res
```

عملگرهای رابطه‌ای با مجموعه‌ها کار می‌کنند. برای مثال، <code dir="ltr">&lt;=</code> بررسی می‌کند که آیا یک مجموعه زیرمجموعه‌ی مجموعه‌ی دیگری است، از جمله امکان اینکه مساوی باشند.

</div>

```python
set('ab') <= set('abc')
```

<div dir="rtl" style="text-align: right;">

با این عملگرها می‌توانیم از مجموعه‌ها برای انجام بعضی تمرین‌های فصل ۷ استفاده کنیم. برای مثال، این نسخه‌ای از <code dir="ltr">uses_only</code> است که حلقه به کار می‌برد.

</div>

```python
def uses_only(word, available):
    for letter in word: 
        if letter not in available:
            return False
    return True
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">uses_only</code> بررسی می‌کند که آیا همه‌ی حرف‌های <code dir="ltr">word</code> در <code dir="ltr">available</code> هستند. با مجموعه‌ها می‌توانیم آن را این‌طور بازنویسی کنیم.

</div>

```python
def uses_only(word, available):
    return set(word) <= set(available)
```

<div dir="rtl" style="text-align: right;">

اگر حرف‌های <code dir="ltr">word</code> زیرمجموعه‌ی حرف‌های <code dir="ltr">available</code> باشند، یعنی <code dir="ltr">word</code> فقط از حرف‌های <code dir="ltr">available</code> استفاده می‌کند.

## ۱۸.۳. defaultdict

ماژول <code dir="ltr">collections</code> همچنین <code dir="ltr">defaultdict</code> را دارد، که شبیه دیکشنری است با این تفاوت که اگر به کلیدی که وجود ندارد دسترسی پیدا کنید، خودکار مقدار جدیدی تولید می‌کند.

وقتی یک <code dir="ltr">defaultdict</code> می‌سازید، تابعی را که برای ساختن مقدارهای جدید به کار می‌رود به آن می‌دهید. تابعی که شیء می‌سازد گاهی **کارخانه** (factory) نامیده می‌شود. توابع داخلی‌ای که فهرست، مجموعه و نوع‌های دیگر را می‌سازند می‌توانند به‌عنوان کارخانه به کار روند.

برای مثال، این یک <code dir="ltr">defaultdict</code> است که هر وقت لازم شود یک <code dir="ltr">list</code> جدید می‌سازد.

</div>

```python
from collections import defaultdict

d = defaultdict(list)
d
```

</div>

```python
t = d['new key']
t
```

اگر در حال ساختن دیکشنری‌ای از فهرست‌ها هستید، اغلب می‌توانید با <code dir="ltr">defaultdict</code> کد ساده‌تری بنویسید.

در یکی از تمرین‌های فصل ۱۱، دیکشنری‌ای ساختم که از یک رشته‌ی مرتب‌شده‌ی حروف به فهرست کلمه‌هایی که با آن حروف قابل هجی‌کردن‌اند نگاشت می‌کند. برای مثال، رشته‌ی <code dir="ltr">'opst'</code> به فهرست <code dir="ltr">['opts', 'post', 'pots', 'spot', 'stop', 'tops']</code> نگاشت می‌شود. این هم کد اصلی.

</div>

```python
def all_anagrams(filename):
    d = defaultdict(list)
    for line in open(filename):
        word = line.strip().lower()
        t = signature(word)
        d[t].append(word)
    return d
```

## ۱۸.۴. عبارت‌های شرطی

دستورهای شرطی اغلب برای انتخاب یکی از دو مقدار به کار می‌روند، مثل این:

</div>

```python
import math
x = 5
```

```python
if x > 0:
    y = math.log(x)
else:
    y = float('nan')
```

```python
y
```

<div dir="rtl" style="text-align: right;">

این دستور بررسی می‌کند که <code dir="ltr">x</code> مثبت است یا نه. اگر باشد، لگاریتمش را حساب می‌کند. اگر نباشد، <code dir="ltr">math.log</code> یک ValueError ایجاد می‌کند. برای اینکه برنامه متوقف نشود، یک <code dir="ltr">NaN</code> تولید می‌کنیم، که مقدار ویژه‌ی ممیز شناوری است و «عدد نیست» (Not a Number) را نشان می‌دهد.

می‌توانیم این دستور را با **عبارت شرطی** (conditional expression) مختصرتر بنویسیم.

</div>

```python
y = math.log(x) if x > 0 else float('nan')
```

```python
y
```

<div dir="rtl" style="text-align: right;">

این خط را تقریباً مثل انگلیسی می‌توانید بخوانید: «<code dir="ltr">y</code> برابر log-<code dir="ltr">x</code> می‌شود اگر <code dir="ltr">x</code> بزرگ‌تر از ۰ باشد؛ وگرنه <code dir="ltr">NaN</code> می‌شود».

توابع بازگشتی گاهی با عبارت‌های شرطی مختصر نوشته می‌شوند. برای مثال، این نسخه‌ای از <code dir="ltr">factorial</code> با _دستور_ شرطی است.

</div>

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

<div dir="rtl" style="text-align: right;">

و این نسخه‌ای با _عبارت_ شرطی است.

</div>

```python
def factorial(n):
    return 1 if n == 0 else n * factorial(n-1)
```

<div dir="rtl" style="text-align: right;">

کاربرد دیگر عبارت‌های شرطی مدیریت آرگومان‌های اختیاری است. برای مثال، این تعریف کلاسی با متد <code dir="ltr">__init__</code> است که از دستور شرطی برای بررسی پارامتری با مقدار پیش‌فرض استفاده می‌کند.

</div>

```python
class Kangaroo:
    def __init__(self, name, contents=None):
        self.name = name
        if contents is None:
            contents = []
        self.contents = contents
```

<div dir="rtl" style="text-align: right;">

و این نسخه‌ای است که از عبارت شرطی استفاده می‌کند.

</div>

```python
def __init__(self, name, contents=None):
    self.name = name
    self.contents = [] if contents is None else contents 
```

<div dir="rtl" style="text-align: right;">

به‌طور کلی، می‌توانید دستور شرطی را با عبارت شرطی جایگزین کنید اگر هر دو شاخه یک عبارت واحد داشته باشند و هیچ دستوری نداشته باشند.

</div>

## ۱۸.۵. شمول فهرست (List comprehensions)

در فصل‌های قبل چند مثال دیدیم که با فهرست خالی شروع می‌کنیم و با متد <code dir="ltr">append</code> عنصرها را یکی‌یکی اضافه می‌کنیم. برای مثال، فرض کنید رشته‌ای داریم که عنوان یک اثر را دارد و می‌خواهیم همه‌ی کلمه‌ها را با حرف بزرگ شروع کنیم.

</div>

```python
title = 'once upon a time in a small town'
```

<div dir="rtl" style="text-align: right;">

می‌توانیم آن را به فهرستی از رشته‌ها بشکنیم، رشته‌ها را پیمایش کنیم، حرف اولشان را بزرگ کنیم و به فهرست اضافه کنیم.

</div>

```python
t = []
for word in title.split():
    t.append(word.capitalize())

' '.join(t)
```

<div dir="rtl" style="text-align: right;">

می‌توانیم همین کار را با **شمول فهرست** (list comprehension) مختصرتر انجام دهیم:

</div>

```python
t = [word.capitalize() for word in title.split()]

' '.join(t)
```

<div dir="rtl" style="text-align: right;">

عملگر کروشه نشان می‌دهد که فهرست جدیدی می‌سازیم. عبارت داخل کروشه عنصرهای فهرست را مشخص می‌کند و بند <code dir="ltr">for</code> نشان می‌دهد که روی چه دنباله‌ای حلقه می‌زنیم.

نحو شمول فهرست ممکن است عجیب به نظر برسد، چون متغیر حلقه – در این مثال <code dir="ltr">word</code> – در عبارت پیش از تعریفش ظاهر می‌شود. اما به آن عادت می‌کنید.

به‌عنوان مثال دیگر، در فصل ۹ از این حلقه برای خواندن کلمه‌ها از یک فایل و افزودنشان به فهرست استفاده کردیم.

</div>

```python
download('https://raw.githubusercontent.com/Alireza-Akhavan/ThinkPython/main/words.txt');
```

```python
word_list = []

for line in open('words.txt'):
    word = line.strip()
    word_list.append(word)
```

```python
len(word_list)
```

<div dir="rtl" style="text-align: right;">

این‌طور می‌توانیم آن را به‌صورت شمول فهرست بنویسیم.

</div>

```python
word_list = [line.strip() for line in open('words.txt')]
```

```python
len(word_list)
```

<div dir="rtl" style="text-align: right;">

شمول فهرست می‌تواند بند <code dir="ltr">if</code> هم داشته باشد که تعیین می‌کند کدام عنصرها در فهرست قرار بگیرند. برای مثال، این حلقه <code dir="ltr">for</code> است که در فصل ۱۰ برای ساختن فهرستی فقط از کلمه‌های <code dir="ltr">word_list</code> که palindrome (کلمه‌ی متقارن) هستند به کار بردیم.

</div>

```python
def is_palindrome(word):
    return list(reversed(word)) == list(word)
```

```python
palindromes = []

for word in word_list:
    if is_palindrome(word):
        palindromes.append(word)
```

```python
palindromes[:10]
```

<div dir="rtl" style="text-align: right;">

این‌طور می‌توانیم همین کار را با شمول فهرست انجام دهیم.

</div>

```python
palindromes = [word for word in word_list if is_palindrome(word)]
```

```python
palindromes[:10]
```

<div dir="rtl" style="text-align: right;">

وقتی شمول فهرست به‌عنوان آرگومان یک تابع به کار می‌رود، اغلب می‌توانیم کروشه‌ها را حذف کنیم. برای مثال، فرض کنید می‌خواهیم $1 / 2^n$ را برای مقدارهای $n$ از ۰ تا ۹ جمع کنیم. می‌توانیم از شمول فهرست این‌طور استفاده کنیم.

</div>

```python
sum([1/2**n for n in range(10)])
```

<div dir="rtl" style="text-align: right;">

یا می‌توانیم کروشه‌ها را این‌طور کنار بگذاریم.

</div>

```python
sum(1/2**n for n in range(10))
```

<div dir="rtl" style="text-align: right;">

در این مثال، آرگومان از نظر فنی یک **عبارت مولد** (generator expression) است، نه شمول فهرست، و هیچ‌وقت واقعاً فهرستی نمی‌سازد. اما جدای از این، رفتار یکسان است.

شمول فهرست و عبارت مولد مختصر و خواندنی هستند، دست‌کم برای عبارت‌های ساده. و معمولاً از حلقه‌های <code dir="ltr">for</code> معادل سریع‌ترند، گاهی خیلی سریع‌تر. پس اگر از اینکه زودتر به آن‌ها اشاره نکردم عصبانی هستید، درک می‌کنم.

اما به دفاع از خودم، شمول فهرست اشکال‌زدایی‌اش سخت‌تر است چون نمی‌توانید داخل حلقه دستور print بگذارید. پیشنهاد می‌کنم فقط وقتی از آن‌ها استفاده کنید که محاسبه آن‌قدر ساده باشد که احتمالاً بار اول درست می‌نویسیدش. یا فکر کنید یک حلقه‌ی <code dir="ltr">for</code> بنویسید و اشکال‌زدایی کنید و بعد آن را به شمول فهرست تبدیل کنید.

</div>

## ۱۸.۶. <code dir="ltr">any</code> و <code dir="ltr">all</code>

پایتون تابع داخلی <code dir="ltr">any</code> را دارد که دنباله‌ای از مقدارهای بولی می‌گیرد و اگر هر یک از مقدارها <code dir="ltr">True</code> باشد <code dir="ltr">True</code> برمی‌گرداند.

</div>

```python
any([False, False, True])
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">any</code> اغلب با عبارت‌های مولد به کار می‌رود.

</div>

```python
any(letter == 't' for letter in 'mount')
```

<div dir="rtl" style="text-align: right;">

این مثال خیلی مفید نیست چون همان کاری را می‌کند که عملگر <code dir="ltr">in</code> می‌کند. اما می‌توانیم از <code dir="ltr">any</code> برای نوشتن راه‌حل‌های مختصر بعضی تمرین‌های فصل ۷ استفاده کنیم. برای مثال، می‌توانیم <code dir="ltr">uses_none</code> را این‌طور بنویسیم.

</div>

```python
def uses_none(word, forbidden):
    """Checks whether a word avoids forbidden letters."""
    return not any(letter in forbidden for letter in word)
```

```python
uses_none('banana', 'xyz')
```

```python
uses_none('apple', 'efg')
```

<div dir="rtl" style="text-align: right;">

این تابع حرف‌های <code dir="ltr">word</code> را پیمایش می‌کند و بررسی می‌کند که آیا هیچ‌کدام در <code dir="ltr">forbidden</code> هستند. استفاده از <code dir="ltr">any</code> با عبارت مولد کارآمد است چون اگر مقدار <code dir="ltr">True</code> پیدا کند بلافاصله متوقف می‌شود، پس لازم نیست کل دنباله را پیمایش کند.

پایتون تابع داخلی دیگری هم دارد، <code dir="ltr">all</code>، که اگر هر عنصر دنباله <code dir="ltr">True</code> باشد <code dir="ltr">True</code> برمی‌گرداند. می‌توانیم از آن برای نوشتن نسخه‌ای مختصر از <code dir="ltr">uses_all</code> استفاده کنیم.

</div>

```python
def uses_all(word, required):
    """Check whether a word uses all required letters."""
    return all(letter in word for letter in required)
```

```python
uses_all('banana', 'ban')
```

```python
uses_all('apple', 'api')
```

<div dir="rtl" style="text-align: right;">

عبارت‌هایی که از <code dir="ltr">any</code> و <code dir="ltr">all</code> استفاده می‌کنند می‌توانند مختصر، کارآمد و خواندنی باشند.

</div>

## ۱۸.۷. چندتایی‌های نام‌دار (Named tuples)

ماژول <code dir="ltr">collections</code> تابعی به نام <code dir="ltr">namedtuple</code> دارد که می‌توان با آن کلاس‌های ساده ساخت. برای مثال، شیء <code dir="ltr">Point</code> در فصل ۱۶ فقط دو صفت دارد، <code dir="ltr">x</code> و <code dir="ltr">y</code>. این‌طور آن را تعریف کردیم.

</div>

```python
class Point:
    """Represents a point in 2-D space."""
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f'({self.x}, {self.y})'
```

<div dir="rtl" style="text-align: right;">

این کد زیادی است برای رساندن مقدار کمی اطلاعات. <code dir="ltr">namedtuple</code> راه مختصرتری برای تعریف کلاس‌هایی مثل این فراهم می‌کند.

</div>

```python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
```

<div dir="rtl" style="text-align: right;">

آرگومان اول نام کلاسی است که می‌خواهید بسازید. دومی فهرست صفت‌هایی است که شیءهای <code dir="ltr">Point</code> باید داشته باشند. نتیجه یک شیء کلاس است، به همین دلیل به متغیری با نام بزرگ‌حرف نسبت داده می‌شود.

کلاسی که با <code dir="ltr">namedtuple</code> ساخته شود، متد <code dir="ltr">__init__</code> دارد که مقدارها را به صفت‌ها نسبت می‌دهد و <code dir="ltr">__str__</code> (در عمل <code dir="ltr">__repr__</code>) که شیء را به شکل خواناپذیر نمایش می‌دهد. پس می‌توانیم یک شیء <code dir="ltr">Point</code> را این‌طور بسازیم و نمایش دهیم.

</div>

```python
p = Point(1, 2)
p
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">Point</code> همچنین متد <code dir="ltr">__eq__</code> دارد که بررسی می‌کند دو شیء <code dir="ltr">Point</code> هم‌ارزند – یعنی صفت‌هایشان یکسان است.

</div>

```python
p == Point(1, 2)
```

<div dir="rtl" style="text-align: right;">

می‌توانید به عنصرهای یک چندتایی نام‌دار با نام یا با اندیس دسترسی پیدا کنید.

</div>

```python
p.x, p.y
```

```python
p[0], p[1]
```

<div dir="rtl" style="text-align: right;">

همچنین می‌توانید یک چندتایی نام‌دار را مثل چندتایی معمولی به کار ببرید، مثل این انتساب.

</div>

```python
x, y = p
x, y
```

<div dir="rtl" style="text-align: right;">

اما شیءهای <code dir="ltr">namedtuple</code> تغییرناپذیرند. بعد از مقداردهی اولیه‌ی صفت‌ها، نمی‌توان آن‌ها را تغییر داد.

</div>

```python
%%expect TypeError

p[0] = 3
```

```python
%%expect AttributeError

p.x = 3
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">namedtuple</code> راهی سریع برای تعریف کلاس‌های ساده است. عیبش این است که کلاس‌های ساده همیشه ساده نمی‌مانند. ممکن است بعداً بخواهید متدهایی به چندتایی نام‌دار اضافه کنید. در آن صورت، می‌توانید کلاس جدیدی تعریف کنید که از چندتایی نام‌دار ارث ببرد.

</div>

```python
class Pointier(Point):
    """This class inherits from Point"""
```

یا در آن مرحله می‌توانید به تعریف کلاس معمولی برگردید.

## ۱۸.۸. بسته‌بندی آرگومان‌های کلیدواژه‌ای

در فصل ۱۱ تابعی نوشتیم که آرگومان‌هایش را در یک چندتایی بسته‌بندی (pack) می‌کند.

</div>

```python
def mean(*args):
    return sum(args) / len(args)
```

<div dir="rtl" style="text-align: right;">

می‌توانید این تابع را با هر تعداد آرگومان فراخوانی کنید.

</div>

```python
mean(1, 2, 3)
```

<div dir="rtl" style="text-align: right;">

اما عملگر <code dir="ltr">*</code> آرگومان‌های کلیدواژه‌ای را بسته‌بندی نمی‌کند. پس فراخوانی این تابع با آرگومان کلیدواژه‌ای باعث خطا می‌شود.

</div>

```python
%%expect TypeError

mean(1, 2, start=3)
```

<div dir="rtl" style="text-align: right;">

برای بسته‌بندی آرگومان‌های کلیدواژه‌ای، می‌توانیم از عملگر <code dir="ltr">**</code> استفاده کنیم:

</div>

```python
def mean(*args, **kwargs):
    print(kwargs)
    return sum(args) / len(args)
```

<div dir="rtl" style="text-align: right;">

پارامتر بسته‌بندی‌کننده‌ی کلیدواژه‌ها می‌تواند هر نامی داشته باشد، اما <code dir="ltr">kwargs</code> انتخاب رایجی است. نتیجه دیکشنری‌ای است که از کلیدواژه‌ها به مقدارها نگاشت می‌کند.

</div>

```python
%%expect TypeError

mean(1, 2, start=3)
```

<div dir="rtl" style="text-align: right;">

در این مثال، مقدار <code dir="ltr">kwargs</code> چاپ می‌شود، اما در غیر این صورت اثری ندارد.

اما عملگر <code dir="ltr">**</code> را می‌توان در فهرست آرگومان‌ها هم برای باز کردن (unpack) دیکشنری به کار برد. برای مثال، این نسخه‌ای از <code dir="ltr">mean</code> است که هر آرگومان کلیدواژه‌ای را که بگیرد بسته‌بندی می‌کند و بعد به‌صورت آرگومان‌های کلیدواژه‌ای برای <code dir="ltr">sum</code> بازشان می‌کند.

</div>

```python
def mean(*args, **kwargs):
    return sum(args, **kwargs) / len(args)
```

<div dir="rtl" style="text-align: right;">

حالا اگر <code dir="ltr">mean</code> را با <code dir="ltr">start</code> به‌عنوان آرگومان کلیدواژه‌ای فراخوانی کنیم، به <code dir="ltr">sum</code> پاس داده می‌شود و <code dir="ltr">sum</code> از آن به‌عنوان نقطه‌ی شروع جمع استفاده می‌کند. در مثال زیر <code dir="ltr">start=3</code> پیش از محاسبه‌ی میانگین، <code dir="ltr">3</code> را به مجموع اضافه می‌کند، پس مجموع <code dir="ltr">6</code> است و نتیجه <code dir="ltr">3</code> می‌شود.

</div>

```python
%%expect TypeError

mean(1, 2, start=3)
```

<div dir="rtl" style="text-align: right;">

به‌عنوان مثال دیگر، اگر دیکشنری‌ای با کلیدهای <code dir="ltr">x</code> و <code dir="ltr">y</code> داشته باشیم، می‌توانیم با عملگر باز کردن از آن برای ساختن شیء <code dir="ltr">Point</code> استفاده کنیم.

</div>

```python
d = dict(x=1, y=2)
Point(**d)
```

<div dir="rtl" style="text-align: right;">

بدون عملگر باز کردن، <code dir="ltr">d</code> به‌عنوان یک آرگومان موقعیتی در نظر گرفته می‌شود، پس به <code dir="ltr">x</code> نسبت داده می‌شود و <code dir="ltr">TypeError</code> می‌گیریم چون آرگومان دومی برای نسبت دادن به <code dir="ltr">y</code> نیست.

</div>

```python
%%expect TypeError

d = dict(x=1, y=2)
Point(d)
```

<div dir="rtl" style="text-align: right;">

وقتی با توابعی کار می‌کنید که تعداد زیادی آرگومان کلیدواژه‌ای دارند، اغلب مفید است دیکشنری‌هایی بسازید و دست‌به‌دست کنید که گزینه‌های پرکاربرد را مشخص می‌کنند.

</div>

```python
def pack_and_print(**kwargs):
    print(kwargs)
    
pack_and_print(a=1, b=2)
```

## ۱۸.۹. اشکال‌زدایی

در فصل‌های قبل از <code dir="ltr">doctest</code> برای آزمودن توابع استفاده کردیم. برای مثال، این تابعی به نام <code dir="ltr">add</code> است که دو عدد می‌گیرد و مجموعشان را برمی‌گرداند. یک doctest دارد که بررسی می‌کند <code dir="ltr">2 + 2</code> برابر <code dir="ltr">4</code> است.

</div>

```python
def add(a, b):
    '''Add two numbers.
    
    >>> add(2, 2)
    4
    '''
    return a + b
```

<div dir="rtl" style="text-align: right;">

این تابع یک شیء تابع می‌گیرد و doctestهایش را اجرا می‌کند.

</div>

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```

<div dir="rtl" style="text-align: right;">

پس می‌توانیم <code dir="ltr">add</code> را این‌طور آزمون کنیم.

</div>

```python
run_doctests(add)
```

<div dir="rtl" style="text-align: right;">

خروجی ندارد، که یعنی همه‌ی آزمون‌ها گذشتند.

پایتون ابزار دیگری برای اجرای آزمون‌های خودکار دارد به نام <code dir="ltr">unittest</code>. استفاده‌اش کمی پیچیده‌تر است، اما این یک مثال است.

</div>

```python
from unittest import TestCase

class TestExample(TestCase):

    def test_add(self):
        result = add(2, 2)
        self.assertEqual(result, 4)
```

<div dir="rtl" style="text-align: right;">

اول <code dir="ltr">TestCase</code> را وارد می‌کنیم، که کلاسی در ماژول <code dir="ltr">unittest</code> است. برای استفاده از آن، باید کلاس جدیدی تعریف کنیم که از <code dir="ltr">TestCase</code> ارث ببرد و دست‌کم یک متد آزمون داشته باشد. نام متد آزمون باید با <code dir="ltr">test</code> شروع شود و باید نشان دهد کدام تابع را آزمون می‌کند.

در این مثال، <code dir="ltr">test_add</code> تابع <code dir="ltr">add</code> را با فراخوانی آن، ذخیره‌ی نتیجه و احضار <code dir="ltr">assertEqual</code> که از <code dir="ltr">TestCase</code> به ارث رسیده آزمون می‌کند. <code dir="ltr">assertEqual</code> دو آرگومان می‌گیرد و بررسی می‌کند که برابرند.

برای اجرای این متد آزمون، باید تابعی در <code dir="ltr">unittest</code> به نام <code dir="ltr">main</code> را اجرا کنیم و چند آرگومان کلیدواژه‌ای به آن بدهیم. تابع زیر جزئیات را نشان می‌دهد – اگر کنجکاو هستید، می‌توانید از یک دستیار مجازی بخواهید توضیح دهد چطور کار می‌کند.

</div>

```python
import unittest

def run_unittest():
    unittest.main(argv=[''], verbosity=0, exit=False)
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">run_unittest</code> کلاس <code dir="ltr">TestExample</code> را به‌عنوان آرگومان نمی‌گیرد – در عوض، دنبال کلاس‌هایی که از <code dir="ltr">TestCase</code> ارث می‌برند می‌گردد. بعد دنبال متدهایی که با <code dir="ltr">test</code> شروع می‌شوند می‌گردد و آن‌ها را اجرا می‌کند. این فرایند **کشف آزمون** (test discovery) نامیده می‌شود.

این چیزی است که وقتی <code dir="ltr">run_unittest</code> را فراخوانی می‌کنیم رخ می‌دهد.

</div>

```python
run_unittest()
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">unittest.main</code> تعداد آزمون‌هایی را که اجرا کرده و نتایج را گزارش می‌کند. در این مورد <code dir="ltr">OK</code> نشان می‌دهد که آزمون‌ها گذشتند.

برای دیدن اینکه وقتی آزمونی شکست می‌خورد چه می‌شود، متد آزمون نادرستی به <code dir="ltr">TestExample</code> اضافه می‌کنیم.

</div>

```python
%%add_method_to TestExample

    def test_add_broken(self):
        result = add(2, 2)
        self.assertEqual(result, 100)
```

<div dir="rtl" style="text-align: right;">

این چیزی است که وقتی آزمون‌ها را اجرا می‌کنیم رخ می‌دهد.

</div>

```python
run_unittest()
```

در تمرین‌های زیر، چند درخواست پیشنهاد می‌کنم که می‌توانید برای پرسیدن اطلاعات بیشتر درباره‌ی <code dir="ltr">unittest</code> از یک دستیار مجازی استفاده کنید.

## ۱۸.۱۰. واژه‌نامه

**کارخانه (factory):** تابعی که برای ساختن شیءها به کار می‌رود، که اغلب به‌عنوان پارامتر به تابع دیگری داده می‌شود.

**عبارت شرطی (conditional expression):** عبارتی که از یک شرط برای انتخاب یکی از دو مقدار استفاده می‌کند.

**شمول فهرست (list comprehension):** راهی مختصر برای پیمایش یک دنباله و ساختن فهرست.

**عبارت مولد (generator expression):** شبیه شمول فهرست، با این تفاوت که فهرست نمی‌سازد.

**کشف آزمون (test discovery):** فرایندی برای یافتن و اجرای آزمون‌ها.

### تمرین ۵

این تابعی است که ضریب دوجمله‌ای را به‌صورت بازگشتی حساب می‌کند.

</div>

```python
def binomial_coeff(n, k):
    """Compute the binomial coefficient "n choose k".

    n: number of trials
    k: number of successes

    returns: int
    """
    if k == 0:
        return 1
    
    if n == 0:
        return 0

    return binomial_coeff(n-1, k) + binomial_coeff(n-1, k-1)
```
