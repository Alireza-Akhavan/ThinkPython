## ۹.۱. فهرست یک دنباله است

مثل رشته، **فهرست** (list) دنباله‌ای از مقدارهاست. در رشته مقدارها نویسه‌اند؛ در فهرست می‌توانند از هر نوعی باشند. مقدارهای داخل فهرست **عنصر** (element) نامیده می‌شوند.

چند راه برای ساختن یک فهرست جدید وجود دارد؛ ساده‌ترین راه محصور کردن عنصرها در کروشه (<code dir="ltr">[</code> و <code dir="ltr">]</code>) است. برای مثال، این فهرستی از دو عدد صحیح است.

</div>

```python
numbers = [42, 123]
```

<div dir="rtl" style="text-align: right;">

و این فهرستی از سه رشته است.

</div>

```python
cheeses = ['Cheddar', 'Edam', 'Gouda']
```

<div dir="rtl" style="text-align: right;">

عنصرهای فهرست لازم نیست از یک نوع باشند. فهرست زیر یک رشته، یک عدد اعشاری، یک عدد صحیح و حتی یک فهرست دیگر را شامل می‌شود.

</div>

```python
t = ['spam', 2.0, 5, [10, 20]]
```

<div dir="rtl" style="text-align: right;">

فهرستی که داخل فهرست دیگر است **تودرتو** (nested) نامیده می‌شود.

فهرستی که هیچ عنصری ندارد فهرست تهی نامیده می‌شود؛ می‌توانید آن را با کروشه‌ی خالی، <code dir="ltr">[]</code>، بسازید.

</div>

```python
empty = []
```

<div dir="rtl" style="text-align: right;">

تابع <code dir="ltr">len</code> طول فهرست را برمی‌گرداند.

</div>

```python
len(cheeses)
```

<div dir="rtl" style="text-align: right;">

طول فهرست تهی <code dir="ltr">0</code> است.

</div>

```python
len(empty)
```

<div dir="rtl" style="text-align: right;">

شکل زیر نمودار وضعیت <code dir="ltr">cheeses</code>، <code dir="ltr">numbers</code> و <code dir="ltr">empty</code> را نشان می‌دهد.

(در کتاب اصلی اینجا نمودار رسم شده است: هر فهرست به‌صورت یک جعبه با کلمه‌ی «list» بیرون آن و عنصرهای شماره‌دار داخل آن؛ <code dir="ltr">cheeses</code> سه عنصر، <code dir="ltr">numbers</code> دو عنصر و <code dir="ltr">empty</code> هیچ عنصری ندارد.)

فهرست‌ها با جعبه‌هایی نشان داده می‌شوند که کلمه‌ی «list» بیرون آن‌ها و عنصرهای شماره‌دار فهرست داخل آن‌هاست.

</div>

## ۹.۲. فهرست‌ها تغییرپذیرند

برای خواندن یک عنصر از فهرست، می‌توانیم از عملگر کروشه استفاده کنیم. اندیس اولین عنصر <code dir="ltr">0</code> است.

</div>

```python
cheeses[0]
```

<div dir="rtl" style="text-align: right;">

برخلاف رشته‌ها، فهرست‌ها **تغییرپذیر** (mutable) هستند. وقتی عملگر کروشه در سمت چپ یک نسبت‌دهی بیاید، مشخص می‌کند کدام عنصر فهرست مقدار جدید می‌گیرد.

</div>

```python
numbers[1] = 17
numbers
```

<div dir="rtl" style="text-align: right;">

عنصر دوم <code dir="ltr">numbers</code> که قبلاً <code dir="ltr">123</code> بود، اکنون <code dir="ltr">17</code> است.

اندیس‌های فهرست مثل اندیس‌های رشته کار می‌کنند:

-   هر عبارت صحیحی می‌تواند به‌عنوان اندیس به کار رود.

-   اگر سعی کنید عنصری را که وجود ندارد بخوانید یا بنویسید، <code dir="ltr">IndexError</code> می‌گیرید.

-   اگر اندیس مقدار منفی داشته باشد، از انتهای فهرست به عقب می‌شمارد.

عملگر <code dir="ltr">in</code> روی فهرست‌ها هم کار می‌کند – بررسی می‌کند که آیا عنصر داده‌شده در هر جای فهرست هست یا نه.

</div>

```python
'Edam' in cheeses
```

```python
'Wensleydale' in cheeses
```

<div dir="rtl" style="text-align: right;">

اگرچه یک فهرست می‌تواند فهرست دیگری را در بر داشته باشد، فهرست تودرتو همچنان یک عنصر به حساب می‌آید – پس در فهرست زیر فقط چهار عنصر وجود دارد.

</div>

```python
t = ['spam', 2.0, 5, [10, 20]]
len(t)
```

<div dir="rtl" style="text-align: right;">

و <code dir="ltr">10</code> عنصر <code dir="ltr">t</code> به حساب نمی‌آید، چون عنصر یک فهرست تودرتوست، نه <code dir="ltr">t</code>.

</div>

```python
10 in t
```

پایتون متدهایی دارد که روی فهرست‌ها عمل می‌کنند. برای مثال، <code dir="ltr">append</code> یک عنصر جدید به انتهای فهرست اضافه می‌کند:

</div>

```python
letters.append('e')
letters
```

<div dir="rtl" style="text-align: right;">

## ۹.۶. فهرست‌ها و رشته‌ها

رشته دنباله‌ای از نویسه‌هاست و فهرست دنباله‌ای از مقدارها، اما فهرستی از نویسه‌ها با رشته یکی نیست. برای تبدیل یک رشته به فهرستی از نویسه‌ها می‌توانید از تابع <code dir="ltr">list</code> استفاده کنید.

</div>

```python
s = 'spam'
t = list(s)
t
```

<div dir="rtl" style="text-align: right;">

تابع <code dir="ltr">list</code> رشته را به حرف‌های جداگانه می‌شکند. اگر می‌خواهید رشته را به کلمه‌ها بشکنید، می‌توانید از متد <code dir="ltr">split</code> استفاده کنید:

</div>

```python
s = 'resting by the quiet river'
t = s.split()
t
```

<div dir="rtl" style="text-align: right;">

یک آرگومان اختیاری به نام **جداکننده** (delimiter) مشخص می‌کند از کدام نویسه‌ها به‌عنوان مرز کلمه‌ها استفاده شود. مثال زیر از خط تیره به‌عنوان جداکننده استفاده می‌کند.

</div>

```python
s = 'ex-student'
t = s.split('-')
t
```

<div dir="rtl" style="text-align: right;">

اگر فهرستی از رشته‌ها دارید، می‌توانید با <code dir="ltr">join</code> آن‌ها را در یک رشته‌ی واحد الحاق کنید. <code dir="ltr">join</code> متد رشته است، پس باید آن را روی جداکننده احضار کنید و فهرست را به‌عنوان آرگومان بدهید.

</div>

```python
delimiter = ' '
t = ['resting', 'by', 'the', 'quiet', 'river']
s = delimiter.join(t)
s
```

<div dir="rtl" style="text-align: right;">

در این مورد جداکننده یک نویسه‌ی فاصله است، پس <code dir="ltr">join</code> بین کلمه‌ها فاصله می‌گذارد. برای الحاق رشته‌ها بدون فاصله می‌توانید از رشته‌ی تهی، <code dir="ltr">''</code>، به‌عنوان جداکننده استفاده کنید.

</div>

## ۹.۷. پیمایش یک فهرست

می‌توانید از دستور <code dir="ltr">for</code> برای پیمایش عنصرهای یک فهرست استفاده کنید.

</div>

```python
for cheese in cheeses:
    print(cheese)
```

<div dir="rtl" style="text-align: right;">

برای مثال، بعد از استفاده از <code dir="ltr">split</code> برای ساختن فهرستی از کلمه‌ها، می‌توانیم از <code dir="ltr">for</code> برای پیمایش آن‌ها استفاده کنیم.

</div>

```python
s = 'resting by the quiet river'

for word in s.split():
    print(word)
```

```python
for x in []:
    print('This never happens.')
```

## ۹.۹. شیءها و مقدارها

اگر این دستورهای نسبت‌دهی را اجرا کنیم:

</div>

```python
a = 'banana'
b = 'banana'
```

<div dir="rtl" style="text-align: right;">

می‌دانیم هر دوی <code dir="ltr">a</code> و <code dir="ltr">b</code> به یک رشته اشاره می‌کنند، اما نمی‌دانیم به رشته‌ی *یکسان* اشاره می‌کنند یا نه. دو وضعیت ممکن وجود دارد که در شکل زیر نشان داده شده‌اند.

(در کتاب اصلی اینجا نمودار رسم شده است: در نمودار سمت چپ، <code dir="ltr">a</code> و <code dir="ltr">b</code> هر کدام به یک رشته‌ی <code dir="ltr">'banana'</code> جداگانه اشاره می‌کنند؛ در نمودار سمت راست، هر دو به یک رشته‌ی واحد اشاره می‌کنند.)

در نمودار سمت چپ، <code dir="ltr">a</code> و <code dir="ltr">b</code> به دو شیء متفاوت اشاره می‌کنند که مقدار یکسانی دارند. در نمودار سمت راست، به یک شیء اشاره می‌کنند. برای بررسی اینکه دو متغیر به یک شیء اشاره می‌کنند یا نه، می‌توانید از عملگر <code dir="ltr">is</code> استفاده کنید.

</div>

```python
a = 'banana'
b = 'banana'
a is b
```

<div dir="rtl" style="text-align: right;">

در این مثال پایتون فقط یک شیء رشته ساخت و هر دوی <code dir="ltr">a</code> و <code dir="ltr">b</code> به آن اشاره می‌کنند. اما وقتی دو فهرست می‌سازید، دو شیء دارید.

</div>

```python
a = [1, 2, 3]
b = [1, 2, 3]
a is b
```

<div dir="rtl" style="text-align: right;">

پس نمودار وضعیت این‌طور است.

(در کتاب اصلی اینجا نمودار رسم شده است: <code dir="ltr">a</code> و <code dir="ltr">b</code> هر کدام به یک فهرست جداگانه‌ی <code dir="ltr">[1, 2, 3]</code> اشاره می‌کنند.)

در این مورد می‌گوییم این دو فهرست **هم‌ارز** (equivalent) هستند، چون عنصرهای یکسانی دارند، اما **یکسان** (identical) نیستند، چون یک شیء نیستند. اگر دو شیء یکسان باشند، هم‌ارز هم هستند، اما اگر هم‌ارز باشند لزوماً یکسان نیستند.

</div>

## ۹.۱۰. هم‌نام‌سازی (aliasing)

اگر <code dir="ltr">a</code> به یک شیء اشاره کند و <code dir="ltr">b = a</code> را نسبت دهید، هر دو متغیر به یک شیء اشاره می‌کنند.

</div>

ارتباط یک متغیر با یک شیء **ارجاع** (reference) نامیده می‌شود. در این مثال، دو ارجاع به یک شیء وجود دارد.

شیئی که بیش از یک ارجاع دارد بیش از یک نام دارد، پس می‌گوییم شیء **هم‌نام** (aliased) شده است. اگر شیء هم‌نام‌شده تغییرپذیر باشد، تغییراتی که با یک نام اعمال می‌شود بر نام دیگر هم اثر می‌گذارد. در این مثال، اگر شیئی را که <code dir="ltr">b</code> به آن اشاره می‌کند تغییر دهیم، شیئی را که <code dir="ltr">a</code> به آن اشاره می‌کند هم تغییر داده‌ایم.

</div>

```python
b[0] = 5
a
```

برای شیءهای تغییرناپذیر مثل رشته‌ها، هم‌نام‌سازی مشکل چندانی نیست. در این مثال:

```python
a = 'banana'
b = 'banana'
```

## ۹.۱۱. آرگومان‌های فهرست

وقتی فهرستی را به یک تابع می‌دهید، تابع یک ارجاع به فهرست می‌گیرد. اگر تابع فهرست را تغییر دهد، فراخواننده تغییر را می‌بیند. برای مثال، <code dir="ltr">pop_first</code> از متد فهرست <code dir="ltr">pop</code> برای حذف اولین عنصر یک فهرست استفاده می‌کند.

```python
letters
```

در فصل قبل فایل <code dir="ltr">words.txt</code> را خواندیم و دنبال کلمه‌هایی با ویژگی‌های مشخص، مثل داشتن حرف <code dir="ltr">e</code>، گشتیم. اما کل فایل را چندین بار خواندیم که کارآمد نیست. بهتر است فایل را یک بار بخوانیم و کلمه‌ها را در یک فهرست بگذاریم. حلقه‌ی زیر نشان می‌دهد چگونه.

```python
download('https://raw.githubusercontent.com/Alireza-Akhavan/ThinkPython/main/words.txt');
```

```python
word_list = []

for line in open('words.txt'):
    word = line.strip()
    word_list.append(word)
    
len(word_list)
```

قبل از حلقه، <code dir="ltr">word_list</code> با یک فهرست تهی مقداردهی اولیه می‌شود. هر بار که حلقه تکرار می‌شود، متد <code dir="ltr">append</code> یک کلمه به انتها اضافه می‌کند. وقتی حلقه تمام می‌شود، بیش از ۱۱۳٬۰۰۰ کلمه در فهرست هست.

دقت کنید که بیشتر متدهای فهرست آرگومان را تغییر می‌دهند و <code dir="ltr">None</code> برمی‌گردانند. این برعکس متدهای رشته است که رشته‌ی جدیدی برمی‌گردانند و رشته‌ی اصلی را دست‌نخورده می‌گذارند.

## ۹.۱۴. واژه‌نامه

**فهرست (list):** شیئی که دنباله‌ای از مقدارها را در بر دارد.

**عنصر (element):** یکی از مقدارهای یک فهرست یا دنباله‌ی دیگر.

**فهرست تودرتو (nested list):** فهرستی که عنصر فهرست دیگری است.

**جداکننده (delimiter):** نویسه یا رشته‌ای که نشان می‌دهد رشته باید از کجا شکسته شود.

**هم‌ارز (equivalent):** داشتن مقدار یکسان.

**یکسان (identical):** یک شیء بودن (که هم‌ارزی را نتیجه می‌دهد).

**ارجاع (reference):** ارتباط بین یک متغیر و مقدارش.

**هم‌نام‌شده (aliased):** اگر بیش از یک متغیر به یک شیء اشاره کنند، شیء هم‌نام شده است.

**صفت (attribute):** یکی از مقدارهای نام‌دار وابسته به یک شیء.

</div>
