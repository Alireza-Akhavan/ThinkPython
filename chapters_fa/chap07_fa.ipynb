## ۷.۱. حلقه‌ها و رشته‌ها

در فصل ۳ یک حلقه‌ی <code dir="ltr">for</code> دیدیم که با تابع <code dir="ltr">range</code> دنباله‌ای از عددها را نمایش می‌داد.

</div>

```python
for i in range(3):
    print(i, end=' ')
```

<div dir="rtl" style="text-align: right;">

این نسخه از آرگومان کلیدواژه‌ای <code dir="ltr">end</code> استفاده می‌کند تا تابع <code dir="ltr">print</code> به‌جای رفتن به خط بعد، بعد از هر عدد یک فاصله بگذارد.

از حلقه‌ی <code dir="ltr">for</code> می‌توانیم برای نمایش حرف‌های یک رشته هم استفاده کنیم.

</div>

```python
for letter in 'Gadsby':
    print(letter, end=' ')
```

<div dir="rtl" style="text-align: right;">

دقت کنید که نام متغیر را از <code dir="ltr">i</code> به <code dir="ltr">letter</code> تغییر دادم، که اطلاعات بیشتری درباره‌ی مقداری که به آن اشاره می‌کند می‌دهد. متغیری که در حلقه‌ی <code dir="ltr">for</code> تعریف می‌شود **متغیر حلقه** (loop variable) نامیده می‌شود.

حالا که می‌توانیم حرف‌های یک کلمه را پیمایش کنیم، می‌توانیم بررسی کنیم که آیا حرف «e» را دارد.

</div>

```python
for letter in "Gadsby":
    if letter == 'E' or letter == 'e':
        print('This word has an "e"')
```

<div dir="rtl" style="text-align: right;">

قبل از ادامه، این حلقه را در یک تابع کپسوله می‌کنیم.

</div>

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            print('This word has an "e"')
```

<div dir="rtl" style="text-align: right;">

و آن را به یک تابع خالص تبدیل می‌کنیم که اگر کلمه «e» داشته باشد <code dir="ltr">True</code> و در غیر این صورت <code dir="ltr">False</code> برگرداند.

</div>

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            return True
    return False
```

<div dir="rtl" style="text-align: right;">

می‌توانیم آن را تعمیم دهیم تا کلمه را به‌عنوان پارامتر بگیرد.

</div>

```python
def has_e(word):
    for letter in word:
        if letter == 'E' or letter == 'e':
            return True
    return False
```

<div dir="rtl" style="text-align: right;">

حالا می‌توانیم آن را این‌طور آزمایش کنیم:

</div>

```python
has_e('Gadsby')
```

```python
has_e('Emma')
```

## ۷.۳. به‌روزرسانی متغیرها

شاید خودتان متوجه شده باشید که مجاز است بیش از یک بار به یک متغیر مقدار نسبت دهید. نسبت‌دهی جدید باعث می‌شود متغیر موجود به مقدار جدیدی اشاره کند (و دیگر به مقدار قبلی اشاره نکند).

برای مثال، این یک نسبت‌دهی اولیه است که متغیر را می‌سازد.

</div>

```python
x = 5
x
```

<div dir="rtl" style="text-align: right;">

و این نسبت‌دهی‌ای است که مقدار متغیر را عوض می‌کند.

</div>

```python
x = 7
x
```

<div dir="rtl" style="text-align: right;">

شکل زیر نشان می‌دهد این نسبت‌دهی‌ها در یک نمودار وضعیت (state diagram) چه شکلی هستند.

(در کتاب اصلی اینجا نمودار رسم شده است: متغیر <code dir="ltr">x</code> با یک پیکان نقطه‌چین به <code dir="ltr">5</code> و یک پیکان پیوسته به <code dir="ltr">7</code>.)

پیکان نقطه‌چین نشان می‌دهد <code dir="ltr">x</code> دیگر به <code dir="ltr">5</code> اشاره نمی‌کند. پیکان پیوسته نشان می‌دهد اکنون به <code dir="ltr">7</code> اشاره می‌کند.

یک نوع رایج نسبت‌دهی **به‌روزرسانی** (update) است، که در آن مقدار جدید متغیر به مقدار قدیمی‌اش بستگی دارد.

</div>

```python
x = 7
```

```python
x = x + 1
x
```

<div dir="rtl" style="text-align: right;">

این دستور یعنی «مقدار فعلی <code dir="ltr">x</code> را بگیر، یکی اضافه کن و نتیجه را دوباره به <code dir="ltr">x</code> نسبت بده.»

اگر سعی کنید متغیری را که وجود ندارد به‌روزرسانی کنید، خطا می‌گیرید، چون پایتون عبارت سمت راست را قبل از نسبت دادن مقدار به متغیر سمت چپ ارزیابی می‌کند.

## ۷.۵. عملگر <code dir="ltr">in</code>

نسخه‌ای از <code dir="ltr">has_e</code> که در این فصل نوشتیم پیچیده‌تر از لازم است. پایتون عملگری به نام <code dir="ltr">in</code> دارد که بررسی می‌کند آیا یک نویسه در یک رشته وجود دارد یا نه.

</div>

```python
word = 'Gadsby'
'e' in word
```

<div dir="rtl" style="text-align: right;">

پس می‌توانیم <code dir="ltr">has_e</code> را این‌طور بازنویسی کنیم.

</div>

```python
def has_e(word):
    if 'E' in word or 'e' in word:
        return True
    else:
        return False
```

<div dir="rtl" style="text-align: right;">

و چون شرط دستور <code dir="ltr">if</code> یک مقدار بولی دارد، می‌توانیم دستور <code dir="ltr">if</code> را حذف کنیم و مقدار بولی را مستقیماً برگردانیم.

</div>

```python
def has_e(word):
    return 'E' in word or 'e' in word
```

<div dir="rtl" style="text-align: right;">

می‌توانیم این تابع را باز هم ساده‌تر کنیم با استفاده از متد <code dir="ltr">lower</code> که حرف‌های یک رشته را به حروف کوچک تبدیل می‌کند. این یک مثال است.

</div>

```python
word.lower()
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">lower</code> یک رشته‌ی جدید می‌سازد – رشته‌ی موجود را تغییر نمی‌دهد – پس مقدار <code dir="ltr">word</code> بدون تغییر می‌ماند.

</div>

```python
word
```

<div dir="rtl" style="text-align: right;">

</div>

```python
def has_e(word):
    return 'e' in word.lower()
```

## ۷.۶. جست‌وجو

بر اساس این نسخه‌ی ساده‌تر <code dir="ltr">has_e</code>، تابع کلی‌تری به نام <code dir="ltr">uses_any</code> می‌نویسیم که پارامتر دومی دارد که رشته‌ای از حرف‌هاست. این تابع اگر کلمه از هر یک از حرف‌ها استفاده کند <code dir="ltr">True</code> و در غیر این صورت <code dir="ltr">False</code> برمی‌گرداند.

</div>

```python
def uses_any(word, letters):
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

<div dir="rtl" style="text-align: right;">

این مثالی است که نتیجه‌اش <code dir="ltr">True</code> است.

</div>

```python
uses_any('banana', 'aeiou')
```

<div dir="rtl" style="text-align: right;">

و این یکی که نتیجه‌اش <code dir="ltr">False</code> است.

</div>

```python
uses_any('apple', 'xyz')
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">uses_any</code> هر دوی <code dir="ltr">word</code> و <code dir="ltr">letters</code> را به حروف کوچک تبدیل می‌کند، پس با هر ترکیبی از حروف بزرگ و کوچک کار می‌کند.

</div>

## ۷.۷. Doctest

در فصل ۴ از یک docstring برای مستندسازی یک تابع استفاده کردیم – یعنی برای توضیح اینکه چه می‌کند. می‌توان از docstring برای *آزمودن* تابع هم استفاده کرد. این نسخه‌ای از <code dir="ltr">uses_any</code> است که docstring آن شامل آزمون‌هاست.

</div>

```python
def uses_any(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any('banana', 'aeiou')
    True
    >>> uses_any('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

<div dir="rtl" style="text-align: right;">

هر آزمون با <code dir="ltr">&gt;&gt;&gt;</code> شروع می‌شود، که در بعضی محیط‌های پایتون به‌عنوان اعلان (prompt) استفاده می‌شود تا نشان دهد کاربر کجا می‌تواند کد را تایپ کند. در یک doctest، بعد از اعلان یک عبارت می‌آید، معمولاً یک فراخوانی تابع. خط بعدی مقداری را که عبارت در صورت درست کار کردن تابع باید داشته باشد نشان می‌دهد.

در مثال اول، <code dir="ltr">'banana'</code> از <code dir="ltr">'a'</code> استفاده می‌کند، پس نتیجه باید <code dir="ltr">True</code> باشد. در مثال دوم، <code dir="ltr">'apple'</code> از هیچ‌کدام از <code dir="ltr">'xyz'</code> استفاده نمی‌کند، پس نتیجه باید <code dir="ltr">False</code> باشد.

برای اجرای این آزمون‌ها باید ماژول <code dir="ltr">doctest</code> را وارد کنیم و تابعی به نام <code dir="ltr">run_docstring_examples</code> را اجرا کنیم. برای اینکه استفاده از این تابع راحت‌تر شود، تابع زیر را نوشتم که یک شیء تابع را به‌عنوان آرگومان می‌گیرد.

</div>

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```

<div dir="rtl" style="text-align: right;">

هنوز درباره‌ی <code dir="ltr">globals</code> و <code dir="ltr">__name__</code> چیزی یاد نگرفته‌ایم – می‌توانید نادیده‌شان بگیرید. حالا می‌توانیم <code dir="ltr">uses_any</code> را این‌طور آزمایش کنیم.

</div>

```python
run_doctests(uses_any)
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">run_doctests</code> عبارت‌های داخل docstring را پیدا می‌کند و ارزیابی‌شان می‌کند. اگر نتیجه با مقدار مورد انتظار یکی باشد، آزمون **قبول می‌شود** (passes). در غیر این صورت **رد می‌شود** (fails).

اگر همه‌ی آزمون‌ها قبول شوند، <code dir="ltr">run_doctests</code> هیچ خروجی‌ای نمایش نمی‌دهد – در آن صورت «بی‌خبری، خوش‌خبری است.» برای دیدن اینکه وقتی یک آزمون رد می‌شود چه اتفاقی می‌افتد، این یک نسخه‌ی نادرست از <code dir="ltr">uses_any</code> است.

</div>

```python
def uses_any_incorrect(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any_incorrect('banana', 'aeiou')
    True
    >>> uses_any_incorrect('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
        else:
            return False     # INCORRECT!
```

<div dir="rtl" style="text-align: right;">

و این چیزی است که هنگام آزمودنش پیش می‌آید.

</div>

```python
run_doctests(uses_any_incorrect)
```

(شماره‌ی خط در گزارش، بسته به محیط اجرا ممکن است متفاوت باشد.)

خروجی شامل مثالی که رد شده، مقداری که تابع باید تولید می‌کرد و مقداری که واقعاً تولید کرد است.

اگر مطمئن نیستید چرا این آزمون رد شد، در یکی از تمرین‌ها فرصت اشکال‌زدایی‌اش را خواهید داشت.

## ۷.۸. واژه‌نامه

**متغیر حلقه (loop variable):** متغیری که در سرآیند یک حلقه‌ی <code dir="ltr">for</code> تعریف می‌شود.

**شیء فایل (file object):** شیئی که یک فایل باز را نشان می‌دهد و پیگیری می‌کند کدام بخش‌های فایل خوانده یا نوشته شده‌اند.

**متد (method):** تابعی که به یک شیء وابسته است و با عملگر نقطه فراخوانی می‌شود.

**به‌روزرسانی (update):** یک دستور نسبت‌دهی که به متغیری که از قبل وجود دارد مقدار جدیدی می‌دهد، به‌جای اینکه متغیر تازه‌ای بسازد.

**مقداردهی اولیه (initialize):** ساختن یک متغیر جدید و دادن مقداری به آن.

**افزایش (increment):** زیاد کردن مقدار یک متغیر.

**کاهش (decrement):** کم کردن مقدار یک متغیر.

**شمارنده (counter):** متغیری که برای شمردن چیزی به کار می‌رود، معمولاً با صفر مقداردهی اولیه می‌شود و بعد افزایش می‌یابد.

**جست‌وجوی خطی (linear search):** الگوی محاسباتی‌ای که دنباله‌ای از عنصرها را می‌پیماید و وقتی چیزی را که دنبالش می‌گردد پیدا کرد متوقف می‌شود.

**قبول شدن (pass):** اگر آزمونی اجرا شود و نتیجه همان مورد انتظار باشد، آزمون قبول می‌شود.

**رد شدن (fail):** اگر آزمونی اجرا شود و نتیجه همان مورد انتظار نباشد، آزمون رد می‌شود.
