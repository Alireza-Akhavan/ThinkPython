## این کتاب برای چه کسی است؟

اگر می‌خواهید برنامه‌نویسی یاد بگیرید، به جای درستی آمده‌اید. پایتون یکی از بهترین زبان‌های برنامه‌نویسی برای مبتدی‌هاست – و همچنین یکی از پرتقاضاترین مهارت‌ها.

همچنین در زمان مناسبی آمده‌اید، چون یادگیری برنامه‌نویسی در حال حاضر احتمالاً آسان‌تر از هر زمان دیگری است. با دستیارهای مجازی مثل ChatGPT، لازم نیست تنها یاد بگیرید. در سراسر این کتاب، راه‌هایی پیشنهاد می‌کنم که با آن‌ها از این ابزارها برای تسریع یادگیری استفاده کنید.

این کتاب در درجه‌ی اول برای کسانی است که هرگز برنامه‌نویسی نکرده‌اند و کسانی که تجربه‌ای در زبان برنامه‌نویسی دیگری دارند. اگر در پایتون تجربه‌ی قابل‌توجهی دارید، ممکن است چند فصل اول را بیش از حد کند بیابید.

یکی از چالش‌های یادگیری برنامه‌نویسی این است که باید *دو* زبان یاد بگیرید: یکی خود زبان برنامه‌نویسی؛ دیگری واژگانی که برای صحبت درباره‌ی برنامه‌ها به کار می‌بریم. اگر فقط زبان برنامه‌نویسی را یاد بگیرید، احتمالاً وقتی لازم باشد پیام خطا را تفسیر کنید، مستندات را بخوانید، با شخص دیگری صحبت کنید یا از دستیارهای مجازی استفاده کنید مشکل خواهید داشت. اگر کمی برنامه‌نویسی کرده‌اید اما این زبان دوم را یاد نگرفته‌اید، امیدوارم این کتاب برایتان مفید باشد.

## هدف‌های کتاب

هنگام نوشتن این کتاب سعی کردم در واژگان دقت کنم. هر اصطلاح را وقتی اولین بار ظاهر می‌شود تعریف می‌کنم. و در پایان هر فصل یک واژه‌نامه هست که اصطلاح‌های معرفی‌شده را مرور می‌کند.

همچنین سعی کردم مختصر باشم. هرچه خواندن کتاب به تلاش ذهنی کمتری نیاز داشته باشد، ظرفیت بیشتری برای برنامه‌نویسی خواهید داشت.

اما نمی‌توانید فقط با خواندن یک کتاب برنامه‌نویسی یاد بگیرید – باید تمرین کنید. به همین دلیل، این کتاب در پایان هر فصل تمرین‌هایی دارد که در آن‌ها می‌توانید آنچه را یاد گرفته‌اید تمرین کنید.

اگر با دقت بخوانید و پیوسته روی تمرین‌ها کار کنید، پیشرفت خواهید کرد. اما از همین حالا هشدار می‌دهم – یادگیری برنامه‌نویسی آسان نیست، و حتی برای برنامه‌نویسان باتجربه هم می‌تواند ناامیدکننده باشد. در طول مسیر، راهبردهایی پیشنهاد می‌کنم که به نوشتن برنامه‌های درست و رفع برنامه‌های نادرست کمک کنند.

## راهنمای پیمایش کتاب

هر فصل این کتاب بر فصل‌های قبلی بنا شده، پس باید آن‌ها را به ترتیب بخوانید و پیش از رفتن به فصل بعد وقت بگذارید و روی تمرین‌ها کار کنید.

شش فصل اول عنصرهای پایه‌ای مثل حساب، شرط‌ها و حلقه‌ها را معرفی می‌کنند. همچنین مهم‌ترین مفهوم برنامه‌نویسی، یعنی توابع، و شیوه‌ی قدرتمندی برای استفاده از آن‌ها، یعنی بازگشت (recursion)، را معرفی می‌کنند.

فصل‌های ۷ و ۸ رشته‌ها را معرفی می‌کنند – که می‌توانند حرف‌ها، کلمه‌ها و جمله‌ها را نشان دهند – و الگوریتم‌هایی برای کار با آن‌ها.

فصل‌های ۹ تا ۱۲ ساختارهای داده‌ی اصلی پایتون – فهرست‌ها، دیکشنری‌ها و چندتایی‌ها – را معرفی می‌کنند، که ابزارهای قدرتمندی برای نوشتن برنامه‌های کارآمد هستند. فصل ۱۲ الگوریتم‌هایی برای تحلیل متن و تولید تصادفی متن جدید ارائه می‌دهد. الگوریتم‌هایی مثل این در هسته‌ی مدل‌های زبانی بزرگ (LLM) قرار دارند، پس این فصل ایده‌ای از شیوه‌ی کار ابزارهایی مثل ChatGPT به شما می‌دهد.

فصل ۱۳ درباره‌ی راه‌های ذخیره‌ی داده در حافظه‌ی بلندمدت است – فایل‌ها و پایگاه‌های داده. به‌عنوان تمرین، می‌توانید برنامه‌ای بنویسید که یک سامانه‌ی فایل را جست‌وجو می‌کند و فایل‌های تکراری را پیدا می‌کند.

فصل‌های ۱۴ تا ۱۷ برنامه‌نویسی شیءگرا (OOP) را معرفی می‌کنند، که شیوه‌ای برای سازمان‌دهی برنامه‌ها و داده‌هایی است که با آن‌ها کار می‌کنند. بسیاری از کتابخانه‌های پایتون به سبک شیءگرا نوشته شده‌اند، پس این فصل‌ها به فهم طراحی آن‌ها کمک می‌کنند – و به تعریف شیءهای خودتان.

هدف این کتاب پوشش دادن کل زبان پایتون نیست. در عوض، روی زیرمجموعه‌ای از زبان تمرکز می‌کنم که بیشترین توانایی را با کمترین مفهوم فراهم می‌کند. با این حال، پایتون امکانات زیادی دارد که می‌توانید برای حل کارآمد مسئله‌های رایج به کار ببرید. فصل ۱۸ بعضی از این امکانات را ارائه می‌دهد.

سرانجام، فصل ۱۹ نظرها و پیشنهادهای پایانی‌ام را برای ادامه‌ی سفر برنامه‌نویسی‌تان ارائه می‌دهد.

## در ویرایش سوم چه چیز جدید است؟

بزرگ‌ترین تغییرها در این ویرایش را دو فناوری جدید رقم زدند – نوت‌بوک‌های Jupyter و دستیارهای مجازی.

هر فصل این کتاب یک نوت‌بوک Jupyter است، که سندی است شامل هم متن معمولی و هم کد. برای من، این نوشتن کد، آزمودن آن و سازگار نگه داشتنش با متن را آسان‌تر می‌کند. برای شما، یعنی می‌توانید کد را اجرا کنید، تغییرش دهید و روی تمرین‌ها کار کنید، همه در یک جا. دستورالعمل کار با نوت‌بوک‌ها در فصل اول است.

تغییر بزرگ دیگر این است که توصیه‌هایی برای کار با دستیارهای مجازی مثل ChatGPT و استفاده از آن‌ها برای تسریع یادگیری افزوده‌ام. وقتی ویرایش قبلی این کتاب در ۲۰۱۶ منتشر شد، پیشینیان این ابزارها بسیار کم‌فایده‌تر بودند و بیشتر مردم از آن‌ها بی‌خبر بودند. حالا آن‌ها ابزار استانداردی برای مهندسی نرم‌افزارند، و فکر می‌کنم ابزاری تحول‌آفرین برای یادگیری برنامه‌نویسی خواهند بود – و یادگیری بسیاری چیزهای دیگر.

تغییرهای دیگر کتاب از حسرت‌هایم درباره‌ی ویرایش دوم برخاست.

اولی این بود که بر آزمون نرم‌افزار تأکید نکرده بودم. این در ۲۰۱۶ هم کوتاهی تأسف‌باری بود، اما با ظهور دستیارهای مجازی، آزمون خودکار حتی مهم‌تر شده است. پس این ویرایش پرکاربردترین ابزارهای آزمون پایتون، یعنی <code dir="ltr">doctest</code> و <code dir="ltr">unittest</code>، را ارائه می‌دهد و چند تمرین دارد که در آن‌ها می‌توانید کار با آن‌ها را تمرین کنید.

حسرت دیگرم این بود که تمرین‌های ویرایش دوم ناهمگون بودند – بعضی جالب‌تر از بقیه بودند و بعضی بیش از حد سخت. رفتن به نوت‌بوک‌های Jupyter به من کمک کرد دنباله‌ای جذاب‌تر و مؤثرتر از تمرین‌ها را توسعه دهم و بیازمایم.

در این بازنگری، ترتیب موضوع‌ها تقریباً همان است، اما چند فصل را جابه‌جا کردم و دو فصل کوتاه را در یکی فشرده کردم. همچنین پوشش رشته‌ها را گسترش دادم تا عبارت‌های منظم (regular expressions) را هم شامل شود.

چند فصل از گرافیک لاک‌پشتی (turtle graphics) استفاده می‌کنند. در ویرایش‌های قبلی از ماژول <code dir="ltr">turtle</code> پایتون استفاده می‌کردم، اما متأسفانه در نوت‌بوک‌های Jupyter کار نمی‌کند. پس آن را با ماژول turtle جدیدی جایگزین کردم که باید استفاده‌اش آسان‌تر باشد.

سرانجام، بخش قابل‌توجهی از متن را بازنویسی کردم، جاهایی که لازم بود را روشن‌تر کردم و در جاهایی که به اندازه‌ی ممکن مختصر نبودم کوتاه کردم.

به این ویرایش جدید خیلی افتخار می‌کنم – امیدوارم خوشتان بیاید!

## شروع کار

برای بیشتر زبان‌های برنامه‌نویسی، از جمله پایتون، ابزارهای زیادی هست که می‌توانید برای نوشتن و اجرای برنامه‌ها به کار ببرید. این ابزارها **محیط‌های توسعه‌ی یکپارچه** (IDE) نامیده می‌شوند. به‌طور کلی دو نوع IDE وجود دارد:

* بعضی با فایل‌هایی که کد دارند کار می‌کنند، پس ابزارهایی برای ویرایش و اجرای این فایل‌ها فراهم می‌کنند.

* بعضی دیگر عمدتاً با نوت‌بوک‌ها کار می‌کنند، که سندهایی‌اند شامل متن و کد.

برای مبتدی‌ها، شروع با یک محیط توسعه‌ی نوت‌بوکی مثل Jupyter را توصیه می‌کنم.

نوت‌بوک‌های این کتاب از یک مخزن آنلاین در <https://allendowney.github.io/ThinkPython> در دسترس‌اند.

دو راه برای استفاده از آن‌ها هست:

* می‌توانید نوت‌بوک‌ها را دانلود کنید و روی کامپیوتر خودتان اجرا کنید. در آن صورت، باید پایتون و Jupyter را نصب کنید، که سخت نیست، اما اگر می‌خواهید پایتون یاد بگیرید، صرف زمان زیاد برای نصب نرم‌افزار می‌تواند ناامیدکننده باشد.

* راه جایگزین اجرای نوت‌بوک‌ها در Colab است، که محیط Jupyter‌ای است که در مرورگر وب اجرا می‌شود، پس لازم نیست چیزی نصب کنید. Colab را Google اداره می‌کند و استفاده از آن رایگان است.

اگر تازه شروع کرده‌اید، قویاً توصیه می‌کنم با Colab شروع کنید.
