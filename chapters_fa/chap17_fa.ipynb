## ۱۷.۱. نمایش ورق‌ها

در یک دسته‌ی ورق استاندارد ۵۲ ورق وجود دارد – هر کدام به یکی از چهار خال (suit) و یکی از سیزده رتبه (rank) تعلق دارد. خال‌ها پیک (Spades)، دل (Hearts)، خشت (Diamonds) و گشنیز (Clubs) هستند. رتبه‌ها آس (Ace)، ۲، ۳، ۴، ۵، ۶، ۷، ۸، ۹، ۱۰، سرباز (Jack)، بی‌بی (Queen) و شاه (King) هستند. بسته به بازی‌ای که می‌کنید، آس می‌تواند بالاتر از شاه یا پایین‌تر از ۲ باشد.

اگر بخواهیم شیء جدیدی برای نمایش یک ورق تعریف کنیم، روشن است که صفت‌ها باید چه باشند: <code dir="ltr">rank</code> و <code dir="ltr">suit</code>. اینکه صفت‌ها از چه نوعی باشند کمتر روشن است. یک امکان استفاده از رشته‌هایی مثل <code dir="ltr">'Spade'</code> برای خال‌ها و <code dir="ltr">'Queen'</code> برای رتبه‌هاست. مشکل این پیاده‌سازی این است که مقایسه‌ی ورق‌ها برای اینکه ببینیم کدام رتبه یا خال بالاتری دارد آسان نخواهد بود.

یک راه جایگزین استفاده از اعداد صحیح برای **کدگذاری** (encode) رتبه‌ها و خال‌هاست. در اینجا «کدگذاری» یعنی نگاشتی بین عددها و خال‌ها یا بین عددها و رتبه‌ها تعریف می‌کنیم. این نوع کدگذاری قرار نیست مخفی باشد (آن «رمزنگاری» (encryption) می‌شد).

برای مثال، این جدول خال‌ها و کدهای صحیح متناظرشان را نشان می‌دهد:

| خال | کد |
| --- | --- |
|  Spades (پیک)      |   3  |
|  Hearts (دل)      |   2  |
|  Diamonds (خشت)    |   1  |
|  Clubs (گشنیز)       |   0  |

با این کدگذاری، می‌توانیم خال‌ها را با مقایسه‌ی کدهایشان مقایسه کنیم.

برای کدگذاری رتبه‌ها، از عدد صحیح <code dir="ltr">2</code> برای رتبه‌ی <code dir="ltr">2</code>، <code dir="ltr">3</code> برای <code dir="ltr">3</code> و همین‌طور تا <code dir="ltr">10</code> استفاده می‌کنیم. جدول زیر کدهای ورق‌های تصویری را نشان می‌دهد.

| رتبه | کد |
| --- | --- |
|  Jack (سرباز)     |   11  |
|  Queen (بی‌بی)   |   12  |
|  King (شاه)      |   13  |

و می‌توانیم برای نمایش آس هم <code dir="ltr">1</code> و هم <code dir="ltr">14</code> به کار ببریم، بسته به اینکه بخواهیم پایین‌تر یا بالاتر از بقیه‌ی رتبه‌ها در نظر گرفته شود.

برای نمایش این کدگذاری‌ها، از دو فهرست رشته استفاده می‌کنیم، یکی با نام خال‌ها و دیگری با نام رتبه‌ها.

این تعریف کلاسی است که یک ورق بازی را نشان می‌دهد، با این فهرست‌های رشته به‌صورت **متغیرهای کلاس** (class variables)، یعنی متغیرهایی که داخل تعریف کلاس تعریف می‌شوند، اما نه داخل یک متد.

</div>

```python
class Card:
    """Represents a standard playing card."""

    suit_names = ['Clubs', 'Diamonds', 'Hearts', 'Spades']
    rank_names = [None, 'Ace', '2', '3', '4', '5', '6', '7', 
                  '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']
```

<div dir="rtl" style="text-align: right;">

اولین عنصر <code dir="ltr">rank_names</code> برابر <code dir="ltr">None</code> است چون ورقی با رتبه‌ی صفر وجود ندارد. با گنجاندن <code dir="ltr">None</code> به‌عنوان نگه‌دارنده‌ی جا، فهرستی با این ویژگی خوب می‌گیریم که اندیس <code dir="ltr">2</code> به رشته‌ی <code dir="ltr">'2'</code> نگاشت می‌شود و همین‌طور الی آخر.

متغیرهای کلاس به کلاس مربوطند، نه به یک نمونه از کلاس، پس می‌توانیم این‌طور به آن‌ها دسترسی داشته باشیم.

</div>

```python
Card.suit_names
```

<div dir="rtl" style="text-align: right;">

می‌توانیم از <code dir="ltr">suit_names</code> برای جست‌وجوی یک خال و گرفتن رشته‌ی متناظر استفاده کنیم.

</div>

```python
Card.suit_names[0]
```

<div dir="rtl" style="text-align: right;">

و از <code dir="ltr">rank_names</code> برای جست‌وجوی یک رتبه.

</div>

```python
Card.rank_names[11]
```

## ۱۷.۲. صفت‌های Card

این یک متد <code dir="ltr">__init__</code> برای کلاس <code dir="ltr">Card</code> است – <code dir="ltr">suit</code> و <code dir="ltr">rank</code> را به‌عنوان پارامتر می‌گیرد و به صفت‌هایی با همان نام‌ها نسبت می‌دهد.

</div>

```python
%%add_method_to Card

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
```

این یک متد <code dir="ltr">__str__</code> برای شیءهای <code dir="ltr">Card</code> است.

</div>

```python
%%add_method_to Card

    def __str__(self):
        rank_name = Card.rank_names[self.rank]
        suit_name = Card.suit_names[self.suit]
        return f'{rank_name} of {suit_name}' 
```

## ۱۷.۴. مقایسه‌ی ورق‌ها

فرض کنید یک شیء <code dir="ltr">Card</code> دوم با همان خال و رتبه بسازیم.

</div>

```python
queen2 = Card(1, 12)
print(queen2)
```

<div dir="rtl" style="text-align: right;">

اگر از عملگر <code dir="ltr">==</code> برای مقایسه‌شان استفاده کنیم، بررسی می‌کند که <code dir="ltr">queen</code> و <code dir="ltr">queen2</code> به یک شیء اشاره می‌کنند یا نه.

</div>

```python
queen == queen2
```

<div dir="rtl" style="text-align: right;">

اشاره نمی‌کنند، پس <code dir="ltr">False</code> برمی‌گرداند. می‌توانیم این رفتار را با تعریف متد ویژه‌ی <code dir="ltr">__eq__</code> تغییر دهیم.

</div>

```python
%%add_method_to Card

    def __eq__(self, other):
        return self.suit == other.suit and self.rank == other.rank
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">__eq__</code> دو شیء <code dir="ltr">Card</code> را به‌عنوان پارامتر می‌گیرد و اگر خال و رتبه‌ی یکسان داشته باشند <code dir="ltr">True</code> برمی‌گرداند، حتی اگر یک شیء نباشند. به عبارت دیگر، بررسی می‌کند که آیا هم‌ارزند، حتی اگر یکسان نباشند.

وقتی عملگر <code dir="ltr">==</code> را با شیءهای <code dir="ltr">Card</code> به کار می‌بریم، پایتون متد <code dir="ltr">__eq__</code> را فراخوانی می‌کند.

</div>

```python
queen == queen2
```

<div dir="rtl" style="text-align: right;">

به‌عنوان آزمون دوم، ورقی با همان خال و رتبه‌ی متفاوت می‌سازیم.

</div>

```python
six = Card(1, 6)
print(six)
```

<div dir="rtl" style="text-align: right;">

می‌توانیم تأیید کنیم که <code dir="ltr">queen</code> و <code dir="ltr">six</code> هم‌ارز نیستند.

</div>

```python
queen == six
```

<div dir="rtl" style="text-align: right;">

اگر از عملگر <code dir="ltr">!=</code> استفاده کنیم، پایتون متد ویژه‌ای به نام <code dir="ltr">__ne__</code> را، در صورت وجود، احضار می‌کند. در غیر این صورت <code dir="ltr">__eq__</code> را احضار می‌کند و نتیجه را معکوس می‌کند – پس اگر <code dir="ltr">__eq__</code> مقدار <code dir="ltr">True</code> برگرداند، نتیجه‌ی عملگر <code dir="ltr">!=</code> برابر <code dir="ltr">False</code> است.

</div>

```python
queen != queen2
```

```python
queen != six
```

<div dir="rtl" style="text-align: right;">

حالا فرض کنید می‌خواهیم دو ورق را مقایسه کنیم تا ببینیم کدام بزرگ‌تر است. اگر از یکی از عملگرهای رابطه‌ای استفاده کنیم، <code dir="ltr">TypeError</code> می‌گیریم.

</div>

```python
%%expect TypeError

queen < queen2
```

<div dir="rtl" style="text-align: right;">

برای تغییر رفتار عملگر <code dir="ltr">&lt;</code>، می‌توانیم متد ویژه‌ای به نام <code dir="ltr">__lt__</code> تعریف کنیم، که مخفف «less than» (کوچک‌تر از) است. برای این مثال، فرض می‌کنیم خال مهم‌تر از رتبه است – پس همه‌ی پیک‌ها از همه‌ی دل‌ها برترند، که از همه‌ی خشت‌ها برترند و همین‌طور الی آخر. اگر دو ورق خال یکسان داشته باشند، آن‌که رتبه‌ی بالاتری دارد برنده است.

برای پیاده‌سازی این منطق، از متد زیر استفاده می‌کنیم که چندتایی‌ای شامل خال و رتبه‌ی یک ورق، به همین ترتیب، برمی‌گرداند.

</div>

```python
%%add_method_to Card

    def to_tuple(self):
        return (self.suit, self.rank)
```

<div dir="rtl" style="text-align: right;">

می‌توانیم از این متد برای نوشتن <code dir="ltr">__lt__</code> استفاده کنیم.

</div>

```python
%%add_method_to Card

    def __lt__(self, other):
        return self.to_tuple() < other.to_tuple()
```

<div dir="rtl" style="text-align: right;">

مقایسه‌ی چندتایی‌ها عنصرهای اول هر چندتایی را، که خال‌ها را نشان می‌دهند، مقایسه می‌کند. اگر یکسان بودند، عنصرهای دوم، که رتبه‌ها را نشان می‌دهند، مقایسه می‌شوند.

حالا اگر از عملگر <code dir="ltr">&lt;</code> استفاده کنیم، متد <code dir="ltr">__lt__</code> احضار می‌شود.

</div>

```python
six < queen
```

<div dir="rtl" style="text-align: right;">

اگر از عملگر <code dir="ltr">&gt;</code> استفاده کنیم، متد ویژه‌ای به نام <code dir="ltr">__gt__</code> را، در صورت وجود، احضار می‌کند. در غیر این صورت <code dir="ltr">__lt__</code> را با آرگومان‌ها به ترتیب معکوس احضار می‌کند.

</div>

```python
%%expect TypeError

queen < queen2
```

```python
queen > queen2
```

<div dir="rtl" style="text-align: right;">

سرانجام، اگر از عملگر <code dir="ltr">&lt;=</code> استفاده کنیم، متد ویژه‌ای به نام <code dir="ltr">__le__</code> را احضار می‌کند.

</div>

```python
%%add_method_to Card

    def __le__(self, other):
        return self.to_tuple() <= other.to_tuple()
```

<div dir="rtl" style="text-align: right;">

پس می‌توانیم بررسی کنیم که آیا یک ورق از دیگری کوچک‌تر یا مساوی با آن است.

</div>

```python
queen <= queen2
```

```python
queen <= six
```

<div dir="rtl" style="text-align: right;">

اگر از عملگر <code dir="ltr">&gt;=</code> استفاده کنیم، در صورت وجود از <code dir="ltr">__ge__</code> استفاده می‌کند. در غیر این صورت، <code dir="ltr">__le__</code> را با آرگومان‌ها به ترتیب معکوس احضار می‌کند.

</div>

```python
queen >= six
```

<div dir="rtl" style="text-align: right;">

همان‌طور که این متدها را تعریف کرده‌ایم، **کامل**‌اند به این معنا که می‌توانیم هر دو شیء <code dir="ltr">Card</code> را مقایسه کنیم، و **سازگار**‌اند به این معنا که نتیجه‌ی عملگرهای متفاوت با هم تناقض ندارند. با این دو ویژگی، می‌توانیم بگوییم شیءهای <code dir="ltr">Card</code> **ترتیب کامل** (totally ordered) دارند. و این یعنی، همان‌طور که به‌زودی خواهیم دید، می‌توان آن‌ها را مرتب کرد.

</div>

## ۱۷.۵. دسته‌های ورق (Decks)

## ۱۷.۸. والدین و فرزندان

وراثت توانایی تعریف کلاس جدیدی است که نسخه‌ی تغییریافته‌ی یک کلاس موجود باشد. برای مثال، فرض کنید می‌خواهیم کلاسی داشته باشیم که یک «دست» (hand) را نشان دهد، یعنی ورق‌هایی که یک بازیکن در دست دارد.

* یک دست شبیه یک دسته است – هر دو از مجموعه‌ای از ورق‌ها تشکیل شده‌اند، و هر دو عملیاتی مثل اضافه و حذف ورق لازم دارند.

* یک دست با یک دسته فرق هم دارد – عملیاتی هست که برای دست‌ها می‌خواهیم و برای دسته معنا ندارد. برای مثال، در پوکر ممکن است دو دست را مقایسه کنیم تا ببینیم کدام برنده است. در بریج (bridge)، ممکن است برای بستن یک مزایده (bid) امتیازی برای یک دست حساب کنیم.

این رابطه بین کلاس‌ها – که یکی نسخه‌ی تخصصی‌شده‌ی دیگری است – برای وراثت مناسب است.

برای تعریف کلاس جدیدی که بر پایه‌ی یک کلاس موجود است، نام کلاس موجود را داخل پرانتز می‌گذاریم.

</div>

```python
class Hand(Deck):
    """Represents a hand of playing cards."""
```

<div dir="rtl" style="text-align: right;">

این تعریف نشان می‌دهد که <code dir="ltr">Hand</code> از <code dir="ltr">Deck</code> ارث می‌برد، یعنی شیءهای <code dir="ltr">Hand</code> می‌توانند به متدهای تعریف‌شده در <code dir="ltr">Deck</code>، مثل <code dir="ltr">take_card</code> و <code dir="ltr">put_card</code>، دسترسی داشته باشند.

<code dir="ltr">Hand</code> متد <code dir="ltr">__init__</code> را هم از <code dir="ltr">Deck</code> به ارث می‌برد، اما اگر <code dir="ltr">__init__</code> را در کلاس <code dir="ltr">Hand</code> تعریف کنیم، آن را که در کلاس <code dir="ltr">Deck</code> است **بازنویسی** (override) می‌کند.

</div>

```python
%%add_method_to Hand

    def __init__(self, label=''):
        self.label = label
        self.cards = []
```

<div dir="rtl" style="text-align: right;">

این نسخه از <code dir="ltr">__init__</code> یک رشته‌ی اختیاری به‌عنوان پارامتر می‌گیرد و همیشه با فهرست خالی از ورق‌ها شروع می‌کند. وقتی یک <code dir="ltr">Hand</code> می‌سازیم، پایتون این متد را احضار می‌کند، نه آن را که در <code dir="ltr">Deck</code> است – که می‌توانیم با بررسی اینکه نتیجه صفت <code dir="ltr">label</code> دارد تأیید کنیم.

</div>

```python
hand = Hand('player 1')
hand.label
```

<div dir="rtl" style="text-align: right;">

برای پخش کردن یک ورق، می‌توانیم از <code dir="ltr">take_card</code> برای برداشتن ورق از یک <code dir="ltr">Deck</code> و از <code dir="ltr">put_card</code> برای اضافه کردن آن به یک <code dir="ltr">Hand</code> استفاده کنیم.

</div>

```python
deck = Deck(cards)
card = deck.take_card()
hand.put_card(card)
print(hand)
```

<div dir="rtl" style="text-align: right;">

این کد را در یک متد <code dir="ltr">Deck</code> به نام <code dir="ltr">move_cards</code> کپسوله می‌کنیم.

</div>

```python
%%add_method_to Deck

    def move_cards(self, other, num):
        for i in range(num):
            card = self.take_card()
            other.put_card(card)
```

<div dir="rtl" style="text-align: right;">

این متد چندریخت است – یعنی با بیش از یک نوع کار می‌کند: <code dir="ltr">self</code> و <code dir="ltr">other</code> هر کدام می‌توانند <code dir="ltr">Hand</code> یا <code dir="ltr">Deck</code> باشند. پس می‌توانیم از این متد برای پخش یک ورق از <code dir="ltr">Deck</code> به <code dir="ltr">Hand</code>، از یک <code dir="ltr">Hand</code> به دیگری، یا از یک <code dir="ltr">Hand</code> به <code dir="ltr">Deck</code> استفاده کنیم.

وقتی کلاس جدیدی از کلاس موجودی ارث می‌برد، کلاس موجود **والد** (parent) و کلاس جدید **فرزند** (child) نامیده می‌شود. به‌طور کلی:

* نمونه‌های کلاس فرزند باید همه‌ی صفت‌های کلاس والد را داشته باشند، اما می‌توانند صفت‌های اضافی هم داشته باشند.

* کلاس فرزند باید همه‌ی متدهای کلاس والد را داشته باشد، اما می‌تواند متدهای اضافی هم داشته باشد.

* اگر کلاس فرزند متدی از کلاس والد را بازنویسی کند، متد جدید باید پارامترهای یکسان بگیرد و نتیجه‌ای سازگار برگرداند.

این مجموعه قواعد «اصل جایگزینی لیسکوف» (Liskov substitution principle) نامیده می‌شود، به نام Barbara Liskov، دانشمند علوم کامپیوتر.

اگر این قواعد را رعایت کنید، هر تابع یا متدی که برای کار با نمونه‌ای از کلاس والد، مثل <code dir="ltr">Deck</code>، طراحی شده، با نمونه‌های کلاس فرزند، مثل <code dir="ltr">Hand</code>، هم کار خواهد کرد. اگر این قواعد را نقض کنید، کدتان مثل قصر ورقی فرو می‌ریزد (house of cards؛ ببخشید، بازی با کلمات شد).

</div>

## ۱۷.۹. تخصصی‌سازی (Specialization)

کلاسی به نام <code dir="ltr">BridgeHand</code> می‌سازیم که یک دست در بریج – بازی ورق پرطرفدار – را نشان می‌دهد. از <code dir="ltr">Hand</code> ارث می‌بریم و متد جدیدی به نام <code dir="ltr">high_card_point_count</code> اضافه می‌کنیم که یک دست را با روش «امتیاز ورق‌های بالا» ارزیابی می‌کند، یعنی برای ورق‌های بالای دست امتیاز جمع می‌کند.

این تعریف کلاسی است که یک دیکشنری را به‌عنوان متغیر کلاس دارد که از نام ورق‌ها به امتیازشان نگاشت می‌کند.

</div>

```python
class BridgeHand(Hand):
    """Represents a bridge hand."""

    hcp_dict = {
        'Ace': 4,
        'King': 3,
        'Queen': 2,
        'Jack': 1,
    }
```

<div dir="rtl" style="text-align: right;">

با داشتن رتبه‌ی یک ورق، مثلاً <code dir="ltr">12</code>، می‌توانیم از <code dir="ltr">Card.rank_names</code> برای گرفتن نمایش رشته‌ای رتبه و بعد از <code dir="ltr">hcp_dict</code> برای گرفتن امتیازش استفاده کنیم.

</div>

```python
rank = 12
rank_name = Card.rank_names[rank]
score = BridgeHand.hcp_dict.get(rank_name, 0)
rank_name, score
```

<div dir="rtl" style="text-align: right;">

متد زیر ورق‌های یک <code dir="ltr">BridgeHand</code> را پیمایش می‌کند و امتیازشان را جمع می‌زند.

</div>

```python
%%add_method_to BridgeHand

    def high_card_point_count(self):
        count = 0
        for card in self.cards:
            rank_name = Card.rank_names[card.rank]
            count += BridgeHand.hcp_dict.get(rank_name, 0)
        return count
```

```python
# This cell makes a fresh Deck and 
# initializes the random number generator

cards = Deck.make_cards()
deck = Deck(cards)
random.seed(3)
```

<div dir="rtl" style="text-align: right;">

برای آزمونش، دستی با پنج ورق پخش می‌کنیم – یک دست بریج معمولاً سیزده ورق دارد، اما آزمودن کد با مثال‌های کوچک آسان‌تر است.

</div>

```python
hand = BridgeHand('player 2')

deck.shuffle()
deck.move_cards(hand, 5)
print(hand)
```

<div dir="rtl" style="text-align: right;">

و این هم امتیاز کل شاه و بی‌بی.

</div>

```python
hand.high_card_point_count()
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">BridgeHand</code> متغیرها و متدهای <code dir="ltr">Hand</code> را به ارث می‌برد و یک متغیر کلاس و یک متد مخصوص بریج اضافه می‌کند. این شیوه‌ی استفاده از وراثت **تخصصی‌سازی** (specialization) نامیده می‌شود چون کلاس جدیدی تعریف می‌کند که برای کاربرد ویژه‌ای، مثل بازی بریج، تخصصی شده است.

## ۱۷.۱۰. اشکال‌زدایی

وراثت ویژگی مفیدی است. بعضی برنامه‌ها که بدون وراثت تکراری می‌شدند، با آن مختصرتر نوشته می‌شوند. همچنین وراثت می‌تواند استفاده‌ی مجدد از کد را آسان کند، چون می‌توانید رفتار کلاس والد را بدون تغییر آن سفارشی کنید. در بعضی موارد ساختار وراثت ساختار طبیعی مسئله را بازتاب می‌دهد، که طراحی را قابل‌فهم‌تر می‌کند.

از سوی دیگر، وراثت می‌تواند خواندن برنامه‌ها را دشوار کند. وقتی متدی احضار می‌شود، گاهی روشن نیست که تعریفش را کجا پیدا کنیم – ممکن است کد مربوط در چند ماژول پخش شده باشد.

هر وقت درباره‌ی جریان اجرای برنامه‌تان مطمئن نیستید، ساده‌ترین راه‌حل افزودن دستورهای print در ابتدای متدهای مربوط است. اگر <code dir="ltr">Deck.shuffle</code> پیامی چاپ کند، مثلاً <code dir="ltr">Running Deck.shuffle</code>، با اجرای برنامه جریان اجرا ردیابی می‌شود.

به‌عنوان یک جایگزین، می‌توانید از تابع زیر استفاده کنید که یک شیء و نام یک متد (به‌صورت رشته) را می‌گیرد و کلاسی را برمی‌گرداند که تعریف متد را فراهم می‌کند.

</div>

```python
def find_defining_class(obj, method_name):
    """Find the class where the given method is defined."""
    for typ in type(obj).mro():
        if method_name in vars(typ):
            return typ
    return f'Method {method_name} not found.'
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">find_defining_class</code> از متد <code dir="ltr">mro</code> استفاده می‌کند تا فهرست شیءهای کلاس (نوع‌ها) را که برای یافتن متدها جست‌وجو می‌شوند بگیرد. «MRO» مخفف «method resolution order» (ترتیب حل متد) است، یعنی دنباله‌ی کلاس‌هایی که پایتون برای «حل» نام یک متد جست‌وجو می‌کند – یعنی پیدا کردن شیء تابعی که نام به آن اشاره می‌کند.

به‌عنوان مثال، یک <code dir="ltr">BridgeHand</code> نمونه‌سازی می‌کنیم و بعد کلاس تعریف‌کننده‌ی <code dir="ltr">shuffle</code> را پیدا می‌کنیم.

</div>

```python
hand = BridgeHand('player 3')
find_defining_class(hand, 'shuffle')
```

<div dir="rtl" style="text-align: right;">

متد <code dir="ltr">shuffle</code> شیء <code dir="ltr">BridgeHand</code> همانی است که در <code dir="ltr">Deck</code> است.

## ۱۷.۱۱. واژه‌نامه

**وراثت (inheritance):** توانایی تعریف کلاس جدیدی که نسخه‌ی تغییریافته‌ی یک کلاس تعریف‌شده‌ی قبلی باشد.

**کدگذاری (encode):** نمایش یک مجموعه از مقدارها با مجموعه‌ی دیگری از مقدارها با ساختن نگاشتی بین آن‌ها.

**متغیر کلاس (class variable):** متغیری که داخل تعریف کلاس، اما نه داخل هیچ متدی، تعریف می‌شود.

**ترتیب کامل (totally ordered):** مجموعه‌ای از شیءها ترتیب کامل دارد اگر بتوانیم هر دو عنصر را مقایسه کنیم و نتیجه‌ها سازگار باشند.

**تفویض (delegation):** وقتی یک متد مسئولیت را به متد دیگری واگذار می‌کند تا بیشتر یا همه‌ی کار را انجام دهد.

**کلاس والد (parent class):** کلاسی که از آن ارث برده می‌شود.

**کلاس فرزند (child class):** کلاسی که از کلاس دیگری ارث می‌برد.

**تخصصی‌سازی (specialization):** شیوه‌ای از استفاده از وراثت برای ساختن کلاس جدیدی که نسخه‌ی تخصصی‌شده‌ی یک کلاس موجود است.

### تمرین ۲

چند تمرین بعدی از شما می‌خواهند توابعی بنویسید که دست‌های پوکر را دسته‌بندی کنند. اگر با پوکر آشنا نیستید، آنچه لازم است بدانید را توضیح می‌دهم. از کلاس زیر برای نمایش دست‌های پوکر استفاده می‌کنیم.

</div>

```python
class PokerHand(Hand):
    """Represents a poker hand."""

    def get_suit_counts(self):
        counter = {}
        for card in self.cards:
            key = card.suit
            counter[key] = counter.get(key, 0) + 1
        return counter
    
    def get_rank_counts(self):
        counter = {}
        for card in self.cards:
            key = card.rank
            counter[key] = counter.get(key, 0) + 1
        return counter    
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">PokerHand</code> دو متد دارد که در تمرین‌ها کمک می‌کنند.

* <code dir="ltr">get_suit_counts</code> ورق‌های <code dir="ltr">PokerHand</code> را پیمایش می‌کند، تعداد ورق‌های هر خال را می‌شمارد و دیکشنری‌ای برمی‌گرداند که از هر کد خال به تعداد دفعاتی که ظاهر می‌شود نگاشت می‌کند.

* <code dir="ltr">get_rank_counts</code> همین کار را با رتبه‌های ورق‌ها انجام می‌دهد و دیکشنری‌ای برمی‌گرداند که از هر کد رتبه به تعداد دفعاتی که ظاهر می‌شود نگاشت می‌کند.

همه‌ی تمرین‌های بعدی را می‌توان فقط با امکانات پایتونی که تا حالا یاد گرفته‌ایم انجام داد، اما بعضی از آن‌ها دشوارتر از بیشتر تمرین‌های قبلی‌اند. تشویقتان می‌کنم از یک دستیار مجازی کمک بگیرید.

```python
%%add_method_to PokerHand

    def has_straight(self, n=5):
        """Checks whether this hand has a straight with at least `n` cards."""
        counter = self.get_rank_counts()
        aces = counter.get(1, 0) + counter.get(14, 0)
        counter[1] = aces
        counter[14] = aces
        
        return False
```

### تمرین ۷

این تمرین حکایتی هشداردهنده درباره‌ی خطای رایجی است که اشکال‌زدایی‌اش می‌تواند دشوار باشد. تعریف کلاس زیر را در نظر بگیرید.

</div>

```python
class Kangaroo:
    """A Kangaroo is a marsupial."""
    
    def __init__(self, name, contents=[]):
        """Initialize the pouch contents.

        name: string
        contents: initial pouch contents.
        """
        self.name = name
        self.contents = contents

    def __str__(self):
        """Return a string representaion of this Kangaroo.
        """
        t = [ self.name + ' has pouch contents:' ]
        for obj in self.contents:
            s = '    ' + object.__str__(obj)
            t.append(s)
        return '\n'.join(t)

    def put_in_pouch(self, item):
        """Adds a new item to the pouch contents.

        item: object to be added
        """
        self.contents.append(item)
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">__init__</code> دو پارامتر می‌گیرد: <code dir="ltr">name</code> الزامی است، اما <code dir="ltr">contents</code> اختیاری است – اگر داده نشود، مقدار پیش‌فرض یک فهرست خالی است.

<code dir="ltr">__str__</code> یک نمایش رشته‌ای از شیء برمی‌گرداند که نام و محتوای کیسه را دارد.

<code dir="ltr">put_in_pouch</code> هر شیئی را می‌گیرد و به <code dir="ltr">contents</code> اضافه می‌کند.

حالا ببینیم این کلاس چطور کار می‌کند. دو شیء <code dir="ltr">Kangaroo</code> با نام‌های <code dir="ltr">'Kanga'</code> و <code dir="ltr">'Roo'</code> می‌سازیم.
