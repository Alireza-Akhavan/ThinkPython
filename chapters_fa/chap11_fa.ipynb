## ۱۱.۳. انتساب چندتایی

می‌توانید در سمت چپ یک نسبت‌دهی چندتایی از متغیرها و در سمت راست چندتایی از مقدارها قرار دهید.

</div>

```python
a, b = 1, 2
```

<div dir="rtl" style="text-align: right;">

مقدارها از چپ به راست به متغیرها نسبت داده می‌شوند – در این مثال <code dir="ltr">a</code> مقدار <code dir="ltr">1</code> و <code dir="ltr">b</code> مقدار <code dir="ltr">2</code> را می‌گیرد. می‌توانیم نتیجه را این‌طور نمایش دهیم:

</div>

```python
a, b
```

<div dir="rtl" style="text-align: right;">

به‌طور کلی‌تر، اگر سمت چپ نسبت‌دهی چندتایی باشد، سمت راست می‌تواند هر نوع دنباله‌ای – رشته، فهرست یا چندتایی – باشد. برای مثال، برای شکستن یک نشانی ایمیل به نام کاربری و دامنه، می‌توانید بنویسید:

</div>

```python
email = 'student@example.org'
username, domain = email.split('@')
```

<div dir="rtl" style="text-align: right;">

مقدار بازگشتی <code dir="ltr">split</code> فهرستی با دو عنصر است – عنصر اول به <code dir="ltr">username</code> و دومی به <code dir="ltr">domain</code> نسبت داده می‌شود.

</div>

```python
username, domain
```

<div dir="rtl" style="text-align: right;">

تعداد متغیرهای سمت چپ و تعداد مقدارهای سمت راست باید برابر باشد – در غیر این صورت <code dir="ltr">ValueError</code> می‌گیرید.

</div>

```python
%%expect ValueError
a, b = 1, 2, 3
```

<div dir="rtl" style="text-align: right;">

انتساب چندتایی وقتی می‌خواهید مقدار دو متغیر را جابه‌جا کنید مفید است. با نسبت‌دهی‌های معمول باید از یک متغیر موقت استفاده کنید، مثل این:

</div>

```python
temp = a
a = b
b = temp
```

<div dir="rtl" style="text-align: right;">

این کار می‌کند، اما با انتساب چندتایی می‌توانیم همین کار را بدون متغیر موقت انجام دهیم.

</div>

```python
a, b = b, a
```

<div dir="rtl" style="text-align: right;">

این کار می‌کند چون همه‌ی عبارت‌های سمت راست قبل از هر نسبت‌دهی ارزیابی می‌شوند.

می‌توانیم از انتساب چندتایی در دستور <code dir="ltr">for</code> هم استفاده کنیم. برای مثال، برای پیمایش موردهای یک دیکشنری، می‌توانیم از متد <code dir="ltr">items</code> استفاده کنیم.

</div>

```python
d = {'one': 1, 'two': 2}

for item in d.items():
    key, value = item
    print(key, '->', value)
```

<div dir="rtl" style="text-align: right;">

هر بار که حلقه تکرار می‌شود، به <code dir="ltr">item</code> یک چندتایی شامل یک کلید و مقدار متناظرش نسبت داده می‌شود.

می‌توانیم این حلقه را مختصرتر این‌طور بنویسیم:

</div>

```python
for key, value in d.items():
    print(key, '->', value)
```

<div dir="rtl" style="text-align: right;">

هر بار که حلقه تکرار می‌شود، یک کلید و مقدار متناظرش مستقیماً به <code dir="ltr">key</code> و <code dir="ltr">value</code> نسبت داده می‌شوند.

</div>

## ۱۱.۴. چندتایی‌ها به‌عنوان مقدار بازگشتی

به‌طور دقیق، یک تابع فقط می‌تواند یک مقدار برگرداند، اما اگر مقدار یک چندتایی باشد، اثرش مثل برگرداندن چند مقدار است. برای مثال، اگر بخواهید دو عدد صحیح را تقسیم کنید و خارج‌قسمت و باقی‌مانده را حساب کنید، ناکارآمد است که اول <code dir="ltr">x//y</code> و بعد <code dir="ltr">x%y</code> را حساب کنید. بهتر است هر دو را همزمان حساب کنید.

تابع داخلی <code dir="ltr">divmod</code> دو آرگومان می‌گیرد و چندتایی‌ای از دو مقدار، یعنی خارج‌قسمت و باقی‌مانده، برمی‌گرداند.

</div>

```python
divmod(7, 3)
```

<div dir="rtl" style="text-align: right;">

می‌توانیم از انتساب چندتایی برای ذخیره‌ی عنصرهای چندتایی در دو متغیر استفاده کنیم.

</div>

```python
quotient, remainder = divmod(7, 3)
quotient
```

```python
remainder
```

<div dir="rtl" style="text-align: right;">

این مثالی از تابعی است که چندتایی برمی‌گرداند.

</div>

```python
def min_max(t):
    return min(t), max(t)
```

<div dir="rtl" style="text-align: right;">

<code dir="ltr">max</code> و <code dir="ltr">min</code> توابع داخلی‌ای هستند که بزرگ‌ترین و کوچک‌ترین عنصر یک دنباله را پیدا می‌کنند. <code dir="ltr">min_max</code> هر دو را حساب می‌کند و چندتایی‌ای از دو مقدار برمی‌گرداند.

</div>

```python
min_max([2, 4, 1, 3])
```

<div dir="rtl" style="text-align: right;">

می‌توانیم نتیجه‌ها را این‌طور به متغیرها نسبت دهیم:

</div>

```python
low, high = min_max([2, 4, 1, 3])
low, high
```

## ۱۱.۵. بسته‌بندی آرگومان‌ها (Argument packing)

توابع می‌توانند تعداد متغیری آرگومان بگیرند. نام پارامتری که با عملگر <code dir="ltr">*</code> شروع شود، آرگومان‌ها را در یک چندتایی **بسته‌بندی** (pack) می‌کند. برای مثال، تابع زیر هر تعداد آرگومان می‌گیرد و میانگین حسابی آن‌ها را حساب می‌کند – یعنی مجموعشان تقسیم بر تعداد آرگومان‌ها.

</div>

```python
def mean(*args):
    return sum(args) / len(args)
```

<div dir="rtl" style="text-align: right;">

پارامتر می‌تواند هر نامی که دوست دارید داشته باشد، اما <code dir="ltr">args</code> قراردادی است. می‌توانیم تابع را این‌طور فراخوانی کنیم:

</div>

```python
mean(1, 2, 3)
```

<div dir="rtl" style="text-align: right;">

اگر دنباله‌ای از مقدارها دارید و می‌خواهید آن‌ها را به‌صورت چند آرگومان به یک تابع بدهید، می‌توانید از عملگر <code dir="ltr">*</code> برای **باز کردن بسته** (unpack) چندتایی استفاده کنید. برای مثال، <code dir="ltr">divmod</code> دقیقاً دو آرگومان می‌گیرد – اگر یک چندتایی را به‌عنوان پارامتر بدهید، خطا می‌گیرید.

</div>

```python
%%expect TypeError
t = (7, 3)
divmod(t)
```

<div dir="rtl" style="text-align: right;">

با اینکه چندتایی دو عنصر دارد، یک آرگومان به حساب می‌آید. اما اگر چندتایی را باز کنید، به‌عنوان دو آرگومان در نظر گرفته می‌شود.

</div>

```python
divmod(*t)
```

<div dir="rtl" style="text-align: right;">

بسته‌بندی و بازکردن بسته وقتی مفید است که بخواهید رفتار یک تابع موجود را تطبیق دهید. برای مثال، این تابع هر تعداد آرگومان می‌گیرد، کوچک‌ترین و بزرگ‌ترین را حذف می‌کند و میانگین بقیه را حساب می‌کند.

</div>

```python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
```

<div dir="rtl" style="text-align: right;">

اول با <code dir="ltr">min_max</code> کوچک‌ترین و بزرگ‌ترین عنصر را پیدا می‌کند. بعد <code dir="ltr">args</code> را به فهرست تبدیل می‌کند تا بتواند از متد <code dir="ltr">remove</code> استفاده کند. در پایان فهرست را باز می‌کند تا عنصرها به‌جای یک فهرست واحد، به‌صورت آرگومان‌های جداگانه به <code dir="ltr">mean</code> داده شوند.

این مثالی است که اثرش را نشان می‌دهد.

</div>

```python
mean(1, 2, 3, 10)
```

```python
trimmed_mean(1, 2, 3, 10)
```

<div dir="rtl" style="text-align: right;">

این نوع میانگین «پیراسته» (trimmed) در بعضی ورزش‌ها با داوری ذهنی – مثل شیرجه و ژیمناستیک – استفاده می‌شود تا اثر داوری که نمره‌اش با بقیه فرق دارد کم شود.

</div>

## ۱۱.۶. Zip

چندتایی‌ها برای پیمایش عنصرهای دو دنباله و انجام عملیات روی عنصرهای متناظر مفیدند. برای مثال، فرض کنید دو تیم یک سری هفت‌بازی انجام می‌دهند و امتیازهایشان را در دو فهرست، یکی برای هر تیم، ثبت می‌کنیم.

</div>

```python
scores1 = [1, 2, 4, 5, 1, 5, 2]
scores2 = [5, 5, 2, 2, 5, 2, 3]
```

## ۱۱.۷. مقایسه و مرتب‌سازی

عملگرهای رابطه‌ای با چندتایی‌ها و دنباله‌های دیگر کار می‌کنند. برای مثال، اگر از عملگر <code dir="ltr">&lt;</code> با چندتایی‌ها استفاده کنید، با مقایسه‌ی عنصر اول از هر دنباله شروع می‌کند. اگر برابر باشند، به جفت عنصر بعدی می‌رود و همین‌طور ادامه می‌دهد تا جفتی را پیدا کند که فرق دارد.

</div>

```python
(0, 1, 2) < (0, 3, 4)
```

<div dir="rtl" style="text-align: right;">

عنصرهای بعدی در نظر گرفته نمی‌شوند – حتی اگر خیلی بزرگ باشند.

</div>

```python
(0, 1, 2000000) < (0, 3, 4)
```

<div dir="rtl" style="text-align: right;">

این روش مقایسه‌ی چندتایی‌ها برای مرتب کردن فهرستی از چندتایی‌ها، یا پیدا کردن کمینه یا بیشینه مفید است. به‌عنوان مثال، پرتکرارترین حرف یک کلمه را پیدا می‌کنیم. در فصل قبل <code dir="ltr">value_counts</code> را نوشتیم که رشته‌ای می‌گیرد و دیکشنری‌ای برمی‌گرداند که هر حرف را به تعداد دفعاتی که ظاهر می‌شود نگاشت می‌کند.

</div>

```python
def value_counts(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

<div dir="rtl" style="text-align: right;">

این نتیجه‌ی رشته‌ی <code dir="ltr">'banana'</code> است.

</div>

```python
counter = value_counts('banana')
counter
```

<div dir="rtl" style="text-align: right;">

با فقط سه مورد، به‌راحتی می‌بینیم که پرتکرارترین حرف <code dir="ltr">'a'</code> است که سه بار ظاهر می‌شود. اما اگر موردهای بیشتری بود، مفید می‌بود که آن‌ها را خودکار مرتب کنیم.

می‌توانیم موردها را این‌طور از <code dir="ltr">counter</code> بگیریم.

```python
sorted(items)
```

<div dir="rtl" style="text-align: right;">

رفتار پیش‌فرض این است که از عنصر اول هر چندتایی برای مرتب کردن فهرست استفاده می‌کند و از عنصر دوم برای شکستن تساوی‌ها.

اما برای پیدا کردن موردهایی با بیشترین شمارش، می‌خواهیم از عنصر دوم برای مرتب کردن فهرست استفاده کنیم. می‌توانیم این کار را با نوشتن تابعی انجام دهیم که چندتایی می‌گیرد و عنصر دوم را برمی‌گرداند.

```python
def second_element(t):
    return t[1]
```

<div dir="rtl" style="text-align: right;">

بعد می‌توانیم آن تابع را به‌عنوان آرگومان اختیاری به نام <code dir="ltr">key</code> به <code dir="ltr">sorted</code> بدهیم، که نشان می‌دهد این تابع باید برای محاسبه‌ی **کلید مرتب‌سازی** (sort key) هر مورد به کار رود.

</div>

```python
sorted_items = sorted(items, key=second_element)
sorted_items
```

کلید مرتب‌سازی ترتیب موردهای فهرست را تعیین می‌کند. حرفی که کمترین شمارش را دارد اول و حرفی که بیشترین شمارش را دارد آخر می‌آید. پس پرتکرارترین حرف را این‌طور پیدا می‌کنیم.

## ۱۱.۸. وارونه کردن یک دیکشنری

فرض کنید می‌خواهید یک دیکشنری را وارونه کنید تا بتوانید یک مقدار را جست‌وجو کنید و کلید متناظر را بگیرید. برای مثال، اگر شمارنده‌ی کلمه‌ای دارید که هر کلمه را به تعداد دفعاتی که ظاهر می‌شود نگاشت می‌کند، می‌توانید دیکشنری‌ای بسازید که عددهای صحیح را به کلمه‌هایی که آن تعداد بار ظاهر می‌شوند نگاشت کند.

اما مشکلی هست – کلیدهای دیکشنری باید یکتا باشند، اما مقدارها لازم نیست. برای مثال، در یک شمارنده‌ی کلمه ممکن است کلمه‌های زیادی شمارش یکسانی داشته باشند.

پس یک راه برای وارونه کردن دیکشنری این است که دیکشنری جدیدی بسازیم که مقدارهایش فهرستی از کلیدهای دیکشنری اصلی باشد. به‌عنوان مثال، حرف‌های <code dir="ltr">parrot</code> را می‌شماریم.

</div>

```python
d =  value_counts('parrot')
d
```

<div dir="rtl" style="text-align: right;">

اگر این دیکشنری را وارونه کنیم، نتیجه باید <code dir="ltr">{1: ['p', 'a', 'o', 't'], 2: ['r']}</code> باشد، که نشان می‌دهد حرف‌هایی که یک بار ظاهر می‌شوند <code dir="ltr">'p'</code>، <code dir="ltr">'a'</code>، <code dir="ltr">'o'</code> و <code dir="ltr">'t'</code> هستند، و حرفی که دو بار ظاهر می‌شود <code dir="ltr">'r'</code> است.

تابع زیر یک دیکشنری می‌گیرد و وارونه‌اش را به‌صورت دیکشنری جدید برمی‌گرداند.

</div>

```python
def invert_dict(d):
    new = {}
    for key, value in d.items():
        if value not in new:
            new[value] = [key]
        else:
            new[value].append(key)
    return new
```

<div dir="rtl" style="text-align: right;">

دستور <code dir="ltr">for</code> کلیدها و مقدارهای <code dir="ltr">d</code> را پیمایش می‌کند. اگر مقدار از قبل در دیکشنری جدید نباشد، اضافه می‌شود و با فهرستی با یک عنصر مرتبط می‌شود. در غیر این صورت به فهرست موجود اضافه می‌شود.

می‌توانیم آن را این‌طور آزمایش کنیم:

</div>

```python
invert_dict(d)
```

<div dir="rtl" style="text-align: right;">

و نتیجه‌ای را که انتظار داشتیم می‌گیریم.

این اولین مثالی است که می‌بینیم مقدارهای دیکشنری فهرست‌اند. مثال‌های بیشتری هم خواهیم دید!

## ۱۱.۱۰. واژه‌نامه

**بسته‌بندی (pack):** جمع کردن چند آرگومان در یک چندتایی.

**بازکردن بسته (unpack):** در نظر گرفتن یک چندتایی (یا دنباله‌ی دیگر) به‌عنوان چند آرگومان.

**شیء zip (zip object):** نتیجه‌ی فراخوانی تابع داخلی <code dir="ltr">zip</code>، که می‌توان از آن برای پیمایش دنباله‌ای از چندتایی‌ها استفاده کرد.

**شیء enumerate (enumerate object):** نتیجه‌ی فراخوانی تابع داخلی <code dir="ltr">enumerate</code>، که می‌توان از آن برای پیمایش دنباله‌ای از چندتایی‌ها استفاده کرد.

**کلید مرتب‌سازی (sort key):** مقدار، یا تابعی که مقداری را حساب می‌کند، که برای مرتب کردن عنصرهای یک مجموعه به کار می‌رود.

**ساختار داده (data structure):** مجموعه‌ای از مقدارها که طوری سازمان‌دهی شده‌اند که عملیات خاصی را کارآمد انجام دهند.
